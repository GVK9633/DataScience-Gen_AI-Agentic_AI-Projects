# 10. Metrics

This is where your paper becomes much stronger.

Don't report only accuracy.

## A. Answer metrics

### Exact Match

Does the predicted answer exactly match the expected answer?

### F1

Measures overlap between predicted and reference answer.

Multi-hop RAG research commonly reports EM and F1. ([DOI][2])

---

## B. Retrieval metrics

Measure:

### Context Recall

Did we retrieve the information required to answer?

### Context Precision

How much of the retrieved information was actually relevant?

### Hit@K

Was the required evidence retrieved within the top K?

These retrieval metrics are important because a bad answer may be caused by **retrieval failure**, not generation failure. RAG evaluation frameworks commonly separate retrieval quality from answer quality using measures such as context precision and recall. ([EvalHub][3])

---

# 11. Faithfulness

Measure whether the answer is actually supported by retrieved evidence.

For example:

```text
Retrieved evidence:
"Company X was founded by John."

Generated answer:
"Company X was founded by John in 1998."

```

If `1998` wasn't in the evidence, the answer contains an unsupported claim.

So:

> **Faithfulness = Is the answer grounded in the retrieved evidence?**

This is particularly important for your future Agentic RAG research. Faithfulness is a core RAG evaluation dimension. ([Springer][4])
