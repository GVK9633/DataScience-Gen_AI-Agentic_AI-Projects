#### Bedrock Based RAG 

```python
from langchain_aws.retrievers import AmazonKnowledgeBasesRetriever
from langchain_aws import ChatBedrockConverse
```

```python

llm = ChatBedrockConverse(
    model=MODEL,
    region_name=REGION
)
retriever = AmazonKnowledgeBasesRetriever(
    knowledge_base_id="TGF3P",
    retrieval_config={"vectorSearchConfiguration": {"numberOfResults": 5}},
)

def get_context(query):
    chunks = retriever.invoke(query)

    context = ""
    for chunk in chunks:
        context += chunk.page_content + "\n"
        
    return context
```

```python
def ask(query):
    context = get_context(query)
    response = llm.invoke(f"""
                          You always provide answer for user question based on the provided context.capitalize
                          if you not found any answer for user question. just say I don't know.
                            Rule: You never say 'Based on the provided context,' Just answer for user question, no extra details.
                          Context: {context}
                          
                          Question: {query}
                          """)
    
    return response.content
    
    
    
```

```python
answer = ask("Wat is the age of patient ?")
```
