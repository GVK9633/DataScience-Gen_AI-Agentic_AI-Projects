# ✅ 1. **JPMorgan Chase**

## Q1. How to decide **single-agent vs multi-agent**?

**Interview answer**

* Use a **single agent** when:

  * the task is linear
  * one goal
  * limited tools
  * no hand-off between responsibilities
    👉 Example: simple FAQ RAG bot

* Use a **multi-agent system** when:

  * different roles are needed (classifier, retriever, validator, executor)
  * reasoning and execution are separated
  * human-in-loop or approval is needed
    👉 Example: policy check → fraud check → document extraction → decision

**One-liner**

> Single agent for simple workflows; multi-agent when tasks need specialization, orchestration and hand-offs.

---

## Q2. How many tools can be used in one MCP server?

**Interview answer**

> There is no hard limit.
> An MCP server can expose **multiple tools**, but in production we keep:

* tools small
* purpose-specific
* and logically grouped

Practical guidance:

* 5–20 tools per server is common
* avoid building a “god-server” with too many unrelated tools

3. Validation agent → format & business rules
4. Fraud agent → duplicate / blacklist / history check

## Memory management in Python

Python uses:

* automatic memory allocation
* reference counting
* garbage collection for cyclic objects

Key idea:

> Memory is released when object reference count becomes zero.

---

## Shallow copy vs Deep copy

**Shallow copy**

* copies only the outer object
* inner objects are shared

**Deep copy**

* copies everything recursively

**One-liner**

> Shallow copy duplicates references, deep copy duplicates objects.
