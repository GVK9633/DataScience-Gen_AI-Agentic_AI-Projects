Interview-Ready Answer (Use This)
“Temperature, Top-K, and Top-P control the randomness of LLM outputs. Temperature scales the probability distribution, Top-K limits choices to the K most likely tokens, and Top-P dynamically selects tokens whose cumulative probability exceeds a threshold. In production systems like RAG, we use low temperature with moderate Top-P for factual consistency.”

Temperature = randomness
Top-K = fixed choices
Top-P = smart choices
