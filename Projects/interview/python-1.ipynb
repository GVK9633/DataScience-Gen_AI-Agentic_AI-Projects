## 2️⃣ Mutable vs Immutable objects

**Answer:**
Mutable → can change after creation
Immutable → cannot change

**Examples:**

```python
list, dict, set   # mutable
int, float, str, tuple  # immutable
```

---

## 3️⃣ What is Python memory management?

**Answer:**
Python uses:

* Private heap space
* Automatic garbage collection
* Reference counting + cyclic GC

---

## 4️⃣ What is the Global Interpreter Lock (GIL)?

**Answer:**

> GIL allows only one thread to execute Python bytecode at a time, preventing race conditions but limiting CPU-bound multithreading.

---

## 5️⃣ Threading vs Multiprocessing

| Threading   | Multiprocessing |
| ----------- | --------------- |
| Same memory | Separate memory |
| GIL limited | No GIL          |
| IO-bound    | CPU-bound       |

---

## 6️⃣ What are decorators?

**Answer:**
Decorators modify function behavior **without changing function code**.

```python
def my_decorator(func):
    def wrapper():
        print("Before")
        func()
    return wrapper
```

---

## 7️⃣ What are generators?

**Answer:**
Generators return values **one at a time** using `yield`.

```python
def gen():
    yield 1
    yield 2
```

## 9️⃣ Deep copy vs Shallow copy

**Answer:**

* Shallow → references same objects
* Deep → copies everything

```python
import copy
copy.copy()
copy.deepcopy()
```
