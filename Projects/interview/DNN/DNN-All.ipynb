## Q1. Traditional ML vs Deep Learning

### Traditional Machine Learning

Usually requires **manual feature engineering**.

Example: Image classification

```text
Image
 ↓
Manually extract features
 ├── Edges
 ├── Color
 ├── Shape
 ↓
ML Model
 ↓
Prediction
```

Algorithms:

* Logistic Regression
* Decision Tree
* Random Forest
* SVM

### Deep Learning

The neural network learns useful features automatically.

```text
Image
 ↓
CNN
 ├── Edges
 ├── Shapes
 ├── Objects
 ↓
Prediction
```

### Comparison

| Traditional ML                                 | Deep Learning                          |
| ---------------------------------------------- | -------------------------------------- |
| Feature engineering often required             | Learns representations automatically   |

| Usually less computationally expensive         | Usually more computationally expensive |
| Random Forest, SVM, Logistic Regression        | CNN, RNN, Transformer                  |

### Interview answer

> "Traditional ML often relies on manually engineered features, whereas deep learning learns hierarchical representations directly from raw data using neural networks."

# Q2. Explain ANN Structure and Forward Propagation

ANN = **Artificial Neural Network**.

Basic structure:

```text
Input Layer
     ↓
Hidden Layer
     ↓
Hidden Layer
     ↓
Output Layer
```

Example:

```text
x1 ──┐
x2 ──┼──> Hidden Neuron ──> Output
x3 ──┘
```

A neuron calculates:

$$
z = w_1x_1+w_2x_2+\cdots+w_nx_n+b
$$

Then:

$$
a=f(z)
$$

where `f` is the activation function.

### Example

Suppose:

```text
x1 = 2
x2 = 3

w1 = 0.5
w2 = 0.2
b = 1
```

Then:

```text
z = (0.5 × 2) + (0.2 × 3) + 1
  = 2.6
```

Activation function produces the neuron's output.

### Forward propagation

```text
Input
 ↓
Weighted Sum
 ↓
Activation
 ↓
Next Layer
 ↓
Output
```

### Interview answer

> "An ANN consists of input, hidden and output layers. During forward propagation, each neuron calculates a weighted sum plus bias, applies an activation function, and passes the result to the next layer."

# Q3. Why is Zero Weight Initialization a Problem?

Suppose every neuron starts with:

```text
w1 = 0
w2 = 0
w3 = 0
```

Consider two neurons in the same layer.

They receive the same inputs and have identical weights.

Therefore:

```text
Neuron 1 → same output
Neuron 2 → same output
Neuron 3 → same output
```

During backpropagation they receive the same gradients.

So they remain identical.

This is called the **symmetry problem**.

### Solution

Use random weight initialization.

Common approaches:

* Xavier/Glorot initialization
* He initialization

### Interview answer

> "Initializing all weights to zero causes neurons in the same layer to learn identical features because they receive identical gradients. Random initialization breaks this symmetry."

### Important

Biases can generally be initialized to zero; the major issue is initializing **all weights** identically.

# Q4. Activation Functions and the Dead ReLU Problem

ReLU:

$$
ReLU(x)=\max(0,x)
$$

Example:

```text id="1a6lks"
x = 5
ReLU(5) = 5

x = -5
ReLU(-5) = 0
```

### Why ReLU is popular?

* Simple
* Computationally efficient
* Helps reduce vanishing-gradient issues compared with sigmoid/tanh in many deep networks

### Dead ReLU

For negative input:

```text
ReLU(x) = 0
```

and the derivative is:

```text
0
```

If a neuron consistently receives negative inputs, its gradient can remain zero and the neuron may stop learning.

This is called:

> **Dying/Dead ReLU**

### Solutions

* Leaky ReLU
* PReLU
* Better initialization
* Appropriate learning rate

### Interview answer

> "ReLU outputs zero for negative inputs, and a neuron can become permanently inactive if it stays in that region, causing the dead ReLU problem. Leaky ReLU and proper initialization can help."

# Q5. How do you detect and prevent Overfitting in Deep Learning?

### Overfitting

Model performs very well on training data but poorly on unseen data.

Example:

```text
Training accuracy = 99%
Validation accuracy = 75%
```

This is a warning sign of overfitting.

### How to detect?

Compare:

```text
Training loss
Validation loss
```

Example:

```text
Training loss
   ↓↓↓↓↓

Validation loss
   ↓↓
      ↑
       ↑
```

If training loss continues decreasing while validation loss starts increasing, overfitting may be occurring.

### Solutions

1. Dropout
2. L1/L2 regularization
3. Early stopping
4. Data augmentation
5. More training data
6. Reduce model complexity

### Interview answer

> "I detect overfitting by comparing training and validation performance. If training performance improves while validation performance deteriorates, I can use techniques such as dropout, regularization, early stopping, data augmentation, or more data."

# Q6. Vanishing vs Exploding Gradients

These problems occur during **backpropagation**, especially in deep or recurrent networks.

## Vanishing Gradient

Gradients become extremely small.

```text
Gradient
 ↓
0.1
 ↓
0.01
 ↓
0.001
 ↓
0.0001
```

Earlier layers learn very slowly.

### Exploding Gradient

Gradients become extremely large.

```text
1
 ↓
10
 ↓
100
 ↓
1000
```

This can cause unstable training.

### Causes

Repeated multiplication through many layers/time steps can cause gradients to shrink or grow dramatically.

### Solutions

Vanishing:

* ReLU-family activations
* Proper initialization
* Batch/layer normalization
* LSTM/GRU for sequence problems

Exploding:

* Gradient clipping
* Appropriate initialization
* Learning-rate tuning
* Normalization

### Interview answer

> "Vanishing gradients become too small for earlier layers to learn effectively, while exploding gradients become excessively large and make training unstable."

# Q7. What is a Receptive Field in CNN?

The **receptive field** is the region of the original input that influences a particular feature-map value.

Imagine:

```text
Original Image
┌───────────────┐
│               │
│    Object     │
│               │
└───────────────┘
```

A CNN's early neuron may see a small region:

```text
3 × 3
```

Deeper layers combine information from larger regions.

```text
Early layer  → small receptive field
Deep layer   → larger receptive field
```

### Why important?

Early layers learn:

```text
Edges
```

Deeper layers can learn:

```text
Shapes → Objects
```

### Interview answer

> "The receptive field is the portion of the input image that can influence a particular neuron or feature-map location, and it generally grows as we move deeper into a CNN."

# Q8. What is the purpose of a 1×1 Convolution?

This is a common CNN interview question.

A `1 × 1` convolution operates **across channels at each spatial location**.

Suppose:

```text
Feature map:
56 × 56 × 128
```

Use:

```text
1 × 1 convolution
```

with 32 filters.

Output:

```text
56 × 56 × 32
```

So it can reduce the number of channels.

### Why use it?

1. Channel reduction
2. Computational efficiency
3. Add non-linearity when followed by activation
4. Channel mixing

### Interview answer

> "A 1×1 convolution performs channel-wise feature mixing at each spatial location and is commonly used to reduce or change the number of channels while keeping spatial dimensions unchanged when stride is 1 and padding preserves size."

# Q9. Sigmoid vs Tanh in LSTMs

LSTMs use both.

### Sigmoid

Output:

```text
0 → 1
```

Used as gates:

```text
Forget gate
Input gate
Output gate
```

It controls **how much information passes through**.

Example:

```text
sigmoid = 0.8
```

means roughly:

> allow a high proportion of the information through.

---

### Tanh

Output:

```text
-1 → +1
```

Used to create/candidate and output bounded state information.

### LSTM concept

```text
Input
  ↓
Sigmoid gates → control information
  ↓
Tanh        → candidate/state transformation
```

### Interview answer

> "LSTMs use sigmoid gates to control how much information is retained or passed, while tanh produces bounded candidate/state values between -1 and 1."

# Q10. What is Teacher Forcing and Exposure Bias?

This is important for sequence models.

Suppose we are training a language model to generate:

```text
I love machine learning
```

During training, after predicting:

```text
I
```

we give the model the **correct previous token**:

```text
I → love
```

instead of its own prediction.

This is called:

> **Teacher Forcing**

### Problem: Exposure Bias

During training:

```text
Correct previous output
```

During inference:

```text
Model's own previous output
```

So there is a mismatch.

```text
Training:
Correct → Correct → Correct

Inference:
Prediction → Prediction → Prediction
```

One wrong prediction can affect subsequent predictions.

### Interview answer

> "Teacher forcing uses the ground-truth previous token during training, which speeds learning, but creates exposure bias because during inference the model must use its own previous predictions."

# Q11. LSTM Limitations and the Attention Fix

LSTMs process sequences sequentially.

For a long sentence:

```text
Token 1
 ↓
Token 2
 ↓
Token 3
 ↓
...
 ↓
Token 100
```

The model must carry information through many steps.

This can make long-range dependencies difficult.

### Attention

Attention allows the model to directly focus on relevant tokens.

Example:

```text
"The animal didn't cross the road because it was tired."
```

To understand `it`, the model may need to focus on `animal`.

Attention provides a direct mechanism to assign higher weight to relevant tokens.

### Interview answer

> "LSTMs process sequences sequentially and can struggle with long-range dependencies. Attention allows a model to directly focus on relevant parts of the sequence, improving long-context dependency handling."

# Q12. Why do we scale Dot-Product Attention?

Attention score:

$$
QK^T
$$

Scaled attention:

$$
\frac{QK^T}{\sqrt{d_k}}
$$

Why?

If `d_k` becomes large, the dot products can become large.

Then softmax can become very sharp:

```text
[10, 1, 0]
```

Softmax becomes heavily concentrated on one value.

This can result in **very small gradients** and make training difficult.

Dividing by:

$$
\sqrt{d_k}
$$

keeps the magnitude more controlled.

### Interview answer

> "We divide the query-key dot product by √dₖ to control the variance of attention scores, preventing softmax from becoming excessively peaked and helping maintain stable gradients."

# Q13. Multi-Head vs Single-Head Attention

### Single Head

One attention mechanism:

```text
Q K V
 ↓
Attention
```

### Multi-Head

Multiple attention mechanisms operate in parallel:

```text
        Q K V
          ↓
 ┌────────┼────────┐
Head 1   Head 2   Head 3
  ↓        ↓        ↓
Attention Attention Attention
 └────────┼────────┘
          ↓
      Concatenate
          ↓
       Linear
```

Different heads can learn different relationships.

For example:

```text
Head 1 → syntactic relationship
Head 2 → semantic relationship
Head 3 → positional/other relationship
```

The exact learned behavior isn't guaranteed to correspond cleanly to these labels, but multiple heads allow different representation subspaces.

### Interview answer

> "Multi-head attention runs several attention mechanisms in parallel, allowing the model to capture different relationships and representations, whereas single-head attention uses one attention mechanism."

# Q14. Why is Positional Encoding needed in Transformers?

This is **very important**.

Transformers use self-attention, which by itself does not inherently encode token order.

Consider:

```text
Dog bites man
```

versus:

```text
Man bites dog
```

The words are the same, but the order changes the meaning.

Therefore, the model needs positional information.

```text
Token + Position
     ↓
Transformer
```

Original Transformer architecture used sinusoidal positional encodings.

Modern Transformer architectures may use other approaches, such as learned positional embeddings or rotary positional embeddings (RoPE).

### Interview answer

> "Positional information is needed because self-attention alone does not inherently represent token order. Positional encodings or related positional mechanisms provide the model with sequence-order information."

# Q15. Pre-Layer Normalization vs Post-Layer Normalization

This is a more advanced Transformer interview question.

Consider a Transformer block.

## Post-LN

Conceptually:

```text
Input
 ↓
Attention
 ↓
Add residual
 ↓
LayerNorm
```

## Pre-LN

```text
Input
 ↓
LayerNorm
 ↓
Attention
 ↓
Add residual
```

### Why is Pre-LN popular?

Pre-LN generally provides **more stable gradient flow during optimization**, especially as Transformer networks become deeper.

### Simplified comparison

| Pre-LN                                        | Post-LN                          |
| --------------------------------------------- | -------------------------------- |
| Norm before sublayer                          | Norm after residual              |
| Often easier to optimize in deep Transformers | Original Transformer formulation |
| Common in modern architectures                | Still used in some architectures |

### Interview answer

> "In Pre-LN Transformers, normalization is applied before the attention/FFN sublayer, while in Post-LN it is applied after the residual connection. Pre-LN is often preferred for more stable optimization of deep Transformers."
