# 📊 Precision@K & Recall@K (RAG Retrieval Metrics)

## 1️⃣ What Problem They Solve

RAGAS evaluates **end-to-end quality**, but **Precision@K / Recall@K** focus purely on **retrieval quality**.

> “Did my retriever fetch the right documents?”

---

## 2️⃣ Definitions (Say This in Interview)

### 🔹 Precision@K

> “Out of the **top K retrieved documents**, how many are actually relevant?”

### 🔹 Recall@K

> “Out of **all relevant documents**, how many were retrieved in the top K?”

---

## 3️⃣ Mathematical Formulas (Interview-safe)

```
Precision@K = Relevant Retrieved Docs / K

Recall@K = Relevant Retrieved Docs / Total Relevant Docs
```

---

## 4️⃣ Simple Example (Return Policy)

### Query:

> “Can I return electronics after 40 days?”

### Relevant Docs (Ground Truth):

```
D1: Electronics can be returned within 30 days
D2: Extended holiday returns allow 45 days
```

### Retrieved Docs @ K=3:

```
D1, D2, D3
```

### Metrics:

```
Precision@3 = 2 / 3 = 0.67
Recall@3 = 2 / 2 = 1.0
```

💡 **Interpretation**

* High Recall ✅ (nothing missed)
* Medium Precision ⚠️ (some noise)

---

# 5️⃣ When to Use (IMPORTANT)

| Scenario           | Metric         |
| ------------------ | -------------- |
| Missed answers     | Recall@K       |
| Noisy retrieval    | Precision@K    |
| Compliance / legal | High Recall    |
| Cost & latency     | High Precision |

# 6️⃣ FULL Python Code (Works with Any Vector DB)

### 🔹 Step 1: Define Ground Truth

```python
ground_truth_docs = {
    "Can I return electronics after 40 days?": [
        "Extended holiday returns allow 45 days"
    ]
}
```

---

### 🔹 Step 2: Run Retriever

```python
def retrieve_docs(retriever, query, k=3):
    docs = retriever.get_relevant_documents(query)
    return [doc.page_content for doc in docs[:k]]
```

---

### 🔹 Step 3: Precision@K / Recall@K Implementation

```python
def precision_at_k(retrieved_docs, relevant_docs, k):
    retrieved_k = retrieved_docs[:k]
    relevant_count = sum(
        1 for doc in retrieved_k if doc in relevant_docs
    )
    return relevant_count / k


def recall_at_k(retrieved_docs, relevant_docs, k):
    retrieved_k = retrieved_docs[:k]
    relevant_count = sum(
        1 for doc in retrieved_k if doc in relevant_docs
    )
    return relevant_count / len(relevant_docs)
```

---

### 🔹 Step 4: Run Evaluation

```python
query = "Can I return electronics after 40 days?"

retrieved = retrieve_docs(retriever, query, k=3)
relevant = ground_truth_docs[query]

p_at_3 = precision_at_k(retrieved, relevant, k=3)
r_at_3 = recall_at_k(retrieved, relevant, k=3)

print("Precision@3:", p_at_3)
print("Recall@3:", r_at_3)
```

## 9️⃣ Optimization Playbook (VERY IMPORTANT)

### 🔹 Low Recall@K

* Increase `k`
* Improve chunking
* Better embeddings
* Metadata filters

### 🔹 Low Precision@K

* Use **HNSW / IVF**
* Reduce chunk size
* Add reranking
* Hybrid search (BM25 + Vector)
