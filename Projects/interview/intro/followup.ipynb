# 🔥 Follow-Up Interview Q&A

**Agentic AI – Return Policy System (Costco)**

---

## Q1️⃣ Why didn’t you use a simple RAG chatbot?

### ✅ Best Answer

> “A simple RAG chatbot can retrieve documents, but it cannot handle **multi-step decision logic**. Our return policy requires validating eligibility, checking time windows, determining refund type, and handling exceptions. That level of reasoning and control required an **Agentic AI system with explicit state and tool orchestration**, which is why we used LangGraph and MCP.”

---

## Q2️⃣ What makes this system *agentic*?

### ✅ Best Answer

> “The system is agentic because it can **reason, act, observe, and decide** across multiple steps. It doesn’t just retrieve information — it executes tools, maintains state, validates constraints, and dynamically decides the next step based on intermediate results.”

---

## Q3️⃣ What agent patterns did you use and why?

### ✅ Best Answer

> “We primarily used **ReAct** for reasoning and tool usage, combined with a **Planner–Executor** pattern for complex queries. This allowed us to break down requests like refund eligibility into smaller steps and ensure each decision followed policy constraints.”

---

## Q4️⃣ How did you prevent hallucinations?

### ✅ Best Answer

> “We controlled hallucinations using three techniques:
>
> 1. **Grounding via RAG** — all responses are based on retrieved policy chunks.
> 2. **Tool-first execution** — eligibility and dates are validated via tools, not LLM guesses.
> 3. **State-driven LangGraph flows** — the agent cannot skip mandatory validation steps.”

---

## Q5️⃣ Why did you choose LangGraph over plain LangChain?

### ✅ Best Answer

> “LangChain chains are linear, but enterprise workflows are not. LangGraph gave us **explicit state management, branching, retries, and failure handling**, which are essential for policy-driven decision systems. It’s much more production-ready for Agentic AI.”

---

## Q6️⃣ What role does the MCP server play?

### ✅ Best Answer

> “The MCP server acts as the **agent runtime and orchestration layer**. It manages tool execution, LLM calls, policy retrieval, and workflow state. This separation makes the system scalable, observable, and easy to extend.”

---

## Q7️⃣ How do you handle structured and unstructured data together?

### ✅ Best Answer

> “We use **RAG for unstructured data** like policy documents and FAQs, and **tool-based queries** for structured data such as order details, purchase dates, and item categories. The agent merges both contexts before generating a final response.”

---

## Q8️⃣ How is vector search optimized?

### ✅ Best Answer

> “We use **ANN-based vector search (HNSW)** to ensure low-latency retrieval. We tuned index parameters and chunk sizes to keep retrieval under acceptable thresholds, which is critical for real-time agent responses.”

---

## Q9️⃣ How do you handle policy updates?

### ✅ Best Answer

> “Policy updates are handled through **re-ingestion and re-embedding pipelines**. Since agent logic is separated from policy content, updates don’t require changes to the agent workflow — only the vector index refresh.”

---

## Q1️⃣0️⃣ How do you ensure scalability?

### ✅ Best Answer

> “We designed the system using **stateless FastAPI services**, externalized memory/state management, and modular agent components. This allows horizontal scaling and independent evolution of tools, models, and workflows.”

---

## Q1️⃣1️⃣ How do you test an Agentic AI system?

### ✅ Best Answer

> “We test at multiple levels:
>
> * **Unit tests** for tools and policy rules
> * **Golden test cases** for known scenarios
> * **Agent trace validation** to ensure correct step execution
> * Regression testing when policies change”

---

## Q1️⃣2️⃣ How do you measure success?

### ✅ Best Answer

> “We track metrics like:
>
> * Reduction in manual policy lookups
> * Accuracy of eligibility decisions
> * Latency per query
> * Escalation rate to human agents
>
> The system reduced manual lookup time by around **40%**, which was a key success metric.”

---

## Q1️⃣3️⃣ What happens if the agent fails?

### ✅ Best Answer

> “Failures are handled through **fallback states** in LangGraph. If a tool fails or confidence is low, the agent escalates to a human or returns a safe, partial response instead of guessing.”

---

## Q1️⃣4️⃣ How do you handle security and access control?

### ✅ Best Answer

> “We enforce role-based access at the API layer, restrict sensitive data access through tools, and ensure the LLM only sees the **minimum required context**.”

---

## Q1️⃣5️⃣ What would you improve if you had more time?

### ✅ Best Answer

> “I would add **confidence scoring**, more advanced evaluation metrics like RAGAS, and proactive agent learning from failure cases to continuously improve decision quality.”

---

# 🧠 Final 15-Second Closing Statement (Very Powerful)

> “This project demonstrates how Agentic AI combined with RAG can move beyond chatbots into **decision automation**, which is where real enterprise value lies.”
