# ✅ 1. MCP SERVER

👉 This exposes tools & resources (order, refund, policy)

### `mcp_server.py`

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(title="MCP Server")

# ------------------------------
# Fake data
# ------------------------------
ORDERS = {
    "ORD123": {
        "order_id": "ORD123",
        "item": "Shoes",
        "delivered": True,
        "return_window_days": 30,
        "days_since_delivery": 10,
        "amount": 2000
    }
}

POLICY_TEXT = """
Returns are allowed within 30 days from delivery.
Refund is issued to original payment method.
Damaged or wrong item is always eligible.
"""

# ------------------------------
# Schemas
# ------------------------------

class OrderRequest(BaseModel):
    order_id: str


class RefundRequest(BaseModel):
    order_id: str
    amount: float


# ------------------------------
# MCP style endpoints
# ------------------------------

@app.get("/mcp/resources/policy")
def get_policy():
    return {"policy": POLICY_TEXT}


@app.post("/mcp/tools/get_order")
def get_order(req: OrderRequest):
    return ORDERS.get(req.order_id, {})


@app.post("/mcp/tools/refund")
def initiate_refund(req: RefundRequest):
    return {
        "status": "SUCCESS",
        "order_id": req.order_id,
        "refunded_amount": req.amount
    }
```

Run:

```
uvicorn mcp_server:app --port 9000
```

# ✅ 2. MCP CLIENT

👉 Used by orchestrator & agents

### `mcp_client.py`

```python
import requests

class MCPClient:

    def __init__(self, base_url="http://localhost:9000"):
        self.base_url = base_url

    def get_policy(self):
        r = requests.get(f"{self.base_url}/mcp/resources/policy")
        return r.json()["policy"]

    def get_order(self, order_id: str):
        r = requests.post(
            f"{self.base_url}/mcp/tools/get_order",
            json={"order_id": order_id}
        )
        return r.json()

    def refund(self, order_id: str, amount: float):
        r = requests.post(
            f"{self.base_url}/mcp/tools/refund",
            json={"order_id": order_id, "amount": amount}
        )
        return r.json()
```

# ✅ 3. AGENTS + ORCHESTRATOR (Supervisor)

### `orchestrator.py`

This is your **multi-agent core**.

```python
from mcp_client import MCPClient


# -------------------------
# Agents
# -------------------------

class PolicyAgent:

    def __init__(self, mcp: MCPClient):
        self.mcp = mcp

    def run(self):
        return self.mcp.get_policy()


class OrderValidationAgent:

    def __init__(self, mcp: MCPClient):
        self.mcp = mcp

    def run(self, order_id: str):
        return self.mcp.get_order(order_id)


class FraudAgent:

    def run(self, order_data: dict):
        # simple mock logic
        return {"risk": "LOW"}


class DecisionAgent:

    def run(self, order_data: dict, fraud_result: dict):
        if not order_data:
            return {"approved": False, "reason": "Order not found"}

        if not order_data["delivered"]:
            return {"approved": False, "reason": "Not delivered"}

        if order_data["days_since_delivery"] > order_data["return_window_days"]:
            return {"approved": False, "reason": "Return window expired"}

        if fraud_result["risk"] == "HIGH":
            return {"approved": False, "reason": "High fraud risk"}

        return {
            "approved": True,
            "refund_amount": order_data["amount"]
        }


class ExecutionAgent:

    def __init__(self, mcp: MCPClient):
        self.mcp = mcp

    def run(self, order_id: str, amount: float):
        return self.mcp.refund(order_id, amount)


# -------------------------
# Orchestrator / Supervisor
# -------------------------

class Orchestrator:

    def __init__(self):
        self.mcp = MCPClient()

        self.policy_agent = PolicyAgent(self.mcp)
        self.order_agent = OrderValidationAgent(self.mcp)
        self.fraud_agent = FraudAgent()
        self.decision_agent = DecisionAgent()
        self.execution_agent = ExecutionAgent(self.mcp)

    def handle_request(self, order_id: str):

        policy = self.policy_agent.run()

        order_data = self.order_agent.run(order_id)

        fraud_result = self.fraud_agent.run(order_data)

        decision = self.decision_agent.run(order_data, fraud_result)

        if not decision["approved"]:
            return {
                "status": "REJECTED",
                "reason": decision["reason"],
                "policy": policy
            }

        execution = self.execution_agent.run(
            order_id,
            decision["refund_amount"]
        )

        return {
            "status": "APPROVED",
            "execution": execution,
            "policy": policy
        }
```

# ✅ 4. FASTAPI – Agent runtime

### `main.py`

This is your **agentic backend**.

```python
from fastapi import FastAPI
from pydantic import BaseModel
from orchestrator import Orchestrator

app = FastAPI()

orchestrator = Orchestrator()


class Query(BaseModel):
    order_id: str


@app.post("/query")
def run_agent(query: Query):
    result = orchestrator.handle_request(query.order_id)
    return result
```

Run:

```
uvicorn main:app --port 8000 --reload
```
