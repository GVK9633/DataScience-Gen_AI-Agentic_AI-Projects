# 🧠 What is RAGAS?

**RAGAS (Retrieval-Augmented Generation Assessment)**
is a framework to **evaluate RAG systems objectively** instead of relying on human judgment.

It evaluates **both retrieval quality and generation quality**.

---

## 🔑 Why RAGAS is Important (Say This in Interview)

> “LLM accuracy alone is not enough. In RAG systems, we must measure whether the **right context was retrieved** and whether the **answer was grounded** in that context. RAGAS provides standardized metrics for this.”

---

# 📊 Core RAGAS Metrics (MUST KNOW)

| Metric                | What it Measures                        | Interview Meaning   |
| --------------------- | --------------------------------------- | ------------------- |
| **Context Precision** | How relevant retrieved chunks are       | Retrieval quality   |
| **Context Recall**    | Whether all required info was retrieved | Missing knowledge   |
| **Faithfulness**      | Is answer grounded in context?          | Hallucination check |
| **Answer Relevancy**  | Does answer match the question?         | User satisfaction   |

# 🧪 Step-by-Step RAGAS Evaluation Code

### 1️⃣ Prepare Evaluation Dataset

```python
from datasets import Dataset

data = {
    "question": [
        "Can I return electronics after 40 days?"
    ],
    "answer": [
        "Yes, electronics can be returned within 45 days during extended holiday returns."
    ],
    "contexts": [[
        "Electronics can be returned within 30 days.",
        "Extended holiday returns allow 45 days."
    ]],
    "ground_truth": [
        "Electronics can be returned within 45 days during extended holiday returns."
    ]
}

dataset = Dataset.from_dict(data)
```

---

### 2️⃣ Define LLM & Embeddings (Same as RAG)

```python
from langchain_openai import ChatOpenAI, OpenAIEmbeddings

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
embeddings = OpenAIEmbeddings()
```

---

### 3️⃣ Run RAGAS Evaluation

```python
from ragas import evaluate
from ragas.metrics import (
    context_precision,
    context_recall,
    faithfulness,
    answer_relevancy
)

results = evaluate(
    dataset,
    metrics=[
        context_precision,
        context_recall,
        faithfulness,
        answer_relevancy
    ],
    llm=llm,
    embeddings=embeddings
)

print(results)
```

---

## 📈 Sample Output (What to Expect)

# 🧠 How to Interpret Results (INTERVIEW GOLD)

| Score      | Meaning            |
| ---------- | ------------------ |
| > 0.9      | Excellent          |
| 0.75 – 0.9 | Acceptable         |
| < 0.7      | Needs optimization |

Say this:

> “If faithfulness drops, it indicates hallucination. If context recall is low, retrieval is missing important documents.”

# ⚙️ How We Improve Scores (Real-World Answer)

### 🔹 Low Context Recall

* Increase `k` in retriever
* Improve chunking
* Add metadata filtering

### 🔹 Low Faithfulness

* Reduce prompt freedom
* Enforce citation-based answers
* Use tool-first logic

### 🔹 Low Precision

* Switch to HNSW
* Improve embedding quality
* Use reranking
