## How do you evaluate an LLM? — Interview Answer

**LLM evaluation means measuring whether the model produces correct, relevant, safe, and useful responses for a given task.**

### 1. What do we evaluate?

| Metric            | What it checks                                   | Example                          |
| ----------------- | ------------------------------------------------ | -------------------------------- |
| **Correctness**   | Is the answer factually correct?                 | Expected: 7 days → Model: 7 days |
| **Relevance**     | Does it answer the question?                     | No unnecessary information       |
| **Faithfulness**  | Is the answer supported by the provided context? | Important for RAG                |
| **Coherence**     | Is the answer logical and understandable?        | No contradictory statements      |
| **Hallucination** | Did the model invent information?                | Fake policy or order status      |
| **Safety**        | Does it avoid harmful/unauthorized responses?    | Doesn't expose customer data     |
| **Latency**       | How long does it take?                           | 1.5 seconds                      |
| **Cost**          | How expensive is each request?                   | Token/API cost                   |

---

## 2. How do you actually evaluate it?

A typical evaluation pipeline is:

```text
Test Dataset
     ↓
Send questions to LLM
     ↓
Get Model Responses
     ↓
Compare with Expected Answer / Context
     ↓
Calculate Metrics
     ↓
Analyze Failures
     ↓
Improve Prompt / RAG / Model
     ↓
Run Evaluation Again
```

### Example

Suppose you are building your **Return Policy Agent**.

Test question:

> "How many days do I have to return a product?"

Expected answer:

> "Products can be returned within 7 days."

Model response:

> "Products can be returned within 7 days."

Evaluation:

```text
Correctness   → PASS
Relevance     → PASS
Faithfulness  → PASS
Hallucination → PASS
```

Another response:

> "Products can be returned within 30 days."

```text
Correctness   → FAIL
Faithfulness  → FAIL
```

# 3. What is LLM-as-a-Judge?

Instead of manually checking every response, we can use another LLM to evaluate the answer.

```text
                User Question
                     ↓
                  LLM A
                     ↓
                Answer
                     ↓
                  LLM B
              "Judge / Evaluator"
                     ↓
             Score + Explanation
```

For example, evaluator prompt:

```text
Question: How many days is the return period?

Expected: 7 days

Model Answer: 30 days

Evaluate correctness from 1-5.
```

Judge:

```text
Score: 1/5
Reason: The answer contradicts the expected policy.
```

### Interview answer

> **"LLM-as-a-Judge uses another LLM to evaluate the quality of the generated response against defined criteria such as correctness, relevance and faithfulness."**

But don't depend only on an LLM judge. Combine it with **deterministic checks and human evaluation**, especially for critical applications.

# 4. RAG Evaluation

For your **RAG/Agentic AI interviews**, this is particularly important.

You evaluate two major parts:

### A. Retrieval evaluation

Did we retrieve the right documents/chunks?

```text
Question
   ↓
Retriever
   ↓
Top-K chunks
```

Metrics include:

* **Precision@K**
* **Recall@K**
* **MRR**
* **NDCG**

### B. Generation evaluation

Did the LLM generate a good answer from those chunks?

Check:

* Faithfulness
* Answer correctness
* Relevance
* Hallucination

```text
          RAG
           |
     ┌─────┴─────┐
     ↓           ↓
Retrieval     Generation
     ↓           ↓
Precision      Faithfulness
Recall         Correctness
MRR            Relevance
```

---

# 5. Agent Evaluation

For an **Agentic AI system**, evaluating only the final text isn't enough.

You should also evaluate:

```text
User Request
     ↓
Agent
     ↓
Planning        ← Was the plan correct?
     ↓
Tool Selection  ← Correct tool?
     ↓
Tool Arguments  ← Correct parameters?
     ↓
Tool Result
     ↓
Final Answer    ← Correct?
```

### Example

User:

> "Check my order status."

Agent should call:

```python
get_order_status(order_id="1234")
```

If it calls:

```python
cancel_order(order_id="1234")
```

the final text could look perfect, but the **agent execution is wrong**.

So evaluate:

* Tool selection
* Tool arguments
* Number of steps
* Task completion
* Final answer
* Safety/authorization

## RAG / Retrieval-Augmented Generation — Q1 to Q15

### Q1. Why is RAG important in modern LLM systems?

**Definition:** RAG combines **retrieval of relevant external information** with **LLM generation**.

**Why important?**

* LLM knowledge may be outdated.
* Reduces hallucination by grounding answers in retrieved documents.
* Allows private/company data without retraining the LLM.
* Knowledge can be updated by updating the data/index.

**Example:**
Customer asks: *“What is our return policy for products after 30 days?”*
RAG retrieves the latest return-policy document → LLM generates the answer.

**Interview answer:**

> “RAG improves LLM responses by retrieving relevant external knowledge and providing it as context to the LLM. It helps ground answers, reduce hallucinations, and use up-to-date or private data.”

---

### Q2. Retriever vs Generator

| Component     | What it does                                            |
| ------------- | ------------------------------------------------------- |
| **Retriever** | Finds relevant documents/chunks                         |
| **Generator** | Uses the retrieved context to generate the final answer |

**Flow:**

```text
User Query
    ↓
Retriever
    ↓
Relevant Chunks
    ↓
LLM / Generator
    ↓
Final Answer
```

**Example:**
Query: *“Can I return a product after 15 days?”*

Retriever → finds return-policy section
Generator → converts that information into a natural-language answer.

**Interview answer:**

> “The retriever finds relevant information from the knowledge base, while the generator, usually an LLM, uses that retrieved context to produce the final answer.”

# Q3. Sparse vs Dense vs Hybrid Retrieval

### Sparse Retrieval

Uses keyword matching.

Example:

```text
Query: "return policy 30 days"
```

BM25 searches for matching terms like `return`, `policy`, `30 days`.

**Good for:** exact keywords, IDs, names, error codes.

### Dense Retrieval

Converts query and documents into **embeddings/vectors** and searches based on semantic similarity.

```text
"How long can I return this?"
        ↓
     Embedding
        ↓
Vector Search
```

**Good for:** semantic meaning.

### Hybrid Retrieval

Combines **sparse + dense** retrieval.

```text
                 Query
                   ↓
          ┌────────┴────────┐
          ↓                 ↓
       BM25             Vector Search
          ↓                 ↓
          └────────┬────────┘
                   ↓
              Reranking
                   ↓
             Top Chunks
```

**Interview answer:**

> “Sparse retrieval is keyword-based, dense retrieval is semantic embedding-based, and hybrid retrieval combines both to improve retrieval coverage and accuracy.”

# Q4. HNSW vs IVF

These are **vector indexing/search strategies**.

|             | HNSW                               | IVF                            |
| ----------- | ---------------------------------- | ------------------------------ |
| Full form   | Hierarchical Navigable Small World | Inverted File Index            |
| Idea        | Graph-based search                 | Cluster-based search           |
| Search      | Traverse graph                     | Search selected clusters       |
| Speed       | Very fast                          | Fast                           |
| Memory      | Usually higher                     | Usually lower                  |
| Best suited | Fast high-quality ANN search       | Large-scale vector collections |

### Simple example

Suppose you have **10 million embeddings**.

**HNSW:**

```text
Vector
  ↓
Graph
  ↓
Navigate nearest nodes
  ↓
Nearest vectors
```

**IVF:**

```text
10M vectors
     ↓
Clusters
 ↓   ↓   ↓
C1  C2  C3 ...
     ↓
Search relevant clusters
```

**Interview answer:**

> “HNSW uses a graph structure to efficiently navigate to nearest vectors, while IVF partitions vectors into clusters and searches selected clusters. HNSW often provides strong recall and low latency, while IVF can be more memory-efficient at large scale.”

# Q5. Parent Document vs Sentence Window Retrieval

Both are techniques to solve a common problem:

> **Small chunks are good for retrieval but may lack enough context.**

### Parent Document Retrieval

Store small chunks for searching but return their **larger parent document/section**.

```text
Parent Document
      ↓
 ┌────┼────┐
Chunk1 Chunk2 Chunk3
   ↓
Retrieve Chunk2
   ↓
Return Parent Document/Section
```

### Sentence Window Retrieval

Retrieve a sentence but provide nearby sentences as context.

```text
S1
S2
S3 ← matched sentence
S4
S5
```

Return:

```text
S2 + S3 + S4
```

**Interview answer:**

> “Parent document retrieval searches using small chunks but returns the larger parent context. Sentence-window retrieval retrieves a matching sentence along with surrounding sentences to preserve context.”

# Q6. How do you solve the "Lost in the Middle" problem?

**Problem:**
When an LLM receives a very long context, information placed in the **middle** may receive less attention than information near the beginning or end.

### Solutions

1. Retrieve fewer but higher-quality chunks.
2. Rerank retrieved documents.
3. Put the most relevant information strategically in the prompt.
4. Remove duplicate/irrelevant context.
5. Use summarization/compression.
6. Use models designed to handle long contexts effectively.

**Example:**

Instead of:

```text
50 irrelevant chunks
+
5 relevant chunks
```

use:

```text
Top 5 highly relevant chunks
        ↓
LLM
```

**Interview answer:**

> “I address lost-in-the-middle by reducing unnecessary context, reranking the retrieved chunks, prioritizing highly relevant information, and using context compression or summarization when necessary.”

# Q7. How does HyDE improve retrieval?

**HyDE = Hypothetical Document Embeddings**

Instead of directly embedding the user's question, an LLM first generates a **hypothetical answer/document**, then embeds that generated text.

### Normal RAG

```text
Question
   ↓
Embedding
   ↓
Vector Search
```

### HyDE

```text
Question
   ↓
LLM generates hypothetical answer
   ↓
Embedding
   ↓
Vector Search
   ↓
Real documents
```

### Example

User:

> "How do I reset my company VPN?"

LLM generates a hypothetical document about VPN reset steps.

That hypothetical document is embedded and used to search the real knowledge base.

**Important:** The hypothetical answer is used for **retrieval**, not blindly given to the user.

**Interview answer:**

> “HyDE generates a hypothetical answer to the query and uses its embedding for retrieval. This can improve semantic matching when the original query is short or poorly aligned with document wording.”

# Q8. What is pre-retrieval query reformulation?

It means **changing the user's query before sending it to the retriever**.

### Example

User asks:

> "How do I fix this?"

Not enough information.

System may reformulate it as:

> "Troubleshooting steps for application login failure"

Then perform retrieval.

### Techniques

* Query expansion
* Query rewriting
* Query decomposition
* Synonym expansion
* Conversation-aware rewriting

**Interview answer:**

> “Pre-retrieval query reformulation improves retrieval by rewriting or expanding the user's query into a clearer search query before sending it to the retriever.”

# Q9. Standard RAG vs CRAG vs Self-RAG

| Type             | Main idea                                                                                     |
| ---------------- | --------------------------------------------------------------------------------------------- |
| **Standard RAG** | Retrieve → generate                                                                           |
| **CRAG**         | Evaluate retrieval quality and take corrective action                                         |
| **Self-RAG**     | Model decides when retrieval is needed and evaluates its own retrieved information/generation |

### Standard RAG

```text
Query
 ↓
Retrieve
 ↓
Generate
```

### CRAG

```text
Query
 ↓
Retrieve
 ↓
Evaluate retrieval
 ↓
Good? ── Yes → Generate
  │
  No
  ↓
Correct / additional retrieval
```

### Self-RAG

```text
Query
 ↓
Should I retrieve?
 ↓
Retrieve
 ↓
Evaluate evidence
 ↓
Generate
 ↓
Evaluate response
```

**Interview answer:**

> “Standard RAG always follows retrieval and generation. CRAG evaluates retrieval quality and performs corrective retrieval when needed, while Self-RAG enables the model to decide when to retrieve and evaluate the retrieved evidence and generated response.”

# Q10. How do you design an Agentic RAG system?

**Traditional RAG:**

```text
Query → Retrieve → Generate
```

**Agentic RAG:**

```text
                 User Query
                     ↓
                AI Agent
                     ↓
              Analyze Intent
                     ↓
          ┌──────────┼──────────┐
          ↓          ↓          ↓
       Vector DB   SQL DB    API/Tool
          ↓          ↓          ↓
          └──────────┼──────────┘
                     ↓
                 Evaluate
                     ↓
              Generate Answer
```

The agent can:

* decide whether retrieval is needed
* choose which knowledge source to use
* perform multiple retrievals
* call tools/APIs
* verify information
* retry when retrieval is poor
* ask for human approval for sensitive actions

**Interview answer:**

> “Agentic RAG adds reasoning and decision-making around retrieval. The agent can select the appropriate knowledge source or tool, perform iterative retrieval, evaluate results, and then generate a grounded response.”

# Q11. Graph RAG vs Traditional Vector RAG

### Vector RAG

Retrieves documents based mainly on **semantic similarity**.

```text
Query
 ↓
Embedding
 ↓
Vector DB
 ↓
Similar chunks
```

### Graph RAG

Represents information as **entities and relationships**.

```text
Customer
   ↓
placed
   ↓
Order
   ↓
contains
   ↓
Product
```

Useful for questions involving **relationships across multiple entities/documents**.

| Vector RAG                    | Graph RAG                    |
| ----------------------------- | ---------------------------- |
| Similarity-based              | Relationship-based           |
| Excellent for semantic search | Good for connected knowledge |
| Simpler architecture          | More complex                 |
| Chunks/embeddings             | Entities/relationships       |

**Interview answer:**

> “Traditional vector RAG retrieves semantically similar chunks, while Graph RAG uses entities and relationships to answer questions requiring connected or multi-hop information.”

# Q12. Do long-context models make RAG obsolete?

**No.**

Long-context models allow more information to be placed into the prompt, but RAG still provides advantages:

* Reduces amount of context sent to LLM.
* Reduces cost.
* Improves retrieval of relevant information.
* Handles very large knowledge bases.
* Supports frequently changing/private information.
* Can provide source citations.

### Example

Company has:

```text
10 million documents
```

You cannot practically send all documents to the LLM.

RAG:

```text
10M documents
      ↓
Retriever
      ↓
Top 10 relevant chunks
      ↓
LLM
```

**Interview answer:**

> “Long-context models reduce some limitations of RAG, but they don't make RAG obsolete. For large, dynamic, private knowledge bases, retrieval remains useful for controlling context, latency, cost, and relevance.”

# Q13. What is RAGAS?

**RAGAS = Retrieval Augmented Generation Assessment**

It is a framework for evaluating RAG systems.

Common evaluation dimensions include:

| Metric                | Meaning                                       |
| --------------------- | --------------------------------------------- |
| **Faithfulness**      | Is the answer supported by retrieved context? |
| **Answer Relevancy**  | Does the answer address the question?         |
| **Context Precision** | Are retrieved chunks relevant?                |
| **Context Recall**    | Did retrieval capture the needed information? |

### Important distinction

You should evaluate **both retrieval and generation**.

```text
              RAG Evaluation
                    ↓
        ┌───────────┴───────────┐
        ↓                       ↓
   Retrieval Quality       Answer Quality
        ↓                       ↓
 Context Recall           Faithfulness
 Context Precision        Answer Relevancy
```

**Interview answer:**

> “RAGAS is used to evaluate RAG pipelines using metrics such as faithfulness, answer relevancy, context precision, and context recall, helping identify whether problems come from retrieval or generation.”

# Q14. How do you optimize RAG for accuracy and latency?

This is a **very important production interview question**.

### Accuracy

* Improve chunking.
* Improve embeddings.
* Use hybrid search.
* Add reranking.
* Tune top-K.
* Use metadata filtering.
* Use query rewriting.
* Remove irrelevant chunks.
* Evaluate retrieval separately from generation.

### Latency

* Use efficient vector indexes such as HNSW/IVF where appropriate.
* Reduce top-K.
* Use caching.
* Parallelize independent retrieval/tool calls.
* Use smaller/faster models where appropriate.
* Reduce prompt/context size.
* Stream responses.

### Production balance

```text
                  RAG
                   ↓
        ┌──────────┴──────────┐
        ↓                     ↓
     Accuracy               Latency
        ↓                     ↓
 Hybrid Search             Caching
 Reranking                 Top-K tuning
 Better chunks             Parallel calls
 Better embeddings         Smaller context
```

**Interview answer:**

> “For accuracy I optimize chunking, embeddings, hybrid retrieval, reranking and query reformulation. For latency I reduce unnecessary context, tune top-K, use efficient indexes, caching, parallel calls and appropriate model sizes.”

# Q15. What are red flags in RAG system design?

These are common things that should make you question the design.

### 🚩 1. Sending huge amounts of documents to the LLM

```text
10,000 chunks → LLM
```

Instead, retrieve and filter relevant chunks.

### 🚩 2. No retrieval evaluation

If you only measure final answers, you don't know whether the problem is:

```text
Data → Retrieval → Prompt → LLM
```

### 🚩 3. No citations/source tracking

Hard to verify whether the answer is grounded.

### 🚩 4. No metadata filtering

Example:

```text
department = Finance
region = India
document_type = Policy
```

could dramatically improve retrieval.

### 🚩 5. Blindly increasing Top-K

More chunks ≠ better answer.

### 🚩 6. No handling of stale documents

Enterprise knowledge changes.

### 🚩 7. No access control

A RAG system must not retrieve documents the user is unauthorized to access.

### 🚩 8. No monitoring/evaluation

Production RAG needs monitoring for:

* retrieval quality
* latency
* token usage
* hallucination/faithfulness
* failures
* user feedback

**Interview answer:**

> “Major RAG red flags include retrieving too much context, no retrieval evaluation, poor chunking, no access control, stale data, blindly increasing top-K, and lack of production monitoring.”
