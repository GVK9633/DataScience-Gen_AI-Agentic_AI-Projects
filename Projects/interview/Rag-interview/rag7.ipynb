# 🧠 Hallucination Detection Architecture (RAG)

```
┌───────────────┐
│   User Query  │
└───────┬───────┘
        │
        ▼
┌──────────────────────┐
│ Query Pre-processing │
│ (rewrite / normalize)│
└────────┬─────────────┘
         │
         ▼
┌────────────────────────────┐
│ Retrieval Layer             │
│  • Vector Search            │
│  • BM25 (optional)          │
│  • RAG Fusion (optional)    │
└────────┬───────────────────┘
         │
         ▼
┌────────────────────────────┐
│ Retrieved Context           │
│ (Top-K validated chunks)    │
└────────┬───────────────────┘
         │
         ▼
┌────────────────────────────┐
│ LLM Answer Generation       │
│ (Context-grounded prompt)   │
└────────┬───────────────────┘
         │
         ▼
┌──────────────────────────────────────────┐
│ Hallucination Detection & Validation      │
│                                          │
│  1️⃣ Keyword / Sentence Grounding         │
│  2️⃣ Semantic Similarity Check            │
│  3️⃣ LLM Faithfulness Judge               │
│  4️⃣ Source Attribution Validation        │
└────────┬─────────────────────────────────┘
         │
   ┌─────▼─────┐         ┌─────────────────────┐
   │  PASS     │         │       FAIL          │
   └─────┬─────┘         └─────────┬───────────┘
         │                         │
         ▼                         ▼
┌──────────────────┐     ┌─────────────────────────┐
│ Final Answer     │     │ Safe Response            │
│ + Citations     │     │ • “Not verifiable”        │
└──────────────────┘     │ • Re-retrieve            │
                          │ • Ask clarification      │
                          └─────────────────────────┘
```

# 🔍 What Each Layer Does (Interview Explanation)

## 1️⃣ Retrieval Layer (Root Cause Fix)

> **Most hallucinations come from bad retrieval**

**Techniques used:**

* Smart chunking
* Metadata filtering
* Hybrid search
* **RAG Fusion**

---

## 2️⃣ Context-Grounded Generation

**Strict rule:**

> “Answer ONLY using provided context. If not found, say so.”

Prevents the LLM from using its memory.

---

## 3️⃣ Hallucination Detection Layer (CRITICAL)

### ✔ Layered Validation

| Check               | Purpose                  |
| ------------------- | ------------------------ |
| Keyword grounding   | Fast rejection           |
| Semantic similarity | Meaning-level validation |
| LLM judge           | Logical faithfulness     |
| Source attribution  | Proof enforcement        |

---

## 4️⃣ Decision Layer

* **PASS** → Return answer with citations
* **FAIL** → Reject, re-retrieve, or ask clarification
