# **“How do you reduce hallucinations in RAG systems?”**

## ✅ Short Interview Answer (30 seconds)

> *“We reduce hallucinations by grounding the LLM strictly in retrieved context, improving retrieval quality, enforcing answer-from-context prompts, validating answer faithfulness, and rejecting responses when supporting evidence is missing.”*

---

## 🧠 Detailed Interview Answer (2–3 minutes)

We reduce hallucinations at **five layers** in real projects.

---

# 1️⃣ Improve Retrieval Quality (MOST IMPORTANT)

> **Most hallucinations come from bad retrieval, not the LLM.**

### 🔹 What We Do

✔ Smart chunking (300–500 tokens)
✔ Metadata filtering
✔ Hybrid search (vector + BM25)
✔ RAG Fusion for higher recall

### 🔹 Interview Line

> “If the right document isn’t retrieved, the LLM will hallucinate.”

---

# 2️⃣ Strict Context-Grounded Prompting

### 🔹 Prompt Rule

```text
Answer ONLY using the provided context.
If the answer is not found, say "Information not available."
```

### 🔹 Why It Works

✔ Prevents model from using internal knowledge
✔ Forces explicit grounding

---

# 3️⃣ Limit Context & Tune `k`

### 🔹 Best Practice

* `k = 3–5`
* Remove duplicates
* Re-rank top documents

### 🔹 Why

Too much context = confusion = hallucinations

---

# 4️⃣ Faithfulness Validation (CRITICAL)

### 🔹 Automated Checks

✔ Answer vs context similarity
✔ LLM-based verification
✔ Keyword overlap checks

### 🔹 Real Behavior

If validation fails → **reject or re-ask**

### 🔹 Interview Line

> “We never blindly trust the LLM output.”

---

# 5️⃣ Enforce Source Attribution

### 🔹 What We Do

* Every answer must cite document IDs
* No citation → no answer

### 🔹 Example

```
Answer:
Electronics can be returned within 30 days.
(Source: ReturnPolicy_v3.pdf)
```

---

# 6️⃣ Confidence-Based Response Control

### 🔹 Strategy

✔ Low confidence → partial answer or refusal
✔ Ask clarifying question if needed

---

# 7️⃣ Guardrails & Safety Rules (Enterprise)

✔ Role-based access control
✔ PII masking
✔ Domain constraints
✔ Refusal policies

---

# 8️⃣ Continuous Monitoring & Feedback Loop

```
User Feedback
   ↓
Hallucination Detection
   ↓
Retrieval Fix
   ↓
Re-index
```

---

# 9️⃣ Real-World Example (Explain This)

**Query:**
“What is the return policy for opened electronics?”

**System Behavior:**
✔ Retrieves policy doc
✔ Finds “opened items not eligible”
✔ Answers correctly
✔ Rejects if doc not found

In production, retrieval accuracy matters more than model size.”*
