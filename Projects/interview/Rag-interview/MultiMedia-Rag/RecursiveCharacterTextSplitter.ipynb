### One-line explanation

→ Creates a splitter that breaks long text into **500-character chunks**, while keeping **100 characters overlapping** between consecutive chunks.

---

### What each part means (very short)

* **`RecursiveCharacterTextSplitter`**
  → Splits text *recursively* using smart boundaries (paragraph → line → sentence → word → char) to avoid cutting in the middle when possible.

* **`chunk_size=500`**
  → Each output chunk will be **at most 500 characters**.

* **`chunk_overlap=100`**
  → The **last 100 characters of the previous chunk** are repeated in the next chunk to preserve context.

---

### Tiny example

Suppose your text is:

```
[0...............................................600]
```

You will get:

```
Chunk 1 → chars 0   to 500
Chunk 2 → chars 400 to 900   (100 overlap from previous)
```

---

### Why this is important for your RAG project

Because overlapping:

* prevents losing context at chunk boundaries
* improves retrieval quality when the answer spans two chunks
