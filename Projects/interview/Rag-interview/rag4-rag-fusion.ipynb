# 3️⃣ SAME Example – Now with RAG Fusion (Code)

---

## 🔹 Step 1: Documents (Same as Before)

```python
from langchain.docstore.document import Document

documents = [
    Document(page_content="Electronics can be returned within 30 days of purchase."),
    Document(page_content="Opened electronic items are not eligible for return."),
    Document(page_content="Refunds are processed within 5 business days."),
    Document(page_content="Damaged items can be replaced within 7 days.")
]
```

---

## 🔹 Step 2: Chunking (Scales to Large Corpora)

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=400,
    chunk_overlap=50
)

chunks = splitter.split_documents(documents)
```

---

## 🔹 Step 3: Vector Store

```python
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS

embeddings = OpenAIEmbeddings()
db = FAISS.from_documents(chunks, embeddings)
```

---

## 🔹 Step 4: RAG Fusion Retriever (KEY PART)

```python
from langchain.retrievers.multi_query import MultiQueryRetriever
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

fusion_retriever = MultiQueryRetriever.from_llm(
    retriever=db.as_retriever(search_kwargs={"k": 3}),
    llm=llm
)
```

### 🎯 What Happens Internally

User query:

```
"What is the return policy for electronic products?"
```

Generated queries:

* "electronics return policy"
* "refund rules for electronics"
* "exchange eligibility for electronic items"

👉 Retrieval happens **for each query**
👉 Results are **fused (RRF)**
👉 Recall ↑ dramatically

---

## 🔹 Step 5: Generate Answer (Grounded Prompt)

```python
def generate_answer(query):
    docs = fusion_retriever.get_relevant_documents(query)
    context = "\n".join(d.page_content for d in docs)

    prompt = f"""
    Answer ONLY from the context below.
    If not found, say "Information not available".

    Context:
    {context}

    Question:
    {query}
    """

    answer = llm.invoke(prompt)
    return answer.content, docs
```

---

## 🔹 Step 6: Faithfulness Validation (Critical)

```python
def validate_answer(answer, docs):
    context_text = " ".join(d.page_content for d in docs)
    return any(sentence.strip() in context_text for sentence in answer.split("."))
```

---

## 🔹 Step 7: End-to-End Execution

```python
query = "What is the return policy for electronic products?"

answer, docs = generate_answer(query)

if validate_answer(answer, docs):
    print("✅ Validated Answer:\n", answer)
else:
    print("❌ Answer rejected (not grounded)")
```

# 5️⃣ REAL RAG FAILURE SCENARIOS & FIXES (VERY IMPORTANT)

---

## ❌ Failure 1: Wrong / Irrelevant Documents Retrieved

### Cause

* Poor chunking
* Wrong embedding model
* No metadata filtering

### Fix

✅ Better chunk size
✅ Metadata filters
✅ Hybrid search (BM25 + vector)

> **Interview line:**
> “We first fix retrieval, not the LLM.”

---

## ❌ Failure 2: Hallucinated Answer

### Cause

* LLM answers from memory
* Weak prompt

### Fix

✅ “Answer ONLY from context”
✅ Faithfulness validation
✅ Reject unverifiable answers

---

## ❌ Failure 3: Missing Important Policy Clauses

### Cause

* Single query retrieval
* Low recall

### Fix

✅ **RAG Fusion**
✅ Increase query diversity
✅ Reciprocal Rank Fusion

---

## ❌ Failure 4: Too Many Documents → Noise

### Cause

* High `k`
* No filtering

### Fix

✅ Tune `k = 3–5`
✅ Metadata pre-filter
✅ Re-ranking

---

## ❌ Failure 5: Slow Response (Large Data)

### Cause

* Huge index
* No caching

### Fix

✅ Index sharding
✅ Cache frequent queries
✅ Hierarchical retrieval
