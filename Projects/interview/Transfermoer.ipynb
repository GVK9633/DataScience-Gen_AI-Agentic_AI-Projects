# ✅ 1. What is a Transformer?

👉 A **Transformer** is a deep learning architecture that uses **attention mechanisms** to process entire sequences in parallel instead of sequentially like RNNs.

👉 It was introduced in the *“Attention Is All You Need”* paper and is the backbone of modern LLMs like GPT-4 and BERT.

👉 It consists of **encoder and decoder blocks with self-attention layers**.

---

# ✅ 2. Why are Transformers better than RNN/LSTM?

👉 Transformers process **all words in parallel**, while RNNs process sequentially → much faster.

👉 They handle **long-range dependencies better** using attention (no vanishing gradient issue).

👉 More scalable and efficient for large datasets → ideal for GenAI.

👉 That’s why RNNs/LSTMs are mostly replaced in modern NLP.

---

# ✅ 3. What is Attention Mechanism?

👉 Attention is a technique that allows the model to **focus on important words in a sentence** while processing.

👉 Instead of treating all words equally, it assigns **weights (importance scores)** to each word.

👉 Example:
In *“The cat sat on the mat”*, for word “sat”, model focuses more on “cat”.

---

# ✅ 4. What is Self-Attention?

👉 Self-attention is a type of attention where **each word attends to every other word in the same sentence**.

👉 It helps understand context and relationships between words.

👉 Example:
“The animal didn’t cross the road because it was tired”
👉 “it” refers to **animal** using self-attention.

# ✅ 5. Difference between Encoder vs Decoder

👉 **Encoder**

* Reads full input at once (bidirectional context)
* Used for **understanding tasks**
* Example: BERT

👉 **Decoder**

* Generates output **token by token (left → right)**
* Uses **causal (masked) attention**
* Used for **text generation**
* Example: GPT-4

👉 **One-liner:** Encoder = understand, Decoder = generate

---

# ✅ 6. What is Causal Attention?

👉 Causal attention ensures that a token can **only attend to previous tokens, not future ones**.

👉 It uses a **mask** to block future words.

👉 Example:

```text
"I love AI"
```

While predicting “AI”, model can see only “I love” ✔

👉 Used in decoder models like GPT.

---

# ✅ 7. What is Multi-Head Attention?

👉 Instead of one attention, the model uses **multiple attention heads in parallel**.

👉 Each head learns **different relationships** (syntax, meaning, position).

👉 Outputs are combined → richer understanding.

👉 Improves performance over single attention.

---

# ✅ 8. What is Positional Encoding?

👉 Transformers don’t understand word order naturally.

👉 Positional encoding adds **position information** to each token.

👉 Example:

```text
"dog bites man" ≠ "man bites dog"
```

👉 Uses sine/cosine functions or learned embeddings.

---

# ✅ 9. What is Tokenization?

👉 Tokenization is the process of **splitting text into smaller units (tokens)**.

👉 Tokens can be:

* words
* subwords
* characters

👉 Example:

```text
"unhappiness" → ["un", "happi", "ness"]
```

👉 First step before feeding text into models.

# ✅ 11. What is Cross-Attention?

👉 Cross-attention is when **decoder attends to encoder output**.

👉 Used in encoder-decoder models (like translation).

👉 Example:

* Input: English sentence (encoder)
* Output: French sentence (decoder uses encoder info)

👉 Helps align input and output sequences.

---

# ✅ 12. How does Transformer handle long context?

👉 Standard transformers have **context window limits** (e.g., 4K, 8K tokens).

👉 Techniques to handle long context:

* Sliding window attention
* Sparse attention
* Chunking + RAG
* Memory mechanisms

👉 New models extend context to **100K+ tokens**

---

# ✅ 13. What is KV Cache?

👉 KV (Key-Value) cache stores **previous attention keys & values** during generation.

👉 Instead of recomputing for every token → reuse past computations.

👉 Improves **latency and speed** significantly in GPT-like models.

👉 Very important for real-time chat systems.

---

# ✅ 14. What is Hallucination in LLMs?

👉 Hallucination is when a model **generates incorrect or made-up information confidently**.

👉 Example:

* Fake facts, wrong citations

👉 Causes:

* Lack of knowledge
* Over-generalization

👉 Solutions:

* RAG (retrieve real data)
* Fine-tuning
* Prompt constraints
