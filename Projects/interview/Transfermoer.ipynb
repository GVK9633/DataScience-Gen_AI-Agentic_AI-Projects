# 🔹 What is a Transformer?

👉 A **Transformer** is a deep learning architecture used in **NLP and GenAI** that processes text using **attention instead of RNN/CNN**.

📌 Introduced in the paper: *“Attention Is All You Need”*

---

## 🔹 Key Idea (Simple)

👉 Instead of reading text word-by-word, it:

* Looks at **all words at once**
* Understands **relationships using attention**

---

## 🔹 Example

Sentence:

```text
"The animal didn’t cross the street because it was tired"
```

👉 Transformer understands:

* “it” → refers to **animal**

This is done using **self-attention**

---

# 🔹 Transformer Architecture

### 1. Encoder (understanding)

* Used in models like BERT

### 2. Decoder (generation)

* Used in models like GPT

---

# 🔹 Important Concepts (MUST KNOW)

### ✅ 1. Self-Attention

👉 Each word looks at other words

```text
"bank" → river bank OR money bank (depends on context)
```

---

### ✅ 2. Multi-Head Attention

👉 Multiple attention layers → capture different meanings

---

### ✅ 3. Positional Encoding

👉 Adds order (since model reads all words together)

---

### ✅ 4. Feed Forward Network

👉 Processes output after attention

---

### ✅ 5. Layer Normalization + Residuals

👉 Helps training deep networks

# 🔥 Practical Coding Question

👉 Implement simple attention (interview favorite)

```python
import numpy as np

def attention(Q, K, V):
    scores = np.dot(Q, K.T)
    weights = softmax(scores)
    return np.dot(weights, V)
```
