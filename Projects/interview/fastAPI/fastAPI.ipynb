# ✅ Top FastAPI Interview Questions (with short answers)

---

## 1️⃣ What is FastAPI and why is it popular?

**Answer**

FastAPI is a **high-performance Python web framework** for building APIs using:

* type hints
* async support
* automatic OpenAPI docs

**Why popular**

* very fast
* very little boilerplate
* automatic validation
* auto Swagger UI

---

## 2️⃣ How is FastAPI different from Flask?

**Answer**

| Feature       | FastAPI              | Flask           |
| ------------- | -------------------- | --------------- |
| Async support | Native               | Not native      |
| Validation    | Automatic (Pydantic) | Manual          |
| Docs          | Auto OpenAPI         | Need extensions |
| Type hints    | Core design          | Optional        |

👉 Interview line:

> FastAPI is built around type hints and async-first design.

---

## 3️⃣ What is ASGI and why does FastAPI use it?

**Answer**

**ASGI = Asynchronous Server Gateway Interface**

It allows:

* async request handling
* websockets
* long-lived connections

FastAPI runs on ASGI servers like `uvicorn`.

---

## 4️⃣ Difference between `async def` and `def` in FastAPI?

**Answer**

Use:

* `async def` → when you call async I/O (DB, HTTP, etc.)
* `def` → when your code is purely CPU / sync

**Example**

```python
@app.get("/items")
async def get_items():
    return {"msg": "ok"}
```

👉 Important interview point:
FastAPI allows **mixing sync and async endpoints safely**.

---

## 5️⃣ What is Pydantic and why is it important?

**Answer**

Pydantic is used for:

* request body validation
* response serialization
* type checking

**Example**

```python
class User(BaseModel):
    name: str
    age: int
```

FastAPI uses it internally to:

* validate JSON
* generate OpenAPI schema

## 6️⃣ What is dependency injection in FastAPI?

**Answer**

FastAPI has a built-in dependency system using `Depends()`.

It is used for:

* auth
* DB session
* common logic

**Example**

```python
def get_db():
    return db

@app.get("/users")
def users(db = Depends(get_db)):
    ...
```

## 9️⃣ What is 422 Unprocessable Entity in FastAPI?

**Answer**

FastAPI returns **422** when:

## 1️⃣1️⃣ How do you handle background tasks?

**Answer**

Using `BackgroundTasks`.

```python
from fastapi import BackgroundTasks

def send_email():
    ...

@app.post("/order")
def order(bg: BackgroundTasks):
    bg.add_task(send_email)
```

Used for:

* notifications
* logging
* async side jobs (lightweight)

---

## 1️⃣2️⃣ How do you handle middleware?

**Answer**

Middleware is used for:

* logging
* auth headers
* tracing
* CORS

```python
@app.middleware("http")
async def log_requests(request, call_next):
    response = await call_next(request)
    return response
```

---

## 1️⃣3️⃣ How do you handle CORS in FastAPI?

**Answer**

Using `CORSMiddleware`.

```python
app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_methods=["*"],
    allow_headers=["*"]
)
```

## 1️⃣4️⃣ How do you return custom response models?

**Answer**

Using `response_model`.

```python
@app.get("/user", response_model=UserOut)
def get_user():
    return user
```

Benefits:

* hides internal fields
* enforces output contract

## 1️⃣5️⃣ How do you handle exceptions globally?

**Answer**

Using exception handlers.

```python
@app.exception_handler(MyException)
async def handler(request, exc):
    return JSONResponse(...)
```

## 2️⃣0️⃣ What is the biggest mistake candidates make in FastAPI?

**Answer**

They write:

```python
async def endpoint():
    blocking_db_call()
```

👉 Blocking I/O inside async endpoints
→ kills performance.
