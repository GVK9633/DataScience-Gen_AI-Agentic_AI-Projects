# Agentic AI & Production Concerns — Interview Q&A

## Q1. What is the difference between Agentic AI and Traditional Generative AI?

### Answer

**Traditional Generative AI** mainly generates a response based on the user's prompt.

**Agentic AI** can **reason, plan, use tools, take actions, observe results, and continue until the task is completed**.

| Traditional GenAI     | Agentic AI                           |
| --------------------- | ------------------------------------ |
| Prompt → Response     | Goal → Plan → Act → Observe → Result |
| Usually one-shot      | Multi-step                           |
| Mainly generates text | Can perform actions                  |
| Limited tool usage    | Uses tools/APIs                      |
| Less autonomous       | More autonomous                      |

### Example

**Traditional GenAI:**

User:

> "What is my order status?"

LLM:

> "I don't know your order status."

**Agentic AI:**

```text
User
 ↓
Agent
 ↓
Call Order API
 ↓
Get order status
 ↓
Reason over result
 ↓
Respond to user
```

Response:

> "Your order #1234 was shipped yesterday and is expected tomorrow."

### Interview answer

> "Traditional GenAI primarily generates content from a prompt, whereas Agentic AI can reason, plan, use tools, observe results, and take multiple steps to achieve a goal."

# Q2. What are Andrew Ng's 4 Agentic Design Patterns?

Andrew Ng described four important patterns for building agentic systems:

1. **Reflection**
2. **Tool Use**
3. **Planning**
4. **Multi-Agent Collaboration**

### 1. Reflection

Agent generates an answer → evaluates it → improves it.

```text
Generate
   ↓
Review
   ↓
Improve
   ↓
Final answer
```

### 2. Tool Use

Agent calls external tools such as:

```text
Database
API
Calculator
Search
MCP tool
```

### 3. Planning

Agent breaks a complex task into smaller tasks.

Example:

> "Plan my trip to Hyderabad."

Agent:

```text
1. Find hotels
2. Find transportation
3. Find places to visit
4. Check availability
5. Create itinerary
```

### 4. Multi-Agent Collaboration

Different agents specialize in different tasks.

```text
Supervisor Agent
      |
 ┌────┼─────┐
 ↓    ↓     ↓
Policy Order Fraud
Agent  Agent Agent
```

### Interview answer

> "The four commonly discussed agentic patterns from Andrew Ng are reflection, tool use, planning, and multi-agent collaboration. They make LLM systems more capable of iterative reasoning and completing complex tasks."

# Q3. What is the ReAct framework?

**ReAct = Reasoning + Acting**

It combines reasoning with tool/action execution.

The agent follows:

```text
Reason
  ↓
Action
  ↓
Observation
  ↓
Reason
  ↓
Action
  ↓
Final Answer
```

### Example

User:

> "Check whether my order is eligible for return."

Agent:

```text
Reason:
I need the order details.

Action:
Call Order API

Observation:
Order delivered 5 days ago.

Reason:
I need the return policy.

Action:
Search Policy Knowledge Base

Observation:
Returns allowed within 7 days.

Final:
The order is eligible for return.
```

### Interview answer

> "ReAct stands for Reasoning and Acting. The agent reasons about what to do, executes an action or tool, observes the result, and uses that observation for the next step."

# Q4. What is the difference between System Prompt and User Prompt?

| System Prompt                   | User Prompt                 |
| ------------------------------- | --------------------------- |
| Defines agent behavior          | Contains user's request     |
| Higher-level instructions       | Specific task               |
| Defines role/rules              | Defines current requirement |
| Usually controlled by developer | Comes from user             |

### Example

**System prompt:**

```text
You are a customer support agent.
Never approve refunds above ₹10,000.
Always verify the order before processing a refund.
```

**User prompt:**

```text
Please refund my order #1234.
```

The agent must follow the system rules while processing the user request.

### Interview answer

> "The system prompt defines the agent's role, behavior, constraints and policies, while the user prompt provides the specific task or request."

# Q5. How do you implement a Reflection Loop?

Reflection means allowing the agent to **review and improve its own output**.

### Basic architecture

```text
          ┌─────────────┐
          │ Generate    │
          │ Answer      │
          └──────┬──────┘
                 ↓
          ┌─────────────┐
          │ Critic /    │
          │ Evaluator   │
          └──────┬──────┘
                 ↓
            Good enough?
             /       \
           Yes        No
            ↓          ↓
          Final     Improve
                       ↓
                    Generate
```

### Example

Suppose an agent creates:

> "Your refund will definitely arrive tomorrow."

Reflection checks:

> "Is this guaranteed by the payment system?"

If not, it changes the answer to:

> "Your refund has been initiated and is expected to be processed within 3–5 business days."

### Interview answer

> "A reflection loop generates an output, evaluates it against defined criteria, and if it fails, sends feedback back to the agent for improvement. I would also set a maximum iteration count to avoid infinite loops."

# Q6. How does Tool Use / Function Calling work?

The LLM itself doesn't directly execute your database or API call.

Instead:

```text
User
 ↓
LLM
 ↓
Decides tool is required
 ↓
Tool call + parameters
 ↓
Application executes tool
 ↓
Tool result
 ↓
LLM
 ↓
Final response
```

### Example

Tool:

```python
get_order_status(order_id)
```

User:

> "Where is order 1234?"

LLM decides:

```text
Tool: get_order_status
Arguments:
order_id = "1234"
```

Application executes:

```text
get_order_status("1234")
```

Result:

```json
{
  "status": "SHIPPED",
  "delivery_date": "2026-10-02"
}
```

LLM then responds:

> "Your order has been shipped and is expected on October 2."

### Interview answer

> "Function calling allows an LLM to request a structured tool invocation. The application executes the actual function or API, returns the result to the LLM, and the LLM uses that result to generate the final response."

# Q7. How do you design Memory in an Agentic AI system?

Agent memory allows the system to retain useful information across interactions.

There are generally two important types:

### Short-term memory

Information within the current conversation.

```text
User: My order number is 1234.

User: Check its status.

Agent remembers:
order = 1234
```

### Long-term memory

Information stored for future conversations.

Example:

```text
User prefers email notifications.
```

This can be stored in a database/vector store depending on the use case.

### Architecture

```text
             Agent
               |
       ┌───────┴────────┐
       ↓                ↓
Short-term         Long-term
Memory             Memory
       ↓                ↓
Conversation       DB / Vector DB
State
```

### Interview answer

> "I would use short-term memory for conversation state and long-term memory for persistent user information. I would carefully control what is stored, apply access controls, and avoid storing unnecessary sensitive information."

# Q8. What is Multi-Agent Orchestration and Routing?

Multi-agent architecture uses multiple specialized agents instead of one agent doing everything.

### Example — Return Policy Project

```text
                Supervisor
                    |
       ┌────────────┼────────────┐
       ↓            ↓            ↓
 Policy Agent   Order Agent   Fraud Agent
       ↓            ↓            ↓
 Policy KB      Order API     ML Model
```

User:

> "Can I return this product?"

Supervisor determines:

```text
Policy Agent → Check return policy
Order Agent  → Check order status
Fraud Agent  → Check risk if required
```

### Routing

Routing decides **which agent should handle a particular task**.

Example:

```text
"Where is my order?"
       ↓
Order Agent

"Can I return it?"
       ↓
Policy Agent

"Why was my refund rejected?"
       ↓
Refund Agent
```

### Interview answer

> "Multi-agent orchestration coordinates multiple specialized agents, while routing determines which agent should handle each task. A supervisor or router can maintain the overall state and control the execution flow."

```text
Prompt → LLM → Parser
```

```text
Agent
 ↓
Tool
 ↓
Agent
 ↓
Decision
 ↓
Another Agent
```

```text
Agent A ↔ Agent B ↔ Agent C
```

Agents communicate with each other to complete a task.

# Q10. What are Evals in Agentic AI?

**Evals = Evaluations**

Evals measure whether the AI system is actually performing correctly.

For an agent, we can evaluate:

* Answer correctness
* Relevance
* Tool selection
* Tool arguments
* Retrieval quality
* Hallucination
* Task completion
* Safety

### Example

Suppose the expected answer is:

> "Return allowed within 7 days."

Agent answers:

> "Return allowed within 30 days."

Evaluation:

```text
Expected: 7 days
Actual: 30 days
Result: FAIL
```

### LLM-as-a-Judge

Another LLM can evaluate the response against criteria.

```text
Agent Response
      ↓
Evaluator LLM
      ↓
Score / Feedback
```

Important: LLM-as-a-Judge itself can be imperfect, so use deterministic checks and human review where appropriate.

### Interview answer

> "Evals are systematic tests used to measure agent quality, such as correctness, relevance, tool usage and task completion. LLM-as-a-Judge can help evaluate subjective qualities, but I would combine it with deterministic tests and human evaluation."

# Q11. What are Tracing, Spans and Observability?

These are extremely important in production Agentic AI.

### Trace

A complete request journey.

```text
User Request
    |
    ├── Agent
    ├── Retrieval
    ├── Tool Call
    ├── Database
    └── LLM
```

### Span

An individual operation inside the trace.

Example:

```text
Trace: Return Request

Span 1 → Intent Agent
Span 2 → Policy Retrieval
Span 3 → Order API
Span 4 → Final LLM
```

### Observability

Helps answer:

> **What happened, where did it fail, and why?**

Monitor:

* Latency
* Token usage
* Cost
* Errors
* Tool failures
* Retrieval quality
* Agent steps
* Model responses

### Interview answer

> "Tracing records the complete execution path of an agent request, while spans represent individual operations within that trace. Observability helps us diagnose latency, failures, incorrect tool calls and cost issues in production."

# Q12. What is the difference between Text Hallucination and Functional Hallucination?

This is a **very important production interview question**.

### Text hallucination

The model generates incorrect information.

Example:

> "Your refund was processed."

But actually it wasn't.

### Functional hallucination

The agent performs an incorrect **action**.

Example:

```text
User:
Cancel my order.

Agent:
Calls cancel_order(order_id=5678)
```

But the user actually meant order `5679`.

This can cause real business impact.

### Comparison

| Text Hallucination            | Functional Hallucination |
| ----------------------------- | ------------------------ |
| Wrong information             | Wrong action             |
| Mainly affects answer quality | Can affect real systems  |
| Example: wrong policy         | Example: wrong refund    |
| Output problem                | Action/execution problem |

### Prevention

Use:

```text
Validation
Authorization
Tool schemas
Confirmation
Guardrails
Human approval
```

### Interview answer

> "Text hallucination is when an agent generates factually incorrect information, while functional hallucination is when it performs or attempts an incorrect action through a tool. Functional hallucinations require stronger validation and authorization because they can change real-world data."

# Q13. How do you handle Non-Deterministic Behavior in Agentic AI?

LLMs are probabilistic, so the same input may sometimes produce different outputs.

### Problems

```text
Request
  ↓
Agent
 ↙   ↘
Path A Path B
```

Different execution paths may cause different results.

### Solutions

**1. Lower temperature**

For deterministic tasks:

```text
temperature ≈ 0
```

**2. Structured output**

Use JSON/schema validation.

**3. Deterministic tools**

Business rules should be implemented in code rather than relying only on LLM reasoning.

**4. Guardrails**

Validate tool calls before execution.

**5. Evals**

Run the same test cases repeatedly.

**6. Retry/fallback**

If an LLM call fails, use controlled retries or fallback models.

### Interview answer

> "I handle non-determinism using lower temperature where appropriate, structured outputs, deterministic business logic, tool validation, guardrails, and continuous evaluation. Critical decisions should not depend solely on probabilistic LLM output."

# Q14. What are Infinite Loops and Agent Sprawl?

## Infinite Loop

An agent repeatedly calls itself or tools without reaching completion.

Example:

```text
Agent
 ↓
Tool
 ↓
Agent
 ↓
Tool
 ↓
Agent
 ↓
Tool
 ↓
...
```

### Prevention

Set:

```text
max_iterations = 5
```

Also use:

* Timeout
* Maximum tool calls
* Budget/token limits
* Loop detection
* Termination conditions

---

## Agent Sprawl

Agent sprawl means creating too many agents when they aren't actually necessary.

Bad architecture:

```text
20 agents
 ↓
Complex routing
 ↓
Hard debugging
 ↓
Higher cost
 ↓
Higher latency
```

Use specialized agents only when there is a clear responsibility boundary.

### Interview answer

> "Infinite loops occur when an agent repeatedly executes actions without reaching a termination condition. I prevent them using iteration limits, timeouts, tool-call limits and explicit termination rules. Agent sprawl means unnecessarily creating too many agents, which increases complexity, cost and debugging difficulty."

# Q15. How do you handle Security, Deployment and Privilege Escalation in Agentic AI?

This is one of the **most important production questions**.

An agent can potentially call powerful tools, so we should follow **least privilege**.

### Example

Suppose an agent has these tools:

```text
get_order()
cancel_order()
refund_order()
delete_customer()
```

Don't give every agent permission to use everything.

Instead:

```text
Order Agent
   ↓
get_order()

Refund Agent
   ↓
get_order()
refund_order()

Admin Agent
   ↓
Restricted admin tools
```

### Security controls

#### 1. Authentication

Verify who is making the request.

#### 2. Authorization

Verify what the user/agent is allowed to do.

```text
User → Authentication
          ↓
      Authorization
          ↓
       Tool Call
```

#### 3. Least privilege

Give an agent only the permissions it needs.

#### 4. Tool validation

Validate:

```text
tool name
parameters
user permissions
business rules
```

#### 5. Human-in-the-loop

For high-risk actions:

```text
Agent
 ↓
Request approval
 ↓
Human
 ↓
Execute
```

Example:

> Refund ₹50,000 → Human approval required.

#### 6. Secrets management

Don't put API keys directly inside prompts or code.

Use secure secret-management mechanisms.

#### 7. Deployment

Typical production architecture:

```text
User
 ↓
API Gateway
 ↓
Authentication
 ↓
Agent Service
 ↓
LLM
 ↓
Tools / APIs / DB
```

Use:

* Containers
* CI/CD
* Monitoring
* Logging
* Rate limiting
* Autoscaling
* Network controls

### Interview answer

> "For production agents, I follow least privilege, strong authentication and authorization, tool-level validation, secure secret management, rate limiting, audit logging and human approval for high-risk operations. The agent should never be able to escalate its own permissions."
