## Q1. Why is the Harmonic Mean used for F1 Score?

F1 score combines:

* **Precision** → How many predicted positives were actually positive?
* **Recall** → How many actual positives did we find?

Formula:

$$
F1 = \frac{2 \times Precision \times Recall}{Precision + Recall}
$$

### Why harmonic mean?

The harmonic mean gives a **low score when either Precision or Recall is low**.

Example:

```text
Precision = 0.9
Recall    = 0.1
```

Arithmetic mean:

```text
(0.9 + 0.1) / 2 = 0.5
```

That looks reasonable, but the model has very poor recall.

F1 will be much lower:

```text
F1 ≈ 0.18
```

So F1 forces us to have a reasonable balance between Precision and Recall.

### Interview answer

> "F1 uses the harmonic mean because it penalizes imbalance between precision and recall and becomes low when either one is low."

# Q2. What is the purpose of an Activation Function?

An activation function introduces **non-linearity** into a neural network.

Without activation functions:

```text
Layer 1
   ↓
Linear
   ↓
Layer 2
   ↓
Linear
```

Multiple linear transformations can still be represented as essentially one linear transformation.

With activation:

```text
Input
 ↓
Linear
 ↓
ReLU
 ↓
Linear
 ↓
ReLU
 ↓
Output
```

The network can learn complex non-linear patterns.

### Common functions

| Function | Common use                   |
| -------- | ---------------------------- |
| ReLU     | Hidden layers                |
| Sigmoid  | Binary classification output |
| Softmax  | Multi-class classification   |
| Tanh     | Some neural networks         |

### Interview answer

> "Activation functions introduce non-linearity, allowing neural networks to learn complex relationships rather than only linear relationships."

Scaling is often important because it is an optimization-based linear model.

Example:

```text
Age       = 30
Salary    = 1,000,000
```

Very different scales can affect optimization and regularization.

So we commonly use:

```text
StandardScaler
```

before Logistic Regression.

Random Forest is based on **decision trees**.

Trees split based on conditions such as:

```text
salary > 500000
```

Scaling doesn't change the ordering of values, so tree-based models generally don't require feature scaling.

### Example

```text
Age:
20, 30, 40

Salary:
20000, 50000, 100000
```

Random Forest can work directly with these values.

### Interview answer

> "Random Forest generally does not require feature scaling because tree splits are based on feature thresholds, while Logistic Regression can benefit from scaling because optimization and regularization are sensitive to feature magnitude."

# Q4. Precision vs Recall in Medical Diagnosis

Suppose we are detecting a disease.

### Precision

> Of the people predicted as having the disease, how many actually have it?

High precision means:

```text
Few false positives
```

### Recall

> Of all people who actually have the disease, how many did we detect?

High recall means:

```text
Few false negatives
```

### Medical diagnosis

Missing a real disease can be serious.

Therefore, depending on the disease and clinical context, **high recall/sensitivity may be prioritized during screening**.

Example:

```text
100 patients have disease
Model detects 98
```

Recall:

```text
98/100 = 98%
```

### Interview answer

> "For medical screening, recall is often important because missing a patient who actually has the disease can be serious. Precision is also important because excessive false positives can lead to unnecessary follow-up."

# Q5. What is the impact of Skewness on Model Performance?

Skewness means the distribution is **not symmetric**.

Example:

```text
Most salaries:
₹3L – ₹10L

Few salaries:
₹50L – ₹1Cr
```

This creates right skew.

### Why can it matter?

Some algorithms and statistical techniques can be affected by:

* Outliers
* Large feature ranges
* Non-normal distributions

### Common solution

Apply transformation such as:

```python
import numpy as np

df["salary"] = np.log1p(df["salary"])
```

This can reduce right skew.

### Important

Tree-based models are generally less sensitive to monotonic transformations than many linear/statistical models, but skewed features can still matter depending on the pipeline and objective.

### Interview answer

> "Highly skewed features can make some models sensitive to outliers and affect optimization or assumptions, so transformations such as log transformation may be useful when appropriate."

# Q6. Lasso vs Ridge

### Lasso = L1

Penalty:

$$
\lambda\sum |w_i|
$$

Can make some coefficients:

```text
w = 0
```

Therefore it can perform **feature selection**.

### Ridge = L2

Penalty:

$$
\lambda\sum w_i^2
$$

Usually shrinks coefficients toward zero without making many exactly zero.

### Example

Suppose:

```text
Feature A → 0.01
Feature B → 5.2
Feature C → 0.02
```

Lasso may produce:

```text
A → 0
B → 4.8
C → 0
```

### Interview answer

> "Lasso uses L1 regularization and can produce sparse models by driving some coefficients to zero, while Ridge uses L2 regularization and mainly shrinks coefficients."

# Q7. Bayesian Optimization vs Grid Search

Both are methods for **hyperparameter tuning**.

### Grid Search

Try every combination.

Example:

```text
learning_rate = [0.01, 0.1, 0.2]
depth          = [3, 5, 7]
```

Total:

```text
3 × 3 = 9 combinations
```

Grid search tests all 9.

### Bayesian Optimization

Uses results from previous trials to decide **which configuration to try next**.

Conceptually:

```text
Trial 1
   ↓
Learn from result
   ↓
Choose promising parameters
   ↓
Trial 2
   ↓
Learn again
```

Therefore it can be more efficient when evaluations are expensive.

### Interview answer

> "Grid Search exhaustively evaluates predefined combinations, while Bayesian Optimization uses previous evaluation results to intelligently select promising hyperparameters."

# Q8. What is Out-of-Bag Error in Random Forest?

Random Forest uses **bootstrap sampling**.

Suppose training data has:

```text
100 samples
```

Each tree randomly samples training examples with replacement.

Some samples are not selected for a particular tree.

These are called:

> **Out-of-Bag (OOB) samples**

Those OOB samples can be used to estimate the model's error.

### Benefit

You can get an internal validation estimate **without creating a separate validation set**, under appropriate settings.

Example:

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    oob_score=True,
    random_state=42
)

model.fit(X_train, y_train)

print(model.oob_score_)
```

### Interview answer

> "OOB error estimates Random Forest generalization performance using samples that were not included in the bootstrap sample for each tree."

# Q9. What is Bagging? What is the No Free Lunch Theorem?

## Bagging

Bagging = **Bootstrap Aggregating**

It trains multiple models on different bootstrap samples and combines their predictions.

```text
Dataset
 ↓
Bootstrap samples
 ↓
Model 1
Model 2
Model 3
...
 ↓
Combine predictions
```

Random Forest is a famous example of bagging with decision trees plus feature randomness.

### Why?

It can reduce **variance** and improve stability.

---

## No Free Lunch Theorem

There is no single machine-learning algorithm that is guaranteed to perform best on **every possible problem**.

Example:

```text
Problem A → Tree model may work well
Problem B → Linear model may work well
Problem C → Neural network may work well
```

### Interview answer

> "Bagging combines multiple models trained on bootstrap samples to reduce variance, while the No Free Lunch theorem states that no single learning algorithm is universally optimal for every problem."

# Q10. Hard Voting vs Soft Voting

Used in ensemble classification.

## Hard Voting

Each model gives a class.

```text
Model 1 → Cat
Model 2 → Dog
Model 3 → Dog
```

Final:

```text
Dog
```

Majority wins.

---

## Soft Voting

Models provide probabilities.

Example:

```text
Model 1 → Dog 0.7
Model 2 → Dog 0.8
Model 3 → Dog 0.6
```

Average probabilities and choose the class with highest combined probability.

### Interview answer

> "Hard voting selects the class with the most model votes, while soft voting combines predicted probabilities and chooses the class with the highest aggregated probability."

# Q11. Why are Weak Learners used in Boosting?

A weak learner is a model that performs **slightly better than random guessing**.

Common example:

```text
Decision stump
```

A decision stump is a tree with very small depth, often depth 1.

Boosting works sequentially:

```text
Weak Model 1
     ↓
Find errors
     ↓
Weak Model 2 focuses on errors
     ↓
Weak Model 3
     ↓
Combine
```

Many weak learners together can create a strong model.

### Example

AdaBoost and Gradient Boosting use this idea.

### Interview answer

> "Boosting uses simple weak learners sequentially, with later learners focusing on errors or residuals from earlier learners, so their combination forms a stronger model."

```text
20 useful features
```

You keep the original features.

Examples:

* Correlation-based selection

* Feature importance

---

Create **new features** from existing features.

Example:

```text
100 original features
        ↓
       PCA
        ↓
20 principal components
```

Suppose you have:

```text
1000 records
```

Example:

```text
5-Fold CV
```

```text
Fold 1 → Validation
Fold 2 → Validation
Fold 3 → Validation
Fold 4 → Validation
Fold 5 → Validation
```

Each fold becomes validation once.

Then calculate:

```text
Average score
```

Example:

```text
Accuracy:
Fold 1 = 90%
Fold 2 = 88%
Fold 3 = 91%
Fold 4 = 89%
Fold 5 = 92%

Average = 90%
```

It provides a more robust estimate of performance than relying on a single split.

# Q15. What is Pruning?

Pruning is mainly associated with **decision trees**.

It removes unnecessary branches to reduce overfitting.

### Example

Without pruning:

```text
             Age
           /     \
        Salary   Salary
        /   \
       ...   ...
```

The tree may become extremely complex and memorize training data.

After pruning:

```text
          Age
        /     \
      Yes      No
```

Simpler tree → potentially better generalization.

### Two approaches

**Pre-pruning**

Stop tree growth early.

Examples:

```text
max_depth
min_samples_split
min_samples_leaf
```

**Post-pruning**

Grow a tree and then remove unnecessary branches.

### Interview answer

> "Pruning reduces unnecessary decision-tree branches to control complexity and overfitting; it can be done during tree growth or after the tree is grown."

# Q17. False Positive vs False Negative

This is extremely important.

### False Positive

Model predicts positive but actual value is negative.

```text
Actual:    No fraud
Predicted: Fraud
```

**False alarm.**

### False Negative

Model predicts negative but actual value is positive.

```text
Actual:    Fraud
Predicted: No fraud
```

**Missed detection.**

### Easy memory

```text
FP → False alarm
FN → Missed positive
```

### Example: Spam detection

FP:

```text
Real email
   ↓
Model says Spam
```

FN:

```text
Spam email
   ↓
Model says Not Spam
```

### Interview answer

> "A false positive is an incorrect positive prediction, while a false negative is an incorrect negative prediction."

# Q20. How do you interpret an ROC-AUC Curve?

Very common interview question.

## ROC

ROC = **Receiver Operating Characteristic**

It plots:

```text
Y-axis → True Positive Rate
X-axis → False Positive Rate
```

### True Positive Rate

Also called **Recall/Sensitivity**.

$$
TPR = \frac{TP}{TP+FN}
$$

### False Positive Rate

$$
FPR = \frac{FP}{FP+TN}
$$

---

## AUC

AUC = **Area Under the Curve**

It measures how well the model separates positive and negative classes across thresholds.

### Interpretation

```text
AUC = 1.0
→ Perfect discrimination

AUC ≈ 0.5
→ Similar to random ranking

AUC < 0.5
→ Worse than random ranking
```

### Example

Model A:

```text
AUC = 0.92
```

Model B:

```text
AUC = 0.70
```

AUC indicates that Model A has stronger ranking/discrimination ability on the evaluated data.

### Important interview point

AUC is **not the same as accuracy**.

A model can have high AUC but still require an appropriate threshold for the application's precision/recall requirements.

### Interview answer

> "ROC-AUC measures a classifier's ability to distinguish between positive and negative classes across different thresholds. ROC plots TPR against FPR, and AUC summarizes the ranking/discrimination ability."
