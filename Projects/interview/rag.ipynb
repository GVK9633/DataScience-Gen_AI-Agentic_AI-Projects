# 🔥 MOST ASKED RAG INTERVIEW Q&A (3 Years Experience)

---

## 1️⃣ What is RAG and why is it needed?

**Answer:**
RAG (Retrieval-Augmented Generation) combines **information retrieval** with **LLMs** to generate answers grounded in **external knowledge**, reducing hallucinations and overcoming knowledge cutoff issues.

**Why needed:**

* LLMs don’t know private data
* Knowledge becomes outdated
* High hallucination risk

---

## 2️⃣ Explain RAG architecture end-to-end

**Answer:**
RAG has two main phases:

**Offline**

* Load data
* Chunk documents
* Generate embeddings
* Store in vector database

**Online**

* Embed user query
* Retrieve relevant chunks
* Add them to prompt
* LLM generates answer

---

## 3️⃣ How does RAG reduce hallucinations?

**Answer:**
RAG grounds LLM responses in **retrieved documents**. The model is instructed to answer **only using the provided context**, which significantly reduces hallucinated outputs.

---

## 4️⃣ What is a vector database and why is it used?

**Answer:**
A vector database stores **embeddings** and supports **similarity search** to quickly retrieve relevant documents.

**Examples:** FAISS, Pinecone, Chroma, Weaviate.

---

## 5️⃣ What are embeddings in RAG?

**Answer:**
Embeddings are **numerical representations of text** that capture semantic meaning and are used to find similar content during retrieval.

---

## 6️⃣ What chunk size is best for RAG?

**Answer:**
There is no fixed size, but commonly:

* **300–800 tokens**
* With **overlap of 50–150 tokens**

Too small → context loss
Too large → poor retrieval

---

## 7️⃣ Difference between RAG and fine-tuning?

| Feature          | RAG     | Fine-Tuning |
| ---------------- | ------- | ----------- |
| Knowledge update | Easy    | Hard        |
| Cost             | Lower   | Higher      |
| Private data     | Yes     | Limited     |
| Hallucinations   | Reduced | Possible    |

**Best practice:** Use **RAG first**, fine-tune only if needed.

---

## 8️⃣ What is similarity search?

**Answer:**
It finds documents whose embeddings are **closest** to the query embedding using distance metrics like:

* Cosine similarity
* Euclidean distance

---

## 9️⃣ What is Top-K in RAG?

**Answer:**
Top-K is the **number of most relevant chunks** retrieved from the vector database.

Typical values: **3–10**

---

## 🔟 What is hybrid search?

**Answer:**
Hybrid search combines:

* **Vector search** (semantic)
* **Keyword search** (BM25)

It improves retrieval accuracy for structured or keyword-heavy data.

---

## 1️⃣1️⃣ What is reranking in RAG?

**Answer:**
Reranking reorders retrieved documents using a **more accurate model** after initial retrieval to improve relevance before passing context to the LLM.

---

## 1️⃣2️⃣ How do you improve RAG answer quality?

**Answer:**

* Better chunking strategy
* Metadata filtering
* Hybrid search
* Reranking
* Prompt engineering
* Query rewriting

---

## 1️⃣3️⃣ What happens if the answer is not in retrieved context?

**Answer:**
The prompt should instruct the LLM to say:

> *“I don’t know based on the provided context.”*

This prevents hallucinations.

---

## 1️⃣4️⃣ What is context window limitation in RAG?

**Answer:**
LLMs have a **maximum token limit**. Too many retrieved documents can exceed this limit, causing truncation or loss of important context.

---

## 1️⃣5️⃣ How do you handle large documents in RAG?

**Answer:**

* Chunking
* Hierarchical retrieval
* Context compression
* Summarization before retrieval

---

## 1️⃣6️⃣ What is Agentic RAG?

**Answer:**
Agentic RAG allows the LLM to **decide when and how to retrieve information**, often making multiple retrieval calls using tools and memory.

---

## 1️⃣7️⃣ What are common RAG challenges?

**RAG Fusion (Retrieval-Augmented Generation Fusion)** is an advanced technique used in **RAG systems** to improve the **quality, relevance, and completeness of retrieved information** before generating an answer with an LLM.

---

## 🔍 Why RAG Fusion is Needed

In standard RAG:

* You send **one query**
* Retrieve top-k documents
* Pass them to the LLM

⚠️ Problem:

* A single query may miss important context
* Retrieved results may be incomplete or biased
* Recall can be low for complex questions

---

## 🚀 What is RAG Fusion?

**RAG Fusion = Multi-query retrieval + intelligent result fusion**

Instead of relying on **one query**, RAG Fusion:

1. **Generates multiple query variants**
2. **Retrieves documents for each query**
3. **Fuses (merges + re-ranks) all results**
4. Sends the **best combined context** to the LLM

---

## 🧠 How RAG Fusion Works (Step-by-Step)

### 1️⃣ Query Expansion

The original user question is rewritten into multiple semantically different queries.

**Example**

```
User query:
"What is the return policy for electronic products?"

Generated queries:
- "electronics product return policy"
- "refund rules for electronic items"
- "return and exchange conditions for electronics"
- "electronics refund eligibility"
```

---

### 2️⃣ Parallel Retrieval

Each query searches the vector database independently.

```
Query 1 → Top 5 docs
Query 2 → Top 5 docs
Query 3 → Top 5 docs
```

Total retrieved docs = many overlapping + unique results

---

### 3️⃣ Fusion (Key Step)

Results are **merged and ranked** using fusion algorithms.

Most common method:

### 🔹 Reciprocal Rank Fusion (RRF)

```
Score(doc) = Σ 1 / (rank + k)
```

Benefits:

* Rewards documents that appear in multiple result sets
* Reduces noise from a single bad query
* Improves recall significantly

---

### 4️⃣ Context Selection

* Remove duplicates
* Pick top-N highest scoring documents
* Pass to the LLM as context

---

### 5️⃣ Answer Generation

LLM generates a grounded response using fused knowledge.

---

## 🧩 RAG vs RAG Fusion

| Feature          | RAG    | RAG Fusion |
| ---------------- | ------ | ---------- |
| Query count      | Single | Multiple   |
| Retrieval recall | Medium | High       |

# 1️⃣ RAG Fusion with **LangChain** (Python)

LangChain has **native RAG Fusion support** using **MultiQueryRetriever + Reciprocal Rank Fusion (RRF)**.

---

## 🔹 Architecture (LangChain RAG Fusion)

```
User Query
   ↓
LLM → Generate Multiple Queries
   ↓
Vector DB (Parallel Retrieval)
   ↓
Reciprocal Rank Fusion (RRF)
   ↓
Top Documents
   ↓
LLM Answer
```

---

## 🔹 Python Code (LangChain RAG Fusion)

### ✅ Install

```bash
pip install langchain langchain-community langchain-openai faiss-cpu
```

---

### ✅ Code Example

```python
from langchain_openai import ChatOpenAI
from langchain.retrievers.multi_query import MultiQueryRetriever
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
from langchain.chains import RetrievalQA
from langchain.docstore.document import Document

# 1. Sample documents
docs = [
    Document(page_content="Electronics can be returned within 30 days."),
    Document(page_content="Refunds are issued within 5 business days."),
    Document(page_content="Opened electronic items are not returnable."),
]

# 2. Vector store
embeddings = OpenAIEmbeddings()
vectorstore = FAISS.from_documents(docs, embeddings)

# 3. LLM
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# 4. RAG Fusion Retriever
retriever = MultiQueryRetriever.from_llm(
    retriever=vectorstore.as_retriever(search_kwargs={"k": 5}),
    llm=llm
)

# 5. QA Chain
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    retriever=retriever,
    chain_type="stuff"
)

# 6. Query
query = "What is the return policy for electronic products?"
response = qa_chain.run(query)

print(response)
```

---

## 🔑 Why This Is RAG Fusion

✔ Multiple query generation
✔ Parallel retrieval
✔ Fused results (RRF internally)
✔ Improved recall & accuracy

# 2️⃣ RAG Fusion with **LlamaIndex** (Python)

LlamaIndex offers **Query Fusion Retriever**, designed specifically for this purpose.

---

## 🔹 Architecture (LlamaIndex)

```
User Query
   ↓
Query Fusion Retriever
   ↓
Vector DB (Multiple Queries)
   ↓
RRF Fusion
   ↓
Context
   ↓
LLM
```

---

## 🔹 Python Code (LlamaIndex RAG Fusion)

### ✅ Install

```bash
pip install llama-index llama-index-llms-openai llama-index-embeddings-openai
```

---

### ✅ Code Example

```python
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
from llama_index.core.retrievers import QueryFusionRetriever
from llama_index.llms.openai import OpenAI
from llama_index.embeddings.openai import OpenAIEmbedding

# 1. Load documents
documents = SimpleDirectoryReader("./data").load_data()

# 2. Create index
index = VectorStoreIndex.from_documents(
    documents,
    embed_model=OpenAIEmbedding()
)

# 3. Base retriever
base_retriever = index.as_retriever(similarity_top_k=5)

# 4. RAG Fusion Retriever
fusion_retriever = QueryFusionRetriever(
    retrievers=[base_retriever],
    llm=OpenAI(model="gpt-4o-mini"),
    similarity_top_k=5,
    num_queries=4,        # Number of generated queries
    mode="reciprocal_rerank"
)

# 5. Query engine
query_engine = index.as_query_engine(
    retriever=fusion_retriever,
    llm=OpenAI(model="gpt-4o-mini")
)

# 6. Query
response = query_engine.query(
    "What is the return policy for electronic products?"
)

print(response)
```
