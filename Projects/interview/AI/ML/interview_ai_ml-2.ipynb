# ✅ 1. Supervised vs Unsupervised vs Reinforcement Learning

**What**

* Supervised: learn mapping (X \rightarrow y)
* Unsupervised: learn structure in (X)
* RL: learn a policy by reward

**Why**

* To solve different problem types

**Equation**

Supervised learning objective:

[
\min_\theta \sum_{i=1}^{N} L(f_\theta(x_i), y_i)
]

RL objective:

[
\max_\pi ; \mathbb{E}\left[\sum_t \gamma^t r_t\right]
]

---

# ✅ 2. Bias vs Variance

**What**

* Bias → error from wrong model assumptions
* Variance → error from sensitivity to data

**Why**

* To understand underfitting vs overfitting

**Equation**

[
\mathbb{E}[(y-\hat y)^2] = \text{Bias}^2 + \text{Variance} + \text{Noise}
]

---

# ✅ 3. Overfitting

**What**

Model performs well on train, poorly on test.

**Why**

We want good generalization.

**Equation (regularized loss)**

[
L = L_{data} + \lambda R(w)
]

---

# ✅ 4. L1 vs L2 regularization

**What**

* L1 → absolute values
* L2 → squared values

**Why**

* control model complexity

**Equation**

L1:

[
\lambda \sum |w_i|
]

L2:

[
\lambda \sum w_i^2
]

---

# ✅ 5. Cross-validation

**What**

Repeated train/validation splits.

**Why**

Stable estimate of performance.

**Equation**

[
Score = \frac{1}{K}\sum_{k=1}^{K} score_k
]

---

# ✅ 6. Logistic Regression vs SVM

### Logistic Regression

**What**

Probabilistic linear classifier.

**Why**

Fast and interpretable.

**Equation**

[
P(y=1|x)=\sigma(w^Tx)
]

---

### SVM

**What**

Maximum margin classifier.

**Why**

Strong in high dimensional space.

**Equation**

[
\min \frac{1}{2}||w||^2
]
subject to
[
y_i(w^Tx_i+b)\ge1
]

---

# ✅ 7. Classification metrics

**What & Why**

To evaluate classifier quality.

**Equations**

Accuracy:

[
\frac{TP+TN}{TP+FP+TN+FN}
]

Precision:

[
\frac{TP}{TP+FP}
]

Recall:

[
\frac{TP}{TP+FN}
]

F1:

[
2\cdot\frac{PR}{P+R}
]

# ✅ 13. Vanishing gradient

**What**

Gradients shrink in deep nets.

**Why**

Early layers stop learning.

**Equation**

Repeated chain rule:

[
\prod_k \frac{\partial a_k}{\partial a_{k-1}} \rightarrow 0
]
