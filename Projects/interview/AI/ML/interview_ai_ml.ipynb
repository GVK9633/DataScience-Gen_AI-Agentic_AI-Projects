# ✅ Machine Learning – Most Asked Q&A

---

### 1. Difference between supervised, unsupervised and reinforcement learning?

**Answer**

| Type          | Description                   | Example                             |
| ------------- | ----------------------------- | ----------------------------------- |
| Supervised    | Trained with labeled data     | Text classification, spam detection |
| Unsupervised  | No labels, discover structure | Clustering, topic modeling          |
| Reinforcement | Learn by reward               | Robotics, game agents               |

---

### 2. Bias vs Variance?

**Answer**

* **Bias** → error from wrong assumptions (underfitting)
* **Variance** → error from sensitivity to training data (overfitting)

---

### 3. How to handle overfitting?

**Answer**

* Regularization (L1 / L2)
* More data / augmentation
* Early stopping
* Dropout (DL)
* Cross-validation

---

### 4. Difference between L1 and L2 regularization?

**Answer**

* L1 → creates sparse models (feature selection)
* L2 → shrinks weights, keeps all features

---

### 5. What is cross-validation?

**Answer**

Splitting data into multiple folds to evaluate model stability and avoid lucky splits.

---

### 6. When would you use Logistic Regression vs SVM?

**Answer**

* Logistic Regression → fast, interpretable, large datasets
* SVM → smaller dataset, high dimensional space

(You already used both in your text-classification project.)

---

### 7. What metrics do you use for classification?

**Answer**

* Accuracy
* Precision
* Recall
* F1-score
* ROC-AUC

---

### 8. Why accuracy is not a good metric for imbalanced data?

**Answer**

Because a model can predict the majority class and still show high accuracy.

---

### 9. Explain precision vs recall.

**Answer**

* Precision → how many predicted positives are correct
* Recall → how many actual positives are found

---

### 10. What is TF-IDF?

**Answer**

TF-IDF measures how important a word is to a document in a corpus.

(Directly matches your resume project.)

# ✅ Deep Learning – Most Asked Q&A

---

### 11. What is a neural network?

**Answer**

A model composed of layers of neurons that apply linear transformation + non-linear activation.

---

### 12. What is backpropagation?

**Answer**

Algorithm to compute gradients and update weights using chain rule.

---

### 13. What is vanishing gradient problem?

**Answer**

Gradients become very small in deep networks → early layers stop learning.

---

### 14. How do you handle vanishing gradients?

**Answer**

* ReLU / LeakyReLU
* Batch normalization
* Better initialization
* Residual connections

---

### 15. Difference between sigmoid, tanh and ReLU?

**Answer**

| Activation | Issue                                    |
| ---------- | ---------------------------------------- |
| Sigmoid    | Vanishing gradients                      |
| Tanh       | Zero centered but still saturates        |
| ReLU       | Faster training, less vanishing gradient |

---

### 16. What is dropout?

**Answer**

Randomly disables neurons during training to reduce overfitting.

---

### 17. What is batch normalization?

**Answer**

Normalizes intermediate layer outputs to stabilize and speed training.

---

### 18. Difference between CNN and RNN?

**Answer**

* CNN → spatial data (images)
* RNN → sequential data (text, time series)

---

### 19. What is LSTM and why is it used?

**Answer**

LSTM solves long-term dependency problem in sequences by gated memory.

---

### 20. What is transfer learning?

**Answer**

Using pre-trained models and fine-tuning them for a new task.

(Connect this to your LoRA/QLoRA fine-tuning.)
