In Python, **OOP (Object-Oriented Programming)** is a way of structuring code using **classes and objects**. It helps in writing reusable, scalable, and maintainable code.

Let’s go step by step with **simple examples** 👇

---

# 🔹 1. Class and Object

```python
class Person:
    def __init__(self, name, age):   # constructor
        self.name = name
        self.age = age

    def display(self):
        print(f"Name: {self.name}, Age: {self.age}")


# Creating object
p1 = Person("Vijay", 30)
p1.display()
```

👉 Output:

```
Name: Vijay, Age: 30
```

---

# 🔹 2. Encapsulation (Data Hiding)

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # private variable

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance


acc = BankAccount(1000)
acc.deposit(500)
print(acc.get_balance())
```

👉 Output:

```
1500
```

---

# 🔹 3. Inheritance

```python
class Animal:
    def speak(self):
        print("Animal makes sound")

class Dog(Animal):   # inherits Animal
    def bark(self):
        print("Dog barks")

d = Dog()
d.speak()
d.bark()
```

👉 Output:

```
Animal makes sound
Dog barks
```

---

# 🔹 4. Polymorphism

```python
class Bird:
    def sound(self):
        print("Bird chirps")

class Cat:
    def sound(self):
        print("Cat meows")

# Same method name, different behavior
for obj in [Bird(), Cat()]:
    obj.sound()
```

👉 Output:

```
Bird chirps
Cat meows
```

---

# 🔹 5. Abstraction

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car starts with key")

c = Car()
c.start()
```

👉 Output:

```
Car starts with key
```

---

# 🔹 6. Constructor & Destructor

```python
class Demo:
    def __init__(self):
        print("Constructor called")

    def __del__(self):
        print("Destructor called")

d = Demo()
del d
```

# 🔹 7. Real-Time Example (Interview Style)

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def yearly_salary(self):
        return self.salary * 12


emp = Employee("Vijay", 50000)
print(emp.yearly_salary())
```
