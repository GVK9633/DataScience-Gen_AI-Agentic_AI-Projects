Here is an **interview-friendly explanation** of the common Big-O time complexities, with very small examples.

| Complexity     | Simple definition                                            | Small example               |
| -------------- | ------------------------------------------------------------ | --------------------------- |
| **O(1)**       | Constant time — takes the same time regardless of input size | Access an array element     |
| **O(log n)**   | Input is reduced significantly at each step                  | Binary search               |
| **O(n)**       | Time grows linearly with input size                          | Loop through an array       |
| **O(n log n)** | Linear work combined with logarithmic splitting              | Merge sort                  |
| **O(n²)**      | Every element is compared/worked with every other element    | Nested loops                |
| **O(2ⁿ)**      | Work approximately doubles for every additional input        | Recursive subset generation |

### 1. O(1) — Constant Time

**Definition:** The operation takes approximately the same amount of time regardless of `n`.

```python
arr = [10, 20, 30, 40, 50]

print(arr[2])
```

Output:

```text
30
```

Whether the array has **5 or 5 million elements**, accessing `arr[2]` is one operation.

**Interview:**

> O(1) means constant time; input size doesn't affect the number of operations.

---

### 2. O(log n) — Logarithmic Time

**Definition:** The algorithm reduces the problem size significantly, usually by half, at every step.

Classic example: **Binary Search**.

```python
arr = [10, 20, 30, 40, 50, 60, 70, 80]

# Search for 70
```

Instead of checking every element:

```text
10 20 30 40 | 50 60 70 80
             ↑
```

We eliminate half the search space each time.

```text
8 elements
↓
4 elements
↓
2 elements
↓
1 element
```
def binary_search(arr, target):

    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid          # found
        elif arr[mid] < target:
            low = mid + 1       # search right half
        else:
            high = mid - 1      # search left half

    return -1                   # not found

arr = [10, 20, 30, 40, 50, 60, 70, 80]

print(binary_search(arr, 70))   # 6

print(binary_search(arr, 25))   # -1 (not found)

So the complexity is **O(log n)**.

**Interview:**

> O(log n) means the input is reduced by a factor, commonly half, at every iteration.

---

### 3. O(n) — Linear Time

**Definition:** The algorithm processes each element once.

```python
arr = [10, 20, 30, 40, 50]

for x in arr:
    print(x)
```

For:

```text
n = 5    → 5 operations
n = 100  → 100 operations
n = 1000 → 1000 operations
```

Therefore:

**O(n)**

**Interview:**

> O(n) means execution time grows linearly with the input size.

---

### 4. O(n log n)

**Definition:** The algorithm performs `log n` levels of work, with approximately `n` work at each level.

Classic example: **Merge Sort**.

```text
              [8, 4, 2, 6]
                 /    \
             [8,4]    [2,6]
             /  \      /  \
           [8] [4]   [2] [6]
```

The array is repeatedly divided:

```text
n → n/2 → n/4 → ... → 1
```

There are approximately **log n levels**, and each level processes `n` elements.

Therefore:

**O(n log n)**

**Interview:**

> O(n log n) usually occurs when we repeatedly divide the input and perform linear work at each level.

---

### 5. O(n²) — Quadratic Time

**Definition:** The algorithm performs work proportional to `n × n`.

Usually seen with **nested loops**.

```python
arr = [1, 2, 3, 4]

for i in arr:
    for j in arr:
        print(i, j)
```

For 4 elements:

```text
4 × 4 = 16 operations
```

For 100 elements:

```text
100 × 100 = 10,000
```

Therefore:

**O(n²)**

A common example is **Bubble Sort**.

**Interview:**

> O(n²) means the work grows quadratically, commonly caused by nested loops.

---

### 6. O(2ⁿ) — Exponential Time

**Definition:** The amount of work approximately doubles for every additional input element.

A simple recursive example:

```python
def solve(n):
    if n == 0:
        return

    solve(n - 1)
    solve(n - 1)
```

Each call creates **2 more calls**:

```text
                 solve(3)
                /       \
           solve(2)    solve(2)
            /   \       /   \
         solve solve  solve solve
```

The number of calls grows approximately:

```text
n = 1  → 2
n = 2  → 4
n = 3  → 8
n = 4  → 16
```

Therefore:

**O(2ⁿ)**

**Interview:**

> O(2ⁿ) means the computation grows exponentially, typically because each recursive call creates multiple additional calls.

---

## Easy way to remember

```text
O(1)       → Same work
O(log n)   → Divide
O(n)       → One loop
O(n log n) → Divide + process
O(n²)      → Nested loops
O(2ⁿ)      → Branching recursion
```

### From fastest to slowest

**O(1) → O(log n) → O(n) → O(n log n) → O(n²) → O(2ⁿ)**

For interviews, remember the **pattern**, not just the definitions.

````python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result

arr = [38, 27, 43, 3, 9, 82, 10]
print(merge_sort(arr))   # [3, 9, 10, 27, 38, 43, 82]
````

## 🔹 1. Unsorted, Search Once → **Linear Search** — O(n)

```python
def linear_search(arr, target):
    for i, val in enumerate(arr):
        if val == target:
            return i
    return -1
```

## 🔹 2. Sorted, Any Search → **Binary Search** — O(log n)

```python
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

## 🔹 3. Unsorted, Many Searches → **Sort + Binary Search** — O(n log n) + O(log n)

```python
def sort_and_search(arr, target):
    arr = sorted(arr)
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

## 🔹 4. Existence Only, Many Lookups → **Hash Set** — O(1) average

```python
def hash_search(arr, target):
    s = set(arr)
    return target in s
```

## 🔹 5. Index Lookup, Many Searches → **Hash Map** — O(1) average

```python
def hashmap_search(arr, target):
    index_map = {val: i for i, val in enumerate(arr)}
    return index_map.get(target, -1)
```

## 🔹 6. First Occurrence (Sorted) → **Bisect Left** — O(log n)

```python
import bisect

def first_occurrence(arr, target):
    arr = sorted(arr)
    i = bisect.bisect_left(arr, target)
    return i if i < len(arr) and arr[i] == target else -1
```

## 🔹 7. Last Occurrence (Sorted) → **Bisect Right** — O(log n)

```python
import bisect

def last_occurrence(arr, target):
    arr = sorted(arr)
    i = bisect.bisect_right(arr, target) - 1
    return i if i >= 0 and arr[i] == target else -1
```

## 🔹 8. First Occurrence (Unsorted) → **Linear Scan** — O(n)

```python
def first_occurrence_unsorted(arr, target):
    for i, val in enumerate(arr):
        if val == target:
            return i
    return -1
```

## 🔹 9. All Occurrences → **Linear Scan (collect all)** — O(n)

```python
def all_occurrences(arr, target):
    return [i for i, val in enumerate(arr) if val == target]
```

## 🔹 10. Rotated Sorted Array → **Modified Binary Search** — O(log n)

```python
def search_rotated(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        if arr[low] <= arr[mid]:          # left half sorted
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:                             # right half sorted
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1
```

## 🔹 12. Infinite/Streaming Sorted → **Exponential Search** — O(log n)

```python
def exponential_search(arr, target):
    if arr[0] == target:
        return 0
    i = 1
    while i < len(arr) and arr[i] <= target:
        i *= 2
    low, high = i // 2, min(i, len(arr) - 1)
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

## 🔹 13. Direct Index Access → **Array Indexing** — O(1)

```python
def direct_index(arr, index):
    if 0 <= index < len(arr):
        return arr[index]
    return -1
```

## 🔹 14. Key-Value Store (Dict) → **Dictionary Lookup** — O(1) average

```python
def dict_lookup(d, key):
    return d.get(key, -1)
```

## 🔹 15. Set Membership → **Set Lookup** — O(1) average

```python
def set_lookup(s, value):
    return value in s
```

## 🔹 Complexity Cheat Sheet (Sorted)

| Complexity | Algorithms |
| :--- | :--- |
| **O(1)** | Array Indexing, Dictionary Lookup, Set Lookup |
| **O(log n)** | Binary Search, Bisect Left/Right, Rotated Binary Search, Exponential Search |
| **O(n)** | Linear Search, First Occurrence (Unsorted), All Occurrences |

**In one line:** O(1) is only achievable with **direct index access**, **hash map lookups**, or **set membership** — everything else is O(log n), O(n), or O(n log n) depending on whether data is sorted and how many times you search.
