__new__ – Object Creation
Responsible for creating the instance
Must return the object (cls)
Rarely used in day-to-day coding

````python
class A:
    def __new__(cls):
        print("__new__ called")
        return super().__new__(cls)

    def __init__(self):
        print("__init__ called")

obj = A()
````

🔹 __init__ – Object Initialization \n
Initializes the already created object \n
Cannot create or return a new object \n
Used to assign values

````python
class Person:
    def __init__(self, name):
        self.name = name
obj = Person("Alice")
print(obj.name)  # Output: Alice
````

🔥 Key Interview Point (VERY IMPORTANT)
❗ If __new__ does not return an instance, __init__ will never be called

````python
class Test:
    def __new__(cls):
        print("new")
        return None

    def __init__(self):
        print("init")

Test()
````

🧠 Real-World Use Cases of __new__
1️⃣ Singleton Pattern

````python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance
s1 = Singleton()
s2 = Singleton()
print(s1 is s2)  # Output: True 
````

🎯 One-Line Interview Answer (MEMORIZE THIS)
__new__ creates the object, __init__ initializes it. __new__ is called first and must return an instance; __init__ configures that instance.

````python
from collections import Counter

def count_characters(s):
    return Counter(s)

print(count_characters("hello"))  # Counter({'l': 2, 'h': 1, 'e': 1, 'o': 1})
````

````python
def count_char(s):
    char_count ={}
    for char in s:
        if char in char_count:
            char_count[char] +=1
        else:
            char_count[char] =1
    return char_count
print(count_char("hello"))  # {'h': 1, 'e': 1, 'l': 2, 'o': 1}
````

````python
## 4️⃣ Find duplicate elements


def find_duplicates(arr):
    return [x for x in set(arr) if arr.count(x) > 1]
print(find_duplicates([1, 2, 3, 2, 4, 5, 1]))  # Output: [1, 2]
````

````python
## 6️⃣ Find second largest number


def second_largest(arr):
    return sorted(set(arr))[-2]
print(second_largest([1, 2, 3, 4, 5]))  # Output: 4
print(second_largest([5, 1, 4, 2, 3]))
````

````python
def second_largest(arr):
    first = second = float('-inf')

    for num in arr:
        if num > first:
            second = first
            first = num
        elif first > num > second:
            second = num

    return second

print(second_largest([5, 1, 4, 2, 3]))
````

4️⃣ What happens if we don’t use self?
Answer:
You cannot access instance variables or methods.

````python
class Test:
    def set_value(self):
        value = 10   # local variable

    def print_value(self):
        print(value)  # ERROR

p = Test()
p.set_value()
p.print_value()
````

````python
class Test:
    def set_value(self):
        self.value = 10   # Instance attribute
    
    def print_value(self):
        print(self.value)  # Access instance attribute

t = Test()
t.set_value()
t.print_value()  # Output: 10
````

Difference between instance variable and local variable?


````python
class Test:
    def __init__(self):
        self.x = 10   # instance variable

    def show(self):
        y = 20        # local variable
        print(self.x, y)

t = Test()
t.show()
````

7️⃣ How does self help in data encapsulation?
Answer:
It binds data and behavior to the same object.

````python
class Bank:
    def __init__(self,balance):
        self.balance = balance
    def deposit(self,amount):
        self.balance +=amount    
p=Bank(10000)
p.deposit(5000)
print(p.balance)  # Output: 15000        
````

````python
class Test:
    x = 10  # class variable

    def show(self):
        print(self.x)
        print(Test.x)

Test().show()
````
