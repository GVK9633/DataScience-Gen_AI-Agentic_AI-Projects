# 🧠 Shallow Copy vs Deep Copy in Python

---

## 🔹 What Is a Copy?

A copy creates a **new object**, but the way **nested objects** are handled is what differentiates **shallow** and **deep** copy.

---

## 1️⃣ Shallow Copy

### ✅ Definition

A **shallow copy** creates a new outer object, but **nested objects are still shared (same references)**.

### 📌 Methods

* `copy.copy()`
* `list.copy()`
* slicing (`[:]`)
* `dict.copy()`

---

### 🔍 Example

```python
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)

shallow[0][0] = 99

print(original)  # [[99, 2], [3, 4]]
print(shallow)   # [[99, 2], [3, 4]]
```

🔴 Change affects **both** because inner list is shared.

---

### 📊 Memory View

```
original ──┐
           ├──> [1,2]
shallow  ──┘

original ──┐
           ├──> [3,4]
shallow  ──┘
```

---

## 2️⃣ Deep Copy

### ✅ Definition

A **deep copy** creates a new object **and recursively copies all nested objects**.

### 📌 Method

* `copy.deepcopy()`

---

### 🔍 Example

```python
import copy

original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)

deep[0][0] = 99

print(original)  # [[1, 2], [3, 4]]
print(deep)      # [[99, 2], [3, 4]]
```

🟢 Changes are **isolated**.

---

### 📊 Memory View

```
original ──> [1,2]
deep     ──> [1,2]

original ──> [3,4]
deep     ──> [3,4]
```

---

## 🔁 Key Differences (Interview Table)

| Feature       | Shallow Copy    | Deep Copy         |
| ------------- | --------------- | ----------------- |
| Outer object  | New             | New               |
| Inner objects | Shared          | New               |
| Performance   | Fast            | Slower            |
| Memory usage  | Low             | High              |
| Risk          | Side effects    | Safe              |
| Use case      | Flat structures | Nested structures |

---

## ⚠️ Important Edge Cases

---

### 🔹 Immutable Objects (No Issue)

```python
a = [1, 2, 3]
b = a.copy()
b[0] = 99

print(a)  # [1, 2, 3]
```

✔ Safe because integers are immutable.

---

### 🔹 Objects with Circular References

```python
a = []
a.append(a)

deep = copy.deepcopy(a)  # Works safely
```

`deepcopy` handles cycles using memoization.

---

### 🔹 Custom Classes

```python
class A:
    def __init__(self, x):
        self.x = x

import copy
obj = A([1,2])
clone = copy.deepcopy(obj)
```

You can customize behavior using:

* `__copy__`
* `__deepcopy__`

---

## 🧠 When to Use What?

### ✅ Use Shallow Copy When:

* Data is **flat**
* Nested objects are immutable
* Performance matters

### ✅ Use Deep Copy When:

* Nested mutable objects exist
* You need full isolation
* Avoiding side effects is critical

---

## 🎤 Interview-Ready Answer (Use This)

> “A shallow copy creates a new object but shares references to nested objects, so changes to inner objects affect both copies. A deep copy recursively copies all nested objects, providing complete isolation. Shallow copies are faster and memory-efficient, while deep copies are safer for complex, nested data structures.”

```
Shallow = new container, same contents
Deep    = new container, new contents
```
