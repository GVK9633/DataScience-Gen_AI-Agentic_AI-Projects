## 1️⃣8️⃣ Flatten a nested list

```python
def flatten(lst):
    res = []
    for i in lst:
        if isinstance(i, list):
            res.extend(flatten(i))
        else:
            res.append(i)
    return res
```

## 1️⃣ Print a Square Star Pattern

**Question:**
Print a 4×4 square using `*`

```
* * * *
* * * *
* * * *
* * * *
```

```python
n = 4
for i in range(n):
    for j in range(n):
        print("*", end=" ")
    print()
```

🧠 **Why asked:** Tests basic nested loop understanding.

## 5️⃣ Print Pyramid Pattern

```
   *
  * *
 * * *
* * * *
```

```python
n = 4
for i in range(1, n + 1):
    for space in range(n - i):
        print(" ", end="")
    for star in range(i):
        print("* ", end="")
    print()
```

## 8️⃣ Matrix Addition Using Double Loop

```python
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
result = [[0, 0], [0, 0]]

for i in range(len(A)):
    for j in range(len(A[0])):
        result[i][j] = A[i][j] + B[i][j]

print(result)
```

---

## 9️⃣ Matrix Multiplication

```python
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
result = [[0, 0], [0, 0]]

for i in range(len(A)):
    for j in range(len(B[0])):
        for k in range(len(B)):
            result[i][j] += A[i][k] * B[k][j]

print(result)
```

🧠 **Advanced:** Time complexity → **O(n³)**

## 1️⃣1️⃣ Count Character Frequency

```python
s = "aabbc"

for i in range(len(s)):
    count = 1
    for j in range(i + 1, len(s)):
        if s[i] == s[j]:
            count += 1
    print(s[i], count)
```

## 1️⃣3️⃣ Find All Pairs With Given Sum

```python
arr = [1, 2, 3, 4, 5]
target = 6

for i in range(len(arr)):
    for j in range(i + 1, len(arr)):
        if arr[i] + arr[j] == target:
            print(arr[i], arr[j])
```

## 1️⃣4️⃣ Print Chessboard Pattern

```
# * # *
* # * #
# * # *
* # * #
```

```python
n = 4
for i in range(n):
    for j in range(n):
        if (i + j) % 2 == 0:
            print("#", end=" ")
        else:
            print("*", end=" ")
    print()
```
