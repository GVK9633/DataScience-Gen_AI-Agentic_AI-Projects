input string ="test"
output : "t","te","tes","test","e","es","est","s","st"

```python
s = "test"

result = []

for i in range(len(s)):
    temp = ""
    for j in range(i, len(s)):
        temp += s[j]
        result.append(temp)

# exclude last single character substring if needed
result = result[:-1]

print(result)
```

reverse number 123

```python
# reverse number 123
n = 231
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n //= 10

print(rev)
```
