## **2. Dictionary merge result**

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}

my_op = dict1 | dict2
print(my_op)
```

👉 **Output:**

```python
{'a': 1, 'b': 3, 'c': 4}
```

✔️ Rule:

* Duplicate key `"b"` → value from `dict2` overrides

## **3. Convert lists into dictionary (unequal length)**

```python
list_1 = [1, 2, 3]
list_2 = ["1", "2", "3", "4"]

result = dict(zip(list_1, list_2))
print(result)
```

👉 **Output:**

```python
{1: '1', 2: '2', 3: '3'}
```

✔️ Important concept:

* `zip()` stops at **shortest list**
* Extra `"4"` is ignored

---

### 🔥 Interview Tip:

```python
from itertools import zip_longest

result = dict(zip_longest(list_1, list_2, fillvalue=None))
print(result)
```

👉 Output:

```python
{1: '1', 2: '2', 3: '3', None: '4'}
```
