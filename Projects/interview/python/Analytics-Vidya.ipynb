# Q3. How do you concatenate lists?

There are two common ways.

### Method 1 — `+`

```python
a = [1, 2]
b = [3, 4]

c = a + b

print(c)
```

Output:

```text
[1, 2, 3, 4]
```

`+` creates a **new list**.

---

### Method 2 — `extend()`

```python
a = [1, 2]
b = [3, 4]

a.extend(b)

print(a)
```

Output:

```text
[1, 2, 3, 4]
```

`extend()` modifies the existing list.

### Important interview difference

| `+`                           | `extend()`             |
| ----------------------------- | ---------------------- |
| Creates new list              | Modifies existing list |
| Doesn't change original lists | Changes first list     |

### Interview answer

> "I use `+` when I want a new combined list and `extend()` when I want to add elements to the existing list."

### `for`

Use when you know or can iterate over a collection/range.

### `while`

Use when execution depends on a condition.

# Q7. Can you pass a function as an argument?

**Yes.**

Python treats functions as **first-class objects**, so we can pass them as arguments.

```python
def square(x):
    return x * x

def calculate(func, value):
    return func(value)

print(calculate(square, 5))
```

Output:

```text
25
```

Here:

```text
square
  ↓
passed to calculate()
  ↓
func(5)
  ↓
25
```

### Interview answer

> "Yes, Python functions are first-class objects, so we can pass a function as an argument, return it, or store it in a variable."

This concept is important for **decorators, callbacks and functional programming**.

# Q9. What is List Comprehension?

List comprehension provides a **short way to create a list**.

Normal:

```python
numbers = []

for i in range(5):
    numbers.append(i * 2)

print(numbers)
```

List comprehension:

```python
numbers = [i * 2 for i in range(5)]

print(numbers)
```

Output:

```text
[0, 2, 4, 6, 8]
```

### With condition

```python
even = [i for i in range(10) if i % 2 == 0]

print(even)
```

Output:

```text
[0, 2, 4, 6, 8]
```

### Interview answer

> "List comprehension is a concise way to create a list using an expression, loop and optionally a condition."

# Q10. What are `*args` and `**kwargs`?

Very important interview question.

Used when you want to pass **multiple positional arguments**.

```python
def add(*args):
    return sum(args)

print(add(1, 2, 3, 4))
```

Used for **multiple keyword arguments**.

Inside the function:

is a dictionary.

### Easy memory

# Q14. Shallow Copy vs Deep Copy

This is an important Python interview topic.

Suppose we have nested objects:

```python
import copy

original = [[1, 2], [3, 4]]
```

### Shallow copy

```python
shallow = copy.copy(original)
```

The outer object is copied, but nested objects are still shared.

### Deep copy

```python
deep = copy.deepcopy(original)
```

Nested objects are also copied.

### Example

```python
import copy

original = [[1, 2], [3, 4]]

shallow = copy.copy(original)
deep = copy.deepcopy(original)

original[0][0] = 100

print(shallow)
print(deep)
```

Result conceptually:

```text
shallow → [[100, 2], [3, 4]]
deep    → [[1, 2], [3, 4]]
```

### Interview answer

> "A shallow copy copies the outer object but shares nested objects, while a deep copy recursively copies nested objects as well."

# Q15. What is a Decorator?

A decorator **adds or modifies behavior of a function without changing its original code**.

Example:

```python
def logger(func):

    def wrapper():
        print("Function started")
        func()
        print("Function completed")

    return wrapper
```

Use it:

```python
@logger
def hello():
    print("Hello")

hello()
```

Output:

```text
Function started
Hello
Function completed
```

### Real-world AI example

A decorator can be used for:

```text
Agent function
     ↓
Decorator
     ↓
Logging
     ↓
Execute agent
```

For example, logging execution time or authentication checks.

### Interview answer

> "A decorator is a function that wraps another function to add behavior without modifying the original function."

# Q17. What is Inheritance?

Inheritance allows a child class to **reuse properties and methods of a parent class**.

```python
class Animal:

    def speak(self):
        print("Animal speaks")


class Dog(Animal):
    pass


dog = Dog()

dog.speak()
```

Output:

```text
Animal speaks
```

Here:

```text
Animal
  ↑
 Dog
```

Dog inherits from Animal.

### Interview answer

> "Inheritance allows a child class to reuse and extend the properties and methods of a parent class."

# Q19. What is Method Overriding and Polymorphism?

## Method Overriding

A child class provides its **own implementation** of a parent method.

```python
class Animal:

    def sound(self):
        print("Animal sound")


class Dog(Animal):

    def sound(self):
        print("Bark")


dog = Dog()

dog.sound()
```

Output:

```text
Bark
```

Dog overrides `sound()`.

---

## Polymorphism

Polymorphism means **same interface/method name, different behavior**.

```python
class Dog:
    def sound(self):
        print("Bark")


class Cat:
    def sound(self):
        print("Meow")


for animal in [Dog(), Cat()]:
    animal.sound()
```

Output:

```text
Bark
Meow
```

Same:

```python
animal.sound()
```

Different behavior.

### Interview answer

> "Method overriding allows a child class to redefine a parent method, while polymorphism allows the same interface to behave differently for different objects."

# Q20. What is `super()`?

`super()` is used to **access methods or constructors of the parent class**.

Example:

```python
class Parent:

    def show(self):
        print("Parent")


class Child(Parent):

    def show(self):
        super().show()
        print("Child")


obj = Child()

obj.show()
```

Output:

```text
Parent
Child
```

### Constructor example

```python
class Parent:

    def __init__(self, name):
        self.name = name


class Child(Parent):

    def __init__(self, name, age):
        super().__init__(name)
        self.age = age
```

Here:

```python
super().__init__(name)
```

calls the parent constructor.

### Interview answer

> "`super()` is used to call a parent class method or constructor from the child class."
