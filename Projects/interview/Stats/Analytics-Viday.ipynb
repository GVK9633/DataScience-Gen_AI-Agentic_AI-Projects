## Q1. What is Bayesian Inference? Explain with Monty Hall.

### Bayesian Inference

Bayesian inference means **updating our belief about something when we receive new evidence**.

Formula:

$$
P(A|B)=\frac{P(B|A)P(A)}{P(B)}
$$

Where:

* `P(A)` = Prior probability
* `P(B|A)` = Likelihood
* `P(A|B)` = Posterior probability
* `P(B)` = Evidence

### Simple example

Suppose a medical test is positive.

Before the test:

```text
Probability of disease = 1%
```

After getting a positive test, we use the test accuracy + disease prevalence to calculate the updated probability.

That's Bayesian inference.

### Monty Hall

There are **3 doors**:

```text
Door 1 → Car
Door 2 → Goat
Door 3 → Goat
```

You select Door 1.

Initially:

```text
P(Car behind Door 1) = 1/3
P(Car behind other doors) = 2/3
```

Host opens one of the other doors and shows a goat.

Now:

```text
Stay → 1/3
Switch → 2/3
```

So switching doubles the probability of winning.

### Interview answer

> "Bayesian inference updates the probability of a hypothesis using new evidence. In the Monty Hall problem, the initial 1/3 probability of our chosen door remains, while the remaining 2/3 probability transfers to the other unopened door."

# Q2. Poisson vs Binomial Distribution

This is extremely common.

| Binomial                       | Poisson                   |
| ------------------------------ | ------------------------- |
| Fixed number of trials         | Fixed time/space interval |
| Each trial has success/failure | Counts number of events   |
| Probability `p`                | Average rate `λ`          |
| Example: 10 coin tosses        | Example: calls per minute |

### Binomial example

You toss a coin 10 times.

Question:

> What is probability of getting exactly 6 heads?

```text
n = 10
p = 0.5
```

That's Binomial.

### Poisson example

A call center receives:

```text
5 calls per minute
```

Question:

> What is probability of receiving exactly 8 calls in one minute?

That's Poisson.

### Interview answer

> "Binomial models the number of successes in a fixed number of independent trials, while Poisson models the number of events occurring in a fixed interval given an average event rate."

# Q3. What is the Central Limit Theorem?

### Simple definition

The **Central Limit Theorem (CLT)** says:

> When we take sufficiently large random samples from a population, the distribution of the sample mean tends to become approximately normal, regardless of the original population distribution.

### Example

Suppose salaries in a company are highly skewed.

Take:

```text
Sample 1 → 30 employees → calculate average
Sample 2 → 30 employees → calculate average
Sample 3 → 30 employees → calculate average
...
```

If we plot all those sample averages, their distribution tends toward a **normal distribution**.

### Why important?

CLT allows us to use:

* Confidence intervals
* Hypothesis testing
* Statistical inference

### Interview answer

> "CLT states that the distribution of sample means approaches a normal distribution as sample size becomes sufficiently large, even if the population itself isn't normally distributed."

# Q4. Stratified Sampling vs Simple Random Sampling

### Simple Random Sampling (SRS)

Everyone has an equal chance of being selected.

Example:

```text
1000 employees
     ↓
Randomly select 100
```

### Stratified Sampling

First divide the population into groups called **strata**, then sample from each group.

Example:

```text
1000 employees

Engineering → 600
Sales       → 300
HR          → 100

Take samples from each group
```

### Why use stratified sampling?

To ensure important groups are represented.

### Interview answer

> "Simple random sampling selects observations randomly from the entire population, while stratified sampling divides the population into groups and samples from each group."

# Q5. Law of Large Numbers vs Gambler's Fallacy

These are often confused.

## Law of Large Numbers

As the number of trials increases, the observed average tends to approach the expected probability.

Example:

Fair coin:

```text
Expected heads = 50%
```

After 10 tosses:

```text
Heads = 70%
```

After 10,000 tosses:

```text
Heads ≈ 50%
```

---

## Gambler's Fallacy

Incorrect belief that previous independent events affect future events.

Example:

```text
Heads
Heads
Heads
Heads
Heads
```

Someone says:

> "Next one must be tails."

Wrong.

The next fair coin toss still has:

```text
P(Heads) = 0.5
P(Tails) = 0.5
```

### Interview answer

> "The Law of Large Numbers describes long-run convergence toward expected averages, while the Gambler's Fallacy incorrectly assumes that previous independent outcomes influence the next outcome."

# Q6. What is a P-value?

A p-value helps determine how compatible the observed data is with the **null hypothesis**.

Usually:

```text
p < 0.05
```

is used as a significance threshold, depending on the chosen significance level.

### Example

Suppose:

```text
H₀: New model and old model have the same accuracy.
```

Experiment gives:

```text
p-value = 0.02
```

At α = 0.05:

```text
0.02 < 0.05
```

So we **reject H₀**.

### Important interview point

A p-value does **NOT** mean:

> "Probability that the null hypothesis is true."

### Interview answer

> "A p-value measures how extreme the observed result would be, or something more extreme, assuming the null hypothesis is true."

# Q7. Type I vs Type II Error

Very important.

|                 | Reality: H₀ True | Reality: H₀ False |
| --------------- | ---------------- | ----------------- |
| Reject H₀       | **Type I**       | Correct           |
| Don't reject H₀ | Correct          | **Type II**       |

### Type I

**False Positive**

You say there is an effect when there isn't.

Example:

```text
No fraud
   ↓
Model says fraud
```

That's Type I.

### Type II

**False Negative**

You fail to detect an effect that actually exists.

```text
Fraud exists
   ↓
Model says no fraud
```

That's Type II.

### Easy memory

```text
Type I  → False Positive
Type II → False Negative
```

### Interview answer

> "Type I error is rejecting a true null hypothesis, while Type II error is failing to reject a false null hypothesis."

# Q8. Confidence Interval vs Prediction Interval

### Confidence Interval

Estimates the range for a **population parameter**, such as the mean.

Example:

```text
Average salary = ₹10 LPA

95% CI:
₹9.5L – ₹10.5L
```

### Prediction Interval

Predicts the range for a **future individual observation**.

Example:

```text
Predicted salary = ₹10 LPA

95% Prediction Interval:
₹7L – ₹13L
```

Prediction interval is usually **wider**.

### Interview answer

> "A confidence interval estimates uncertainty around a population parameter, while a prediction interval estimates the range where a future individual observation may fall."

# Q9. How do you determine sample size for A/B testing?

Suppose:

```text
A → Current website
B → New website
```

You want to know whether B improves conversion.

Sample size depends mainly on:

1. Baseline conversion rate
2. Minimum detectable effect
3. Significance level α
4. Statistical power
5. Variability

Typical values:

```text
α = 0.05
Power = 80%
```

Example:

```text
Current conversion = 10%
Expected conversion = 12%
```

You need enough users in each group to reliably detect that difference.

### Interview answer

> "For an A/B test, sample size is determined using the baseline rate, minimum detectable effect, significance level, desired statistical power and variability."

### Important

Don't simply say:

> "More users is always better."

More users can increase statistical power, but also increase cost/time.

# Q10. Parametric vs Non-Parametric Tests

### Parametric

Assumes a particular distribution or structure, often normality.

Examples:

```text
t-test
ANOVA
```

### Non-parametric

Makes fewer distributional assumptions.

Examples:

```text
Mann-Whitney U
Wilcoxon
Kruskal-Wallis
```

### Example

If data approximately satisfies assumptions for a t-test:

```text
Use t-test
```

If assumptions are badly violated or data is ordinal/non-normal:

```text
Consider a non-parametric alternative
```

### Interview answer

> "Parametric tests rely on assumptions about the underlying distribution and parameters, while non-parametric tests make fewer distributional assumptions."

# Q11. What is Bias-Variance Trade-off?

This is **extremely important for ML interviews**.

### Bias

Error from an overly simple model.

```text
High Bias
   ↓
Underfitting
```

### Variance

Error from a model being too sensitive to training data.

```text
High Variance
   ↓
Overfitting
```

### Example

Imagine predicting house prices.

Very simple model:

```text
Price = 2 × area
```

May have:

```text
High Bias
Low Variance
```

Very complex model memorizing training examples:

```text
Very low training error
High test error
```

May have:

```text
Low Bias
High Variance
```

### Goal

Find a balance.

```text
Underfitting ← Good model → Overfitting
High Bias       Balance       High Variance
```

### Interview answer

> "Bias-variance trade-off is the balance between underfitting caused by high bias and overfitting caused by high variance; the goal is to minimize generalization error."

# Q12. L1 vs L2 Regularization

Very important for ML.

## L1 — Lasso

Penalty:

$$
\lambda \sum |w_i|
$$

It can drive some weights exactly to **zero**.

Therefore:

> L1 can perform feature selection.

---

## L2 — Ridge

Penalty:

$$
\lambda \sum w_i^2
$$

It generally makes weights smaller but usually doesn't make them exactly zero.

### Comparison

| L1                 | L2                     |           |              |
| ------------------ | ---------------------- | --------- | ------------ |
| Lasso              | Ridge                  |           |              |
| `                  | w                      | ` penalty | `w²` penalty |
| Can make weights 0 | Shrinks weights        |           |              |
| Feature selection  | Controls large weights |           |              |
| Sparse model       | Usually non-sparse     |           |              |

### Example

Suppose:

```text
w1 = 0.001
w2 = 5
w3 = 0.02
```

L1 may produce:

```text
w1 = 0
w2 = 4.5
w3 = 0
```

### Interview answer

> "L1 regularization adds the absolute values of weights to the loss and can produce sparse models, while L2 adds squared weights and generally shrinks weights toward zero."

# Q13. What is Simpson's Paradox?

Simpson's Paradox occurs when:

> A trend appears in multiple groups but reverses or disappears when the groups are combined.

### Simple example

Suppose two treatments:

```text
Treatment A
Group 1 → 90%
Group 2 → 60%

Treatment B
Group 1 → 80%
Group 2 → 50%
```

A appears better in both groups.

But if the groups have very different sizes, the **overall aggregated result can potentially reverse**.

### Why?

A hidden/confounding variable affects the aggregation.

### Interview answer

> "Simpson's Paradox occurs when an association observed within separate groups reverses or changes when the groups are aggregated, often because of a confounding variable or unequal group sizes."

# Q14. What is Berkson's Paradox?

Berkson's Paradox is a form of **selection bias** caused by conditioning on a selected population.

### Simple example

Suppose:

```text
Skill
 ↑
 │
 │       High performers
 │      /
 │     /
 └────────────→ Experience
```

Now suppose we only look at people selected for a highly competitive job.

Selection depends on both:

```text
Skill
+
Experience
```

Among the selected people, skill and experience can appear negatively associated even if they aren't negatively associated in the overall population.

### Real-world example

Looking only at hospital patients can create misleading relationships between diseases because **hospital admission itself is a selection condition**.

### Interview answer

> "Berkson's Paradox is a selection-bias phenomenon where conditioning on a selected population creates or reverses an association between variables."

# Q15. How do you handle missing data?

This is very important in **real ML projects**.

First determine:

> **Why is the data missing?**

Three important categories:

### 1. MCAR

**Missing Completely At Random**

Missingness is unrelated to observed or unobserved data.

Example:

```text
Random system failure
```

---

### 2. MAR

**Missing At Random**

Missingness depends on observed variables.

Example:

```text
Older customers are less likely
to provide income information.
```

---

### 3. MNAR

**Missing Not At Random**

Missingness depends on the missing value itself or unobserved factors.

Example:

```text
People with very high income
may be less willing to report income.
```

---

## Common imputation techniques

### Numerical

Mean:

```python
df["age"].fillna(df["age"].mean())
```

Median:

```python
df["age"].fillna(df["age"].median())
```

### Categorical

```python
df["city"].fillna(df["city"].mode()[0])
```

### Advanced

* KNN imputation
* Regression imputation
* Multiple imputation
* Model-based approaches

### Important interview point

Don't blindly use mean imputation.

For example, if salary has outliers:

```text
₹3L
₹4L
₹5L
₹50L
```

Median may be more appropriate than mean.

### Interview answer

> "For missing data, I first understand the missingness mechanism, then choose an appropriate strategy such as deletion, median/mode imputation, KNN, regression or multiple imputation while avoiding data leakage."
