# 🔥 MUST-ASKED MCP SERVER INTERVIEW Q&A (3 Yrs Exp)

---

## 1️⃣ What is MCP and why is it needed?

**Answer:**
MCP (Model Context Protocol) is a standard that defines how **LLMs securely interact with tools, data, and context**. It ensures controlled access, reduces hallucinations, and makes LLM systems production-safe.

**One-liner:**

> MCP standardizes how models consume context and call tools safely.

---

## 2️⃣ How is MCP different from a normal LLM API call?

| Normal LLM         | MCP                |
| ------------------ | ------------------ |
| Free-form prompts  | Structured context |
| No control         | Tool-controlled    |
| Hallucination risk | Safer              |
| Hard to audit      | Traceable          |

---

## 3️⃣ Explain MCP server architecture

**Answer:**
An MCP server exposes:

* Context providers (read-only data)
* Tools (controlled functions)
* Schemas (input/output contracts)

The LLM interacts only through these interfaces.

---

## 4️⃣ What are MCP tools?

**Answer:**
MCP tools are **server-side functions** exposed to the LLM that perform controlled operations like fetching data or validating rules.

**Example:**
`get_order_details(order_id)`

---

## 5️⃣ Where should business logic live in MCP?

**Answer:**
Business logic must live **inside MCP tools**, not in prompts or the LLM.

**One-liner:**

> LLM explains; MCP decides.

---

## 6️⃣ How does MCP prevent hallucinations?

**Answer:**
By:

* Restricting model access to approved tools
* Supplying only validated context
* Enforcing structured outputs

---

## 7️⃣ MCP vs RAG – what’s the difference?

| MCP              | RAG              |
| ---------------- | ---------------- |
| Deterministic    | Probabilistic    |
| Decision systems | Knowledge Q&A    |
| Tool-driven      | Retrieval-driven |
| Safer            | Flexible         |

👉 **Use MCP for decisions, RAG for explanations.**

---

## 8️⃣ Can MCP and RAG work together?

**Answer:**
Yes. MCP controls decision logic, while RAG retrieves documents for explanation or support.

---

## 9️⃣ What is a context provider in MCP?

**Answer:**
A context provider supplies **read-only information** like policies, user data, or metadata to the LLM.

---

## 🔟 What kind of data should NOT be exposed via MCP?

**Answer:**

* Raw database access
* Secrets / credentials
* Write operations
* Unvalidated user inputs

---

## 1️⃣1️⃣ How do you handle errors in MCP tools?

**Answer:**
By returning **structured error responses**, not exceptions.

```json
{
  "error": "ORDER_NOT_FOUND"
}
```

---

## 1️⃣2️⃣ How do you make MCP responses auditable?

**Answer:**

* Log tool calls
* Store inputs/outputs
* Return structured JSON
* Track decision reasoning

---

## 1️⃣3️⃣ How do you version MCP tools?

**Answer:**

* Version tool schemas
* Backward compatibility
* Deprecate safely

Example:
`validate_return_v2`

---

## 1️⃣4️⃣ What are schemas in MCP?

**Answer:**
Schemas define **exact input/output formats**, ensuring predictable behavior.

---

## 1️⃣5️⃣ How does MCP improve security?

**Answer:**

* No free-text execution
* Controlled tool access
* Least-privilege design
* No direct DB exposure

---

## 1️⃣6️⃣ What is the role of the LLM in MCP?

**Answer:**
The LLM:

* Chooses tools
* Formats responses
* Explains outcomes
  ❌ Does not enforce rules

---

## 1️⃣7️⃣ MCP in real-world projects – example?

**Answer:**
Return approval, loan eligibility, refunds, compliance checks, fraud analysis.

---

## 1️⃣8️⃣ How do you test MCP tools?

**Answer:**

* Unit test tools
* Mock LLM calls
* Validate schema responses

---

## 1️⃣9️⃣ What happens if MCP server is down?

**Answer:**

* Fail safely
* Fallback logic
* No LLM decision without MCP

---

## 2️⃣0️⃣ Common mistakes in MCP implementation

❌ Business logic in prompt
❌ LLM deciding outcomes
❌ No schema validation
❌ No logging
