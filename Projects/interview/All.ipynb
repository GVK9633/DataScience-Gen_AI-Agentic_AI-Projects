## 1. Explain a RAG Pipeline End-to-End

**RAG = Retrieval-Augmented Generation**

RAG combines **retrieval** (searching a knowledge base) with **generation** (LLM writing an answer).

### 📊 Diagram: End-to-End RAG Pipeline

```
┌─────────────────────────────────────────────────────────────────────────┐
│                         RAG PIPELINE                                     │
├─────────────────────────────────────────────────────────────────────────┤
│                                                                          │
│   ┌──────────────┐                                                       │
│   │  DOCUMENTS   │  (PDFs, websites, databases, etc.)                    │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐    ┌──────────────┐    ┌──────────────┐              │
│   │   CHUNKING   │───▶│  EMBEDDING   │───▶│   VECTOR DB  │              │
│   │  (Split text)│    │ (Convert to  │    │  (Store and  │              │
│   │              │    │  vectors)    │    │   index)     │              │
│   └──────────────┘    └──────────────┘    └──────┬───────┘              │
│                                                   │                      │
│                      INDEXING PHASE               │                      │
│                                                   │                      │
│   ┌──────────────┐                                │                      │
│   │  USER QUERY  │                                │                      │
│   └──────┬───────┘                                │                      │
│          │                                        │                      │
│          ▼                                        │                      │
│   ┌──────────────┐    ┌──────────────┐            │                      │
│   │   EMBEDDING  │───▶│  SIMILARITY  │◀───────────┘                      │
│   │  (Query →    │    │   SEARCH     │                                   │
│   │   vector)    │    │ (Find top-k) │                                   │
│   └──────────────┘    └──────┬───────┘                                   │
│                               │                                          │
│                               ▼                                          │
│                      ┌──────────────┐                                    │
│                      │  RETRIEVED   │                                    │
│                      │  DOCUMENTS   │                                    │
│                      └──────┬───────┘                                    │
│                               │                                          │
│                               ▼                                          │
│                      ┌──────────────┐    ┌──────────────┐               │
│                      │   PROMPT     │───▶│     LLM      │               │
│                      │  (Query +    │    │  (Generate   │               │
│                      │  Context)    │    │   answer)    │               │
│                      └──────────────┘    └──────┬───────┘               │
│                                                   │                      │
│                                                   ▼                      │
│                                           ┌──────────────┐              │
│                                           │   ANSWER     │              │
│                                           └──────────────┘              │
│                                                                          │
│                      RETRIEVAL + GENERATION PHASE                        │
└─────────────────────────────────────────────────────────────────────────┘
```

### 📝 Explanation:

| Phase | What Happens |
|-------|--------------|
| **Indexing** | Documents are split into chunks, converted to embeddings, and stored in a vector database |
| **Retrieval** | User query is converted to an embedding, and similar documents are fetched |
| **Generation** | Retrieved documents + query are passed to LLM, which generates an answer |

### 🧠 Simple Analogy:
> RAG is like an **open-book exam**. The LLM is the student, the vector database is the textbook, and retrieval is finding the right page before answering.

## 2. What Happens During the Indexing Phase of RAG?

### 📊 Diagram: Indexing Phase

```
┌─────────────────────────────────────────────────────────────────────────┐
│                         INDEXING PHASE                                   │
├─────────────────────────────────────────────────────────────────────────┤
│                                                                          │
│   ┌──────────────┐                                                       │
│   │  RAW DOCS    │  (PDF, HTML, TXT, DOCX)                               │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │   PARSING    │  (Extract text, tables, images)                       │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │   CHUNKING   │  (Split into smaller pieces)                          │
│   │              │  e.g., 512 tokens per chunk                           │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │  EMBEDDING   │  (Convert each chunk to a vector)                     │
│   │   MODEL      │  e.g., [0.2, -0.5, 0.8, ...]                         │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │  VECTOR DB   │  (Store vectors + metadata)                           │
│   │  (Index)     │  e.g., Pinecone, Weaviate, FAISS                      │
│   └──────────────┘                                                       │
│                                                                          │
└─────────────────────────────────────────────────────────────────────────┘
```

### 📝 Steps:

| Step | Action | Example |
|------|--------|---------|
| 1 | **Parse** | Extract text from PDF |
| 2 | **Chunk** | Split into 512-token pieces |
| 3 | **Embed** | Convert each chunk to a 768-dim vector |
| 4 | **Store** | Save vector + original text in DB |

### 🧠 One-Liner:
> Indexing = **Prepare the library** – parse, chunk, embed, and store documents so they can be searched later.

## 3. What Happens During the Retrieval/Query Phase?

```
┌─────────────────────────────────────────────────────────────────────────┐
│                      RETRIEVAL / QUERY PHASE                             │
├─────────────────────────────────────────────────────────────────────────┤
│                                                                          │
│   ┌──────────────┐                                                       │
│   │  USER QUERY  │  "What is the refund policy?"                         │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │  EMBEDDING   │  Query → [0.1, -0.3, 0.7, ...]                        │
│   │   MODEL      │                                                       │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │  SIMILARITY  │  Compare query vector with all stored vectors         │
│   │   SEARCH     │  (Cosine similarity, dot product, etc.)               │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │  TOP-K       │  Return top 3-5 most similar chunks                   │
│   │  RESULTS     │                                                       │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │  (Optional)  │  Rerank results for better quality                    │
│   │  RERANKER    │                                                       │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │  CONTEXT +   │  Pass to LLM for answer generation                    │
│   │  QUERY       │                                                       │
│   └──────────────┘                                                       │
│                                                                          │
└─────────────────────────────────────────────────────────────────────────┘
```

### 📝 Steps:

| Step | Action |
|------|--------|
| 1 | Convert query to embedding |
| 2 | Search vector DB for similar vectors |
| 3 | Retrieve top-k chunks |
| 4 | (Optional) Rerank for better relevance |
| 5 | Pass to LLM with query |

### 🧠 One-Liner:
> Retrieval = **Find the right pages** – embed the query, search the vector DB, and return the most relevant chunks.

## 4. How Do You Decide the Right Chunk Size and Chunk Overlap?

### 📊 Diagram: Chunk Size vs Overlap

```
Chunk Size = 512 tokens, Overlap = 50 tokens

Document: [========================================================]
           |--- Chunk 1 (512 tokens) ---|
                    |--- Chunk 2 (512 tokens) ---|
                             |--- Chunk 3 (512 tokens) ---|
           |overlap|  |overlap|  |overlap|
            (50)       (50)       (50)
```

### 📝 Guidelines:

| Factor | Recommendation | Reason |
|--------|----------------|--------|
| **Chunk Size** | 256–1024 tokens | Too small = lost context; Too large = noise |
| **Overlap** | 10–20% of chunk size | Prevents cutting sentences mid-thought |
| **Domain** | Technical docs: smaller chunks; Narrative: larger chunks | Technical = precise; Narrative = context |

### 🧠 Simple Rule:
> **Small chunks** = precise retrieval but may miss context.  
> **Large chunks** = more context but may include irrelevant info.  
> **Overlap** = ensures no information is lost at boundaries.

## 5. Compare Fixed-Size, Recursive, Semantic, and Parent-Child Chunking

### 📊 Comparison Table

| Method | How It Works | Pros | Cons |
|--------|--------------|------|------|
| **Fixed-Size** | Split every N tokens | Simple, fast | Cuts sentences mid-way |
| **Recursive** | Split by paragraphs → sentences → words | Respects structure | Still may split ideas |
| **Semantic** | Split based on meaning (embeddings) | Preserves meaning | Slow, complex |
| **Parent-Child** | Small chunks for search, large for context | Best of both | More storage |

### 📊 Diagram: Parent-Child Chunking

```
Parent Chunk (1024 tokens):
┌─────────────────────────────────────────────────────────────┐
│  "The refund policy states that customers can return..."    │
│                                                              │
│  Child Chunk 1 (256 tokens):  "refund policy states..."     │
│  Child Chunk 2 (256 tokens):  "customers can return..."     │
│  Child Chunk 3 (256 tokens):  "within 30 days..."           │
└─────────────────────────────────────────────────────────────┘
         │
         ▼
Search uses child chunks (precise)
LLM uses parent chunk (full context)
```

### 🧠 One-Liner:
> **Fixed** = simple; **Recursive** = structure-aware; **Semantic** = meaning-aware; **Parent-Child** = precision + context.

## 7. What is an Embedding? How Does an Embedding Model Represent Text?

### 📊 Diagram: Text → Embedding

```
Text: "The cat sat on the mat"
          │
          ▼
┌─────────────────────┐
│   EMBEDDING MODEL   │  (e.g., BERT, OpenAI, Sentence-BERT)
│   (Neural Network)  │
└──────────┬──────────┘
           │
           ▼
Vector: [0.2, -0.5, 0.8, 0.1, -0.3, ...]  (768 dimensions)
```

### 📝 Explanation:

| Concept | Meaning |
|---------|---------|
| **Embedding** | A **dense vector** of numbers representing text |
| **Dimensions** | Usually 384, 768, 1024, or 1536 |
| **Similarity** | Similar meanings → similar vectors |
| **Model** | Neural network trained to map text to vectors |

### 🧠 Simple Analogy:
> An embedding is like a **GPS coordinate for meaning**. Words with similar meanings are close together in this high-dimensional space.

**Example:**
- "king" - "man" + "woman" ≈ "queen"
- "cat" and "kitten" are close
- "cat" and "car" are far

## 8. Explain Cosine Similarity, Dot Product, and Euclidean Distance

### 📊 Diagram: Similarity Metrics

```
         Vector A
            ↑
            │
            │  θ (angle)
            │ /
            │/
            └──────────→ Vector B

Cosine Similarity = cos(θ) = (A · B) / (|A| × |B|)
Dot Product = A · B = |A| × |B| × cos(θ)
Euclidean Distance = |A - B| = √(Σ(aᵢ - bᵢ)²)
```

### 📝 Comparison:

| Metric | Formula | Range | Use Case |
|--------|---------|-------|----------|
| **Cosine Similarity** | \(\frac{A \cdot B}{\|A\| \|B\|}\) | [-1, 1] | Text similarity (most common) |
| **Dot Product** | \(A \cdot B\) | \((-\infty, \infty)\) | When magnitude matters |
| **Euclidean Distance** | \(\sqrt{\sum (a_i - b_i)^2}\) | \([0, \infty)\) | When position matters |

### 🧠 When to Use:

| Metric | Best For |
|--------|----------|
| **Cosine** | Comparing documents of different lengths |
| **Dot Product** | When vector magnitude is important |
| **Euclidean** | When absolute position matters |

```
┌─────────────────────────────────────────────────────────────────────────┐
│                         VECTOR DATABASE                                  │
├─────────────────────────────────────────────────────────────────────────┤
│                                                                          │
│   ┌──────────────┐                                                       │
│   │  VECTOR 1    │  [0.2, -0.5, 0.8, ...]  + metadata                   │
│   │  VECTOR 2    │  [0.1, 0.3, -0.2, ...]  + metadata                   │
│   │  VECTOR 3    │  [-0.4, 0.7, 0.1, ...]  + metadata                   │
│   │  ...         │                                                       │
│   └──────────────┘                                                       │
│                                                                          │
│   Query Vector: [0.15, -0.2, 0.5, ...]                                   │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │  SIMILARITY  │  Compare query with all vectors                       │
│   │   SEARCH     │  (Approximate Nearest Neighbor - ANN)                 │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │  TOP-K       │  Return most similar vectors                          │
│   │  RESULTS     │                                                       │
│   └──────────────┘                                                       │
│                                                                          │
└─────────────────────────────────────────────────────────────────────────┘
```

| 3 | At query time, compute similarity |
| 4 | Return top-k nearest neighbors |

### 🧠 Popular Vector DBs:
- **Pinecone** (managed)
- **Weaviate** (open-source)
- **FAISS** (Facebook, fast)
- **Chroma** (lightweight)

## 11. What is Reranking? Why Do We Need a Reranker?

### 📊 Diagram: Reranking

```
Stage 1: Retrieval (fast, approximate)
Query → Vector DB → Top 100 results

Stage 2: Reranking (slow, precise)
Query + Top 100 → Cross-Encoder → Top 5 results

┌─────────────────────────────────────────────────────────────┐
│  WITHOUT RERANKER          │  WITH RERANKER                  │
├────────────────────────────┼─────────────────────────────────┤
│  Top 5 from vector search  │  Top 5 from reranker            │
│  (may include irrelevant)  │  (much more relevant)           │
└────────────────────────────┴─────────────────────────────────┘
```

### 📝 Why Reranking?

| Reason | Explanation |
|--------|-------------|
| **Vector search is approximate** | May miss subtle relevance |
| **Bi-encoder limitation** | Encodes query and doc separately |
| **Cross-encoder** | Encodes query + doc together → better score |
| **Trade-off** | Slower but more accurate |

### 🧠 One-Liner:
> **Retrieval** = fast but rough; **Reranking** = slow but precise.

## 12. What is Hybrid Search? Why Combine Keyword and Vector Search?

### 📊 Diagram: Hybrid Search

```
┌─────────────────────────────────────────────────────────────────────────┐
│                         HYBRID SEARCH                                    │
├─────────────────────────────────────────────────────────────────────────┤
│                                                                          │
│   Query: "Python list append"                                            │
│          │                                                               │
│          ├──────────────────────┬──────────────────────┐                 │
│          ▼                      ▼                      ▼                 │
│   ┌──────────────┐      ┌──────────────┐      ┌──────────────┐          │
│   │  KEYWORD     │      │   VECTOR     │      │  COMBINE     │          │
│   │  SEARCH      │      │   SEARCH     │      │  (Weighted   │          │
│   │  (BM25)      │      │  (Semantic)  │      │   Sum)       │          │
│   └──────┬───────┘      └──────┬───────┘      └──────┬───────┘          │
│          │                      │                      │                 │
│          ▼                      ▼                      ▼                 │
│   Exact match:           Semantic match:        Final ranked             │
│   "append"               "adding to list"       results                  │
│                                                                          │
└─────────────────────────────────────────────────────────────────────────┘
```

### 📝 Why Combine?

| Search Type | Strength | Weakness |
|-------------|----------|----------|
| **Keyword (BM25)** | Exact matches, rare words | Misses synonyms |
| **Vector** | Semantic meaning | Misses exact terms |
| **Hybrid** | Both | More complex |

### 🧠 One-Liner:
> **Keyword** finds exact words; **Vector** finds meaning. **Hybrid** = best of both.

## 13. How Would You Implement Metadata Filtering in RAG?

### 📊 Diagram: Metadata Filtering

```
Vector DB Entry:
┌─────────────────────────────────────────────────────────────┐
│  Vector: [0.2, -0.5, 0.8, ...]                              │
│  Text: "Refund policy for 2024..."                          │
│  Metadata: {                                                │
│    "source": "policy.pdf",                                  │
│    "date": "2024-01-15",                                    │
│    "category": "refund",                                    │
│    "department": "customer_service"                         │
│  }                                                          │
└─────────────────────────────────────────────────────────────┘

Query: "refund policy" + Filter: {date: "2024", category: "refund"}
          │
          ▼
┌─────────────────────────────────────────────────────────────┐
│  Step 1: Filter by metadata (date, category)                │
│  Step 2: Vector search within filtered results              │
│  Step 3: Return top-k                                       │
└─────────────────────────────────────────────────────────────┘
```

### 📝 Implementation:

| Step | Action |
|------|--------|
| 1 | Store metadata with each chunk |
| 2 | At query time, apply filters |
| 3 | Search only within filtered subset |

### 🧠 One-Liner:
> Metadata filtering = **narrow the search space** before vector search.

## 14. What is Query Rewriting and When Would You Use It?

### 📊 Diagram: Query Rewriting

```
Original Query: "How do I fix it?"
          │
          ▼
┌─────────────────────────────────────────────────────────────┐
│                    QUERY REWRITING                           │
│  (LLM rewrites query to be more specific)                    │
└─────────────────────────────────────────────────────────────┘
          │
          ▼
Rewritten Query: "How do I fix the error 'ModuleNotFoundError' 
                  when importing pandas in Python?"
```

### 📝 When to Use:

| Situation | Example |
|-----------|---------|
| **Vague query** | "Fix it" → "Fix ModuleNotFoundError" |
| **Multi-hop** | "Compare A and B" → "What is A?" + "What is B?" |

| **HyDE** | Generate hypothetical answer, then search |

### 🧠 One-Liner:
> Query rewriting = **improve the question** before searching.

## 📊 Retrieval Metrics

### 1. Precision@k

**One-liner:** Out of the top k retrieved documents, how many are actually relevant?

**Diagram:**
```
Query: "refund policy"
Top 5 retrieved: [Relevant, Irrelevant, Relevant, Relevant, Irrelevant]

Precision@5 = 3 relevant / 5 retrieved = 0.60
```

---

### 2. Recall@k

**One-liner:** Out of all relevant documents in the database, how many did we retrieve in top k?

**Diagram:**
```
Total relevant docs in DB = 10
Top 5 retrieved: [Relevant, Irrelevant, Relevant, Relevant, Irrelevant]

Recall@5 = 3 retrieved relevant / 10 total relevant = 0.30
```

---

### 3. MRR (Mean Reciprocal Rank)

**One-liner:** On average, how high is the first relevant document ranked?

**Diagram:**
```
Query 1: First relevant at position 2 → 1/2 = 0.5
Query 2: First relevant at position 1 → 1/1 = 1.0
Query 3: First relevant at position 4 → 1/4 = 0.25

MRR = (0.5 + 1.0 + 0.25) / 3 = 0.583
```

---

### 4. NDCG (Normalized Discounted Cumulative Gain)

**One-liner:** Rewards relevant documents that appear higher in the ranking, normalized to [0,1].

**Diagram:**
```
Rank 1: Relevant (score 3) → DCG = 3/log₂(2) = 3.0
Rank 2: Irrelevant (score 0) → 0
Rank 3: Relevant (score 2) → 2/log₂(4) = 1.0
DCG = 4.0

Ideal DCG (if sorted perfectly) = 3/log₂(2) + 2/log₂(3) = 4.26
NDCG = 4.0 / 4.26 = 0.94
```

---

## 📊 Generation Metrics

### 5. Faithfulness

**One-liner:** Does the generated answer stick to the retrieved context, or does it hallucinate?

**Diagram:**
```
Context: "Refunds are processed within 30 days."
Answer: "Refunds take 30 days." → Faithful ✅
Answer: "Refunds take 7 days." → Not faithful ❌
```

---

### 6. Answer Relevance

**One-liner:** Does the generated answer actually address the user's question?

**Diagram:**
```
Query: "What is the refund policy?"
Answer: "Refunds take 30 days." → Relevant ✅
Answer: "The weather is nice." → Not relevant ❌
```

---

### 7. Context Relevance

**One-liner:** Is the retrieved context actually useful for answering the query?

**Diagram:**
```
Query: "Refund policy for electronics"
Retrieved: "Electronics refund takes 15 days." → Relevant ✅
Retrieved: "Store hours are 9-5." → Not relevant ❌
```

---

## 📊 Similarity Metrics

### 8. Cosine Similarity

**One-liner:** Measures the angle between two vectors; 1 = same direction, 0 = orthogonal, -1 = opposite.

**Diagram:**
```
Vector A = [1, 0], Vector B = [1, 1]
cos(θ) = (1×1 + 0×1) / (1 × √2) = 0.707
```

---

### 9. Dot Product

**One-liner:** Measures alignment and magnitude; larger = more similar if vectors are normalized.

**Diagram:**
```
A = [1, 2], B = [3, 4]
A · B = (1×3) + (2×4) = 3 + 8 = 11
```

---

### 10. Euclidean Distance

**One-liner:** Straight-line distance between two points; smaller = more similar.

**Diagram:**
```
A = [1, 2], B = [4, 6]
d = √((4-1)² + (6-2)²) = √(9 + 16) = √25 = 5
```
