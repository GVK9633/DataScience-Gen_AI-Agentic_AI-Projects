## 1. Explain a RAG Pipeline End-to-End

**RAG = Retrieval-Augmented Generation**

RAG combines **retrieval** (searching a knowledge base) with **generation** (LLM writing an answer).

### 📊 Diagram: End-to-End RAG Pipeline

```
┌─────────────────────────────────────────────────────────────────────────┐
│                         RAG PIPELINE                                     │
├─────────────────────────────────────────────────────────────────────────┤
│                                                                          │
│   ┌──────────────┐                                                       │
│   │  DOCUMENTS   │  (PDFs, websites, databases, etc.)                    │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐    ┌──────────────┐    ┌──────────────┐              │
│   │   CHUNKING   │───▶│  EMBEDDING   │───▶│   VECTOR DB  │              │
│   │  (Split text)│    │ (Convert to  │    │  (Store and  │              │
│   │              │    │  vectors)    │    │   index)     │              │
│   └──────────────┘    └──────────────┘    └──────┬───────┘              │
│                                                   │                      │
│                      INDEXING PHASE               │                      │
│                                                   │                      │
│   ┌──────────────┐                                │                      │
│   │  USER QUERY  │                                │                      │
│   └──────┬───────┘                                │                      │
│          │                                        │                      │
│          ▼                                        │                      │
│   ┌──────────────┐    ┌──────────────┐            │                      │
│   │   EMBEDDING  │───▶│  SIMILARITY  │◀───────────┘                      │
│   │  (Query →    │    │   SEARCH     │                                   │
│   │   vector)    │    │ (Find top-k) │                                   │
│   └──────────────┘    └──────┬───────┘                                   │
│                               │                                          │
│                               ▼                                          │
│                      ┌──────────────┐                                    │
│                      │  RETRIEVED   │                                    │
│                      │  DOCUMENTS   │                                    │
│                      └──────┬───────┘                                    │
│                               │                                          │
│                               ▼                                          │
│                      ┌──────────────┐    ┌──────────────┐               │
│                      │   PROMPT     │───▶│     LLM      │               │
│                      │  (Query +    │    │  (Generate   │               │
│                      │  Context)    │    │   answer)    │               │
│                      └──────────────┘    └──────┬───────┘               │
│                                                   │                      │
│                                                   ▼                      │
│                                           ┌──────────────┐              │
│                                           │   ANSWER     │              │
│                                           └──────────────┘              │
│                                                                          │
│                      RETRIEVAL + GENERATION PHASE                        │
└─────────────────────────────────────────────────────────────────────────┘
```

### 📝 Explanation:

| Phase | What Happens |
|-------|--------------|
| **Indexing** | Documents are split into chunks, converted to embeddings, and stored in a vector database |
| **Retrieval** | User query is converted to an embedding, and similar documents are fetched |
| **Generation** | Retrieved documents + query are passed to LLM, which generates an answer |

### 🧠 Simple Analogy:
> RAG is like an **open-book exam**. The LLM is the student, the vector database is the textbook, and retrieval is finding the right page before answering.

## 2. What Happens During the Indexing Phase of RAG?

### 📊 Diagram: Indexing Phase

```
┌─────────────────────────────────────────────────────────────────────────┐
│                         INDEXING PHASE                                   │
├─────────────────────────────────────────────────────────────────────────┤
│                                                                          │
│   ┌──────────────┐                                                       │
│   │  RAW DOCS    │  (PDF, HTML, TXT, DOCX)                               │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │   PARSING    │  (Extract text, tables, images)                       │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │   CHUNKING   │  (Split into smaller pieces)                          │
│   │              │  e.g., 512 tokens per chunk                           │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │  EMBEDDING   │  (Convert each chunk to a vector)                     │
│   │   MODEL      │  e.g., [0.2, -0.5, 0.8, ...]                         │
│   └──────┬───────┘                                                       │
│          │                                                               │
│          ▼                                                               │
│   ┌──────────────┐                                                       │
│   │  VECTOR DB   │  (Store vectors + metadata)                           │
│   │  (Index)     │  e.g., Pinecone, Weaviate, FAISS                      │
│   └──────────────┘                                                       │
│                                                                          │
└─────────────────────────────────────────────────────────────────────────┘
```

### 📝 Steps:

| Step | Action | Example |
|------|--------|---------|
| 1 | **Parse** | Extract text from PDF |
| 2 | **Chunk** | Split into 512-token pieces |
| 3 | **Embed** | Convert each chunk to a 768-dim vector |
| 4 | **Store** | Save vector + original text in DB |

### 🧠 One-Liner:
> Indexing = **Prepare the library** – parse, chunk, embed, and store documents so they can be searched later.

| Step | Action |
|------|--------|

| 5 | Pass to LLM with query |

### 🧠 One-Liner:
> Retrieval = **Find the right pages** – embed the query, search the vector DB, and return the most relevant chunks.

## 4. How Do You Decide the Right Chunk Size and Chunk Overlap?

### 📊 Diagram: Chunk Size vs Overlap

```
Chunk Size = 512 tokens, Overlap = 50 tokens

Document: [========================================================]
           |--- Chunk 1 (512 tokens) ---|
                    |--- Chunk 2 (512 tokens) ---|
                             |--- Chunk 3 (512 tokens) ---|
           |overlap|  |overlap|  |overlap|
            (50)       (50)       (50)
```

### 📝 Guidelines:

| Factor | Recommendation | Reason |
|--------|----------------|--------|
| **Chunk Size** | 256–1024 tokens | Too small = lost context; Too large = noise |
| **Overlap** | 10–20% of chunk size | Prevents cutting sentences mid-thought |
| **Domain** | Technical docs: smaller chunks; Narrative: larger chunks | Technical = precise; Narrative = context |

### 🧠 Simple Rule:
> **Small chunks** = precise retrieval but may miss context.  
> **Large chunks** = more context but may include irrelevant info.  
> **Overlap** = ensures no information is lost at boundaries.

## 5. Compare Fixed-Size, Recursive, Semantic, and Parent-Child Chunking

### 📊 Comparison Table

| Method | How It Works | Pros | Cons |
|--------|--------------|------|------|
| **Fixed-Size** | Split every N tokens | Simple, fast | Cuts sentences mid-way |
| **Recursive** | Split by paragraphs → sentences → words | Respects structure | Still may split ideas |
| **Semantic** | Split based on meaning (embeddings) | Preserves meaning | Slow, complex |
| **Parent-Child** | Small chunks for search, large for context | Best of both | More storage |

### 📊 Diagram: Parent-Child Chunking

```
Parent Chunk (1024 tokens):
┌─────────────────────────────────────────────────────────────┐
│  "The refund policy states that customers can return..."    │
│                                                              │
│  Child Chunk 1 (256 tokens):  "refund policy states..."     │
│  Child Chunk 2 (256 tokens):  "customers can return..."     │
│  Child Chunk 3 (256 tokens):  "within 30 days..."           │
└─────────────────────────────────────────────────────────────┘
         │
         ▼
Search uses child chunks (precise)
LLM uses parent chunk (full context)
Metric	Question
Context Precision	“Of what I retrieved, how much was relevant?”
Context Recall	“Did I retrieve all the important information I needed?”
Interview one-liner:
“Context precision measures how much of the retrieved context is relevant to the query and whether relevant chunks are ranked higher than irrelevant chunks.”


```

### 🧠 One-Liner:
> **Fixed** = simple; **Recursive** = structure-aware; **Semantic** = meaning-aware; **Parent-Child** = precision + context.

To check whether a retrieved chunk is **relevant**, you compare the **user's question** with the **information contained in the chunk**.

### Simple example from your Return Policy project

**User question:**

> "Can I return my shoes after 20 days?"

Suppose your Policy Agent retrieves these chunks:

| Chunk | Content                                             | Relevant?    | Why?                                          |
| ----- | --------------------------------------------------- | ------------ | --------------------------------------------- |
| 1     | "Products can be returned within 30 days."          | ✅ Yes        | Directly answers return period                |
| 2     | "Refunds are processed within 5–7 business days."   | ⚠️ Partially | Related to refund, but not return eligibility |
| 3     | "Shoes are available in black, blue and white."     | ❌ No         | Doesn't answer the question                   |
| 4     | "Items must be unused and have original packaging." | ✅ Yes        | Relevant return condition                     |
| 5     | "Customer support is available 24/7."               | ❌ No         | Not relevant                                  |

So your system should ideally prioritize **Chunk 1 and Chunk 4**.

## How does a RAG system actually check this?

There are several methods.

### 1. Similarity score

When you create embeddings:

```text
Question
"Can I return shoes after 20 days?"
        ↓
     Embedding
        ↓
[0.21, 0.84, 0.31, ...]
```

Each document chunk also has an embedding.

The vector database calculates similarity, commonly using **cosine similarity**.

```text
Question ─────────────── Chunk 1
             similarity = 0.91 ✅

Question ─────────────── Chunk 2
             similarity = 0.82 ✅

Question ─────────────── Chunk 3
             similarity = 0.23 ❌
```

Higher similarity generally means the chunk is more semantically related, but **similarity alone does not prove that the chunk actually answers the question**.

---

### 2. Reranker

A **reranker** looks at the question and retrieved chunks together and gives a relevance score.

```text
Question
   +
Retrieved Chunk
   ↓
Reranker
   ↓
Relevance Score
```

Example:

```text
Chunk 1 → 0.96 ✅
Chunk 4 → 0.91 ✅
Chunk 2 → 0.63
Chunk 3 → 0.18 ❌
Chunk 5 → 0.10 ❌
```

Then you might send only the top 2–3 chunks to the LLM.

---

### 3. LLM-based evaluation

You can also ask an evaluator LLM:

> "Does this chunk contain information relevant to answering the question?"

For example:

```text
Question:
Can I return shoes after 20 days?

Chunk:
"Products can be returned within 30 days."

Evaluator:
Relevant = YES
```

This is useful for **offline evaluation**, although you should not blindly trust an LLM judge.

---

## In your project

Your **Policy Agent** can follow:

```text
Customer Question
       ↓
Embedding
       ↓
Vector Search
       ↓
Top 10 chunks
       ↓
Reranker
       ↓
Top 3 relevant chunks
       ↓
Policy Agent
       ↓
LLM
       ↓
Answer
```

### Very important distinction

**Retrieval score ≠ relevance automatically.**

For example:

```text
Similarity = 0.90
```

doesn't necessarily mean:

```text
"This chunk contains the correct policy."
```

The chunk could be semantically similar but still not contain the required answer.

## 7. What is an Embedding? How Does an Embedding Model Represent Text?

### 📊 Diagram: Text → Embedding

```
Text: "The cat sat on the mat"
          │
          ▼
┌─────────────────────┐
│   EMBEDDING MODEL   │  (e.g., BERT, OpenAI, Sentence-BERT)
│   (Neural Network)  │
└──────────┬──────────┘
           │
           ▼
Vector: [0.2, -0.5, 0.8, 0.1, -0.3, ...]  (768 dimensions)
```

### 📝 Explanation:

| Concept | Meaning |
|---------|---------|
| **Embedding** | A **dense vector** of numbers representing text |
| **Dimensions** | Usually 384, 768, 1024, or 1536 |
| **Similarity** | Similar meanings → similar vectors |
| **Model** | Neural network trained to map text to vectors |

### 🧠 Simple Analogy:
> An embedding is like a **GPS coordinate for meaning**. Words with similar meanings are close together in this high-dimensional space.

**Example:**
- "king" - "man" + "woman" ≈ "queen"
- "cat" and "kitten" are close
- "cat" and "car" are far

## 8. Explain Cosine Similarity, Dot Product, and Euclidean Distance

### 📊 Diagram: Similarity Metrics

```
         Vector A
            ↑
            │
            │  θ (angle)
            │ /
            │/
            └──────────→ Vector B

Cosine Similarity = cos(θ) = (A · B) / (|A| × |B|)
Dot Product = A · B = |A| × |B| × cos(θ)
Euclidean Distance = |A - B| = √(Σ(aᵢ - bᵢ)²)
```

### 📝 Comparison:

| Metric | Formula | Range | Use Case |
|--------|---------|-------|----------|
| **Cosine Similarity** | \(\frac{A \cdot B}{\|A\| \|B\|}\) | [-1, 1] | Text similarity (most common) |
| **Dot Product** | \(A \cdot B\) | \((-\infty, \infty)\) | When magnitude matters |
| **Euclidean Distance** | \(\sqrt{\sum (a_i - b_i)^2}\) | \([0, \infty)\) | When position matters |

### 🧠 When to Use:

| Metric | Best For |
|--------|----------|
| **Cosine** | Comparing documents of different lengths |
| **Dot Product** | When vector magnitude is important |
| **Euclidean** | When absolute position matters |

## 11. What is Reranking? Why Do We Need a Reranker?

### 📊 Diagram: Reranking

```
Stage 1: Retrieval (fast, approximate)
Query → Vector DB → Top 100 results

Stage 2: Reranking (slow, precise)
Query + Top 100 → Cross-Encoder → Top 5 results

┌─────────────────────────────────────────────────────────────┐
│  WITHOUT RERANKER          │  WITH RERANKER                  │
├────────────────────────────┼─────────────────────────────────┤
│  Top 5 from vector search  │  Top 5 from reranker            │
│  (may include irrelevant)  │  (much more relevant)           │
└────────────────────────────┴─────────────────────────────────┘
```

### 📝 Why Reranking?

| Reason | Explanation |
|--------|-------------|
| **Vector search is approximate** | May miss subtle relevance |
| **Bi-encoder limitation** | Encodes query and doc separately |
| **Cross-encoder** | Encodes query + doc together → better score |
| **Trade-off** | Slower but more accurate |

### 🧠 One-Liner:
> **Retrieval** = fast but rough; **Reranking** = slow but precise.

## 📊 Retrieval Metrics

### 1. Precision@k

**One-liner:** Out of the top k retrieved documents, how many are actually relevant?

**Diagram:**
```
Query: "refund policy"
Top 5 retrieved: [Relevant, Irrelevant, Relevant, Relevant, Irrelevant]

Precision@5 = 3 relevant / 5 retrieved = 0.60
```

---

### 2. Recall@k

**One-liner:** Out of all relevant documents in the database, how many did we retrieve in top k?

**Diagram:**
```
Total relevant docs in DB = 10
Top 5 retrieved: [Relevant, Irrelevant, Relevant, Relevant, Irrelevant]

Recall@5 = 3 retrieved relevant / 10 total relevant = 0.30
```

---

### 3. MRR (Mean Reciprocal Rank)

**One-liner:** On average, how high is the first relevant document ranked?

**Diagram:**
```
Query 1: First relevant at position 2 → 1/2 = 0.5
Query 2: First relevant at position 1 → 1/1 = 1.0
Query 3: First relevant at position 4 → 1/4 = 0.25

MRR = (0.5 + 1.0 + 0.25) / 3 = 0.583
```

---

### 4. NDCG (Normalized Discounted Cumulative Gain)

**One-liner:** Rewards relevant documents that appear higher in the ranking, normalized to [0,1].

**Diagram:**
```
Rank 1: Relevant (score 3) → DCG = 3/log₂(2) = 3.0
Rank 2: Irrelevant (score 0) → 0
Rank 3: Relevant (score 2) → 2/log₂(4) = 1.0
DCG = 4.0

Ideal DCG (if sorted perfectly) = 3/log₂(2) + 2/log₂(3) = 4.26
NDCG = 4.0 / 4.26 = 0.94
```

## 23. What is multi-head attention?

Definition: Multi-head attention runs several attention mechanisms in parallel. Each head can learn different relationships between tokens.

For example:

* Head 1: Learns syntactic relationships.

* Head 2: Learns relationships between entities.

* Head 3: Learns longer-range dependencies.

These are illustrative roles, not fixed assignments for every trained head.

### Formula ⭐

MultiHead(Q,K,V)=Concat(head1,…,headh)WO\boxed{ MultiHead(Q,K,V)= Concat(head_1,\ldots,head_h)W^O }MultiHead(Q,K,V)=Concat(head1,…,headh)WO

Where:

headi=Attention(QWiQ,KWiK,VWiV)head_i=Attention(QW_i^Q,KW_i^K,VW_i^V)headi=Attention(QWiQ,KWiK,VWiV)

### Interview answer

"Multi-head attention allows the Transformer to attend to information from multiple representation subspaces simultaneously. The outputs of all heads are concatenated and projected to produce the final representation."

## 24. What is a context window?

Definition: A context window is the maximum number of tokens that an LLM can process as input context for a single request, including relevant prompt content and, depending on the API, generated output.

### Example

Suppose a model supports a context window of 128,000 tokens.

It can process a combination of:

* System instructions

* User prompt

* Conversation history

* Retrieved RAG documents

* Generated response

within the applicable token limit.

Important: Context window size is measured in tokens, not characters or words.

## 25. What happens when a prompt exceeds the model's context window?

If the total tokens exceed the model's allowed limit, the request may fail with a context-length error. Some applications may truncate or summarize content before sending it, depending on their implementation.

### Example

```
Model context window = 8,000 tokens
Prompt + history + documents = 10,000 tokens
```

The request exceeds the limit by:

10,000−8,000=2,000 tokens10,000-8,000=\boxed{2,000\text{ tokens}}10,000−8,000=2,000 tokens

### How to solve it

1. Remove unnecessary conversation history.

2. Summarize older messages.

3. Retrieve only relevant documents.

4. Reduce document chunk size.

5. Use a model with a larger context window, if appropriate.

Interview point: A larger context window does not automatically mean better answers. Relevant and high-quality context still matters.

## 26. Explain temperature, Top-K, and Top-P

## Temperature

* Low temperature: More predictable outputs.

* High temperature: More varied outputs.

## 27. Why do LLMs hallucinate?

Definition: Hallucination occurs when an LLM generates information that is false, unsupported, or not grounded in reliable evidence, often while sounding confident.

### Main reasons

1. Next-token prediction: The model is trained to generate likely sequences, not to guarantee factual truth.

2. Incomplete training data: The required information may be absent or poorly represented.

3. Ambiguous prompts: The model may interpret the question incorrectly.

4. Outdated knowledge: The model may not know recent events.

5. Weak grounding: It may generate an answer without consulting trusted sources.

6. Overconfident generation: The model may produce a plausible answer even when uncertain.

### Example

User asks:

> "What is the latest price of a product?"

Without access to current pricing data, the model may produce an outdated or invented price.

Interview answer: "LLMs hallucinate because they optimize for generating plausible token sequences rather than directly verifying facts. Missing knowledge, ambiguous prompts, and lack of external grounding can increase hallucinations."

# 46. How would you reduce latency in an LLM application?

Latency means:

> **How long the user waits for the response.**

For your project:

```text
User
 ↓
Intent Agent
 ↓
Orchestrator
 ↓
Policy RAG
 ↓
Order API
 ↓
Risk Model
 ↓
Decision LLM
 ↓
Response
```

There can be multiple latency points.

### Ways to reduce it

| Technique                | How it helps                        |
| ------------------------ | ----------------------------------- |
| Smaller model            | Faster inference                    |
| Streaming                | User sees response immediately      |
| Caching                  | Avoid repeated work                 |
| Reduce prompt size       | Fewer input tokens                  |
| Retrieve fewer chunks    | Faster RAG                          |
| Reranking efficiently    | Reduce retrieval overhead           |
| Parallel tool calls      | Run independent operations together |
| Avoid unnecessary agents | Reduce number of calls              |
| Batch processing         | Improve throughput where applicable |
| Connection pooling       | Reduce network overhead             |

### Important example

Suppose:

```text
Policy Agent = 300 ms
Order API = 200 ms
Risk Model = 150 ms
```

If Order and Risk are independent:

**Sequential:**

```text
300 + 200 + 150 = 650 ms
```

**Parallel:**

```text
300 + max(200,150)
= 500 ms
```

### Interview answer

> “I would first measure where the latency is coming from. Then I would reduce unnecessary LLM calls, use smaller models where appropriate, cache repeated queries, reduce prompt and retrieval size, stream responses, and parallelize independent operations. In my Return Policy system, Order Validation and Fraud/Risk checks could potentially run in parallel when there is no dependency between them.”

# 47. How would you handle millions of documents in a RAG system?

You should **not put millions of documents directly into the LLM context**.

Instead:

```text
Millions of Documents
        ↓
Document Processing
        ↓
Chunking
        ↓
Embeddings
        ↓
Distributed Vector DB
        ↓
Metadata Index
        ↓
Hybrid Retrieval
        ↓
Top-K
        ↓
Reranker
        ↓
LLM
```

### Key techniques

**1. Chunking**

Break documents into manageable pieces.

```text
Document
   ↓
Chunk 1
Chunk 2
Chunk 3
...
```

**2. Embeddings**

Convert chunks into vectors.

```text
"Return within 30 days"
        ↓
[0.21, 0.84, 0.13, ...]
```

**3. Metadata**

Store:

```text
product_category
country
policy_version
effective_date
document_type
```

This allows filtering before/while retrieval.

**4. Hybrid search**

Use:

```text
Vector Search + Keyword Search
```

This is useful because exact terms such as **"30 days"**, **"order ID"**, or policy codes may be important.

**5. Reranking**

Retrieve perhaps 20 candidates and rerank them to select the most relevant 3–5.

### For your Return Policy project

You might have:

```text
Return Policy
Refund Policy
Exchange Policy
Product-specific Policy
Country-specific Policy
Warranty Policy
```

The Policy Agent retrieves only the relevant policy sections.

### Interview answer

> “For millions of documents, I would use distributed document processing, chunking, embeddings, a scalable vector database, metadata filtering, hybrid search and reranking. I would also implement document versioning and incremental indexing so that we don't reprocess the entire corpus whenever a policy changes.”

# 49. How would you design an LLM system for high concurrency?

**Concurrency = many users sending requests at the same time.**

For example:

```text
1,000 users
     ↓
Load Balancer
     ↓
+----+----+----+
|    |    |    |
API  API  API
     |
     ↓
Queue / Async processing
     |
     ↓
LLM Service
```

### Important techniques

### 1. Horizontal scaling

Instead of:

```text
1 server → 10,000 requests
```

use:

```text
10 servers → distributed requests
```

### 2. Load balancing

Distribute requests across instances.

### 3. Rate limiting

Prevent one customer/system from consuming all resources.

### 4. Autoscaling

Increase instances when traffic increases.

```text
Low traffic → 3 instances
High traffic → 20 instances
```

### 5. Caching

For repeated questions:

> "What is the return period?"

Return cached result instead of calling the LLM every time.

### 6. Streaming

Send generated tokens progressively rather than waiting for the complete response.

### 7. Model routing

Use:

```text
Simple question → smaller/faster model
Complex reasoning → larger model
```

### 8. Async processing

Long-running tasks can be placed into queues.

### Interview answer

> “For high concurrency, I would use stateless application servers behind a load balancer, horizontal autoscaling, rate limiting, caching and connection pooling. I would use queues for asynchronous tasks and model routing to send simple requests to smaller models. I would also monitor P95/P99 latency and error rates to automatically scale the system.”

# 50. Poor answers after deployment — how identify whether problem is retrieval, prompt, model, data or code?

This is one of the **most important debugging questions** for your RAG project.

Don't randomly change the prompt.

Use a **layer-by-layer diagnosis**.

```text
                 Poor Answer
                     |
              Trace the request
                     |
       +-------------+-------------+
       |             |             |
    Retrieval      Prompt        Model
       |             |             |
       +-------------+-------------+
                     |
                   Data
                     |
                  Code
```

## Step 1 — Check the retrieved documents

Question:

> Did the RAG system retrieve the correct policy?

Example:

Customer asks:

> "Can I return an item after 20 days?"

Expected retrieval:

```text
Return period = 30 days
```

But retrieved:

```text
Warranty policy
Exchange policy
```

Then the problem is probably **retrieval**.

---

## Step 2 — Check the source data

Maybe the correct policy wasn't even indexed.

```text
Original Policy
       ↓
Document ingestion
       ↓
Chunking
       ↓
Embedding
       ↓
Vector DB
```

Check:

* Was the document uploaded?
* Was chunking correct?
* Was embedding successful?
* Is the latest policy version indexed?
* Is metadata correct?

If the source data is wrong/outdated → **data problem**.

---

## Step 3 — Check the prompt

Suppose correct context was retrieved:

```text
Context:
Return allowed within 30 days.
```

But prompt says poorly:

```text
Answer the customer.
```

The model may not be sufficiently constrained to use the provided context.

Then test a better grounded prompt.

If quality improves → **prompt problem**.

---

## Step 4 — Check the model

Use the **same**:

```text
Input
Retrieved documents
Prompt
```

and test another model.

If:

```text
Model A → poor
Model B → good
```

then model capability/configuration may be contributing.

---

## Step 5 — Check application code

Sometimes the RAG and prompt are correct but the application passes the wrong data.

For example:

```python
retrieved_docs = retrieve(query)

prompt = build_prompt(user_query, retrieved_docs)

response = llm.invoke(prompt)
```

A coding bug might accidentally send:

```text
user_query
```

without:

```text
retrieved_docs
```

Then the LLM receives no policy context.

---

# Best debugging framework

| Layer          | What to check                                     |
| -------------- | ------------------------------------------------- |
| **Data**       | Is source document correct/current?               |
| **Retrieval**  | Did we retrieve the right chunks?                 |
| **Prompt**     | Did we correctly pass/instruct the model?         |
| **Model**      | Can the model answer correctly with same context? |
| **Code**       | Is data flowing correctly between components?     |
| **Tools**      | Did order/API/tool calls return correct results?  |
| **Evaluation** | Does the issue reproduce on a known test set?     |

I would create a test case like:
