# 🔹 First Say This in Interview (Very Important)

> **We do NOT fine-tune the model for business decisions.**
> We fine-tune it only for **language quality, tone, and structured output**, while MCP enforces the rules and RAG provides knowledge.

This one line shows **maturity**.

---

# 🔹 Where Fine-Tuning Fits in MCP + RAG

```
Business Rules → MCP (NO fine-tuning)
Policy Knowledge → RAG (NO fine-tuning)
Language & Format → Fine-Tuned LLM (YES)
```

---

# 🔹 What We Fine-Tune in This Project ✅

Fine-tuning is used to improve:

1. **Decision explanation style**
2. **Customer-friendly language**
3. **Consistent JSON output**
4. **Tone (polite, compliant, brand-specific)**

---

# 🔹 What We Do NOT Fine-Tune ❌

❌ Return approval logic
❌ Policy rules
❌ Eligibility conditions
❌ Database facts

These remain **deterministic** in MCP.

---

| Aspect                | Fine-Tuning | RAG     |

| Knowledge update      | ❌           | ✅       |
| Style control         | ✅           | ❌       |
| Hallucination control | ❌           | Partial |
| Business logic        | ❌           | ❌       |

# 🔹 How to Explain in 30 Seconds (Memorize ⭐)

> In our return-policy project, we don’t fine-tune the model for decision logic. MCP handles approvals deterministically, and RAG retrieves policy text. We fine-tune the LLM only to generate consistent, customer-friendly explanations and structured JSON outputs, which improves response quality without compromising safety.

---

# 🔹 Interview Follow-Up Questions & Smart Answers

### ❓ Why not fine-tune everything?

Because fine-tuning embeds rules in the model, making updates risky and non-transparent.

---

### ❓ When is fine-tuning justified?

When we need consistent tone, format, or domain-specific phrasing at scale.

---

### ❓ How much data is needed?

Typically **500–2,000 high-quality examples** for formatting/tone improvements.

---

### ❓ Can fine-tuned models hallucinate?

Yes. That’s why MCP still controls decisions.

---

# 🔹 Common Mistakes (Say You Avoided)

❌ Fine-tuning for rules
❌ Using RAG for decisions
❌ Letting LLM approve returns
❌ No fallback to base model

---

# ✅ Final Interview Takeaway

> **Fine-tuning improves how the answer is said, not what the answer is.**
