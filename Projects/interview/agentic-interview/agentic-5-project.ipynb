# 🏗️ Clean Separation: Why This Is Production-Grade

### Core Principle

> **LLMs reason and explain. MCP servers own data and rules. LangGraph orchestrates.**

This prevents:

* Hallucinations
* Tight coupling
* Security leaks
* Un-testable logic

---

# 📁 Layered Architecture Explained

## 1️⃣ MCP Servers Layer (Authoritative Systems)

📂 `mcp_servers/`

### Responsibilities

* Data ownership
* Business rules
* Security & audit boundaries

### Why separate?

* These services already exist in enterprises
* LLMs must *never* bypass them

#### Examples

* `order_mcp.py` → OMS
* `policy_mcp.py` → Legal / policy store
* `finance_mcp.py` → Refund calculations

✅ Deterministic
✅ Auditable
✅ Replaceable with real MCP servers later

---

## 2️⃣ Agent Layer (LLM-Oriented Logic)

📂 `agents/`

Each agent has **one responsibility**.

| Agent             | Responsibility        |
| ----------------- | --------------------- |
| Order Agent       | Fetch facts           |
| Policy Agent      | Retrieve policy (RAG) |
| Eligibility Agent | Apply rules           |
| Refund Agent      | Call finance          |
| Explanation Agent | User-facing response  |

### Why this matters

* Easy testing per agent
* Safe LLM usage
* Clear ownership

> 💡 Notice: **LLM is used ONLY in `explanation_agent`**

This is a **key production rule**.

---

## 3️⃣ Orchestration Layer (LangGraph)

📂 `graph/return_graph.py`

### Why LangGraph?

* Deterministic DAG execution
* Retry, observability, replay
* Clear audit trail

### Pattern Used

✅ **Hierarchical Manager–Worker**

LangGraph = Manager
Agents = Workers

---

## 4️⃣ Entry Point

📄 `main.py`

* No business logic
* Just wiring and invocation

This allows:

* API wrapping (FastAPI)
* Batch processing
* Event-driven triggers
