## 🔹 Core Multi-Agent Orchestration Pattern Names

### 1️⃣ **ReAct Pattern**

**(Reason → Act → Observe → Loop)**

* Thinks, uses tools, observes results
* Loop until completion

📌 Used in:
LangChain ReAct, LangGraph loops

---

### 2️⃣ **Planner–Executor Pattern**

* **Planner Agent** breaks task into steps
* **Executor Agent** performs steps

📌 Used in:
Task automation, workflows

---

### 3️⃣ **Manager–Worker Pattern**

* **Manager (Orchestrator) Agent**
* Multiple **Worker Agents**

📌 Used in:
Software dev agents, enterprise systems

---

### 4️⃣ **Hierarchical Agent Pattern**

* Tree-like agent structure
* Top agent controls sub-agents

📌 Used in:
Complex decision systems

---

### 5️⃣ **Sequential Chain Pattern**

* Agents run one after another

📌 Used in:
Report generation, ETL pipelines

---

### 6️⃣ **Conditional Routing Pattern**

* Next agent depends on output or state

📌 Used in:
Decision trees, validation flows

---

### 7️⃣ **Parallel / Ensemble Pattern**

* Multiple agents run simultaneously
* Results aggregated

📌 Used in:
Research, fact-checking, voting systems

---

### 8️⃣ **Reflection Pattern**

* Agent reviews and improves its own output

📌 Used in:
Quality improvement, self-critique

---

### 9️⃣ **Debate / Consensus Pattern**

* Agents argue or critique each other
* Final answer via consensus

📌 Used in:
High-accuracy reasoning

---

### 🔟 **Tool-Augmented Agent Pattern**

* Agent dynamically chooses tools

📌 Used in:
Search, math, APIs, databases

---

### 1️⃣1️⃣ **Memory-Augmented Agent Pattern**

* Uses embeddings / vector DB for recall

📌 Used in:
RAG, long-term memory agents

---

### 1️⃣2️⃣ **Event-Driven Agent Pattern**

* Agents triggered by events or signals

📌 Used in:
Real-time systems, monitoring

---
