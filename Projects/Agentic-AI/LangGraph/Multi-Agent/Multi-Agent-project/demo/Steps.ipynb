Great question 👍 Let’s go step by step to validate this **dynamic agent framework** on your local system.

---

## ✅ Steps to Validate Locally

### 1. **Unzip the Project**

```bash
unzip dynamic_agent_framework.zip -d dynamic_agent_framework
cd dynamic_agent_framework
```

---

### 2. **Setup Virtual Environment (Recommended)**

```bash
python -m venv venv
source venv/bin/activate   # Mac/Linux
venv\Scripts\activate      # Windows
```

---

### 3. **Install Dependencies**

If I include a `requirements.txt`, you just run:

```bash
pip install -r requirements.txt
```

Otherwise, install the core ones manually:

```bash
pip install langgraph langchain openai httpx pydantic
```

(If you are using Gemini → you need `google-generativeai`.)

---

### 4. **Configure API Keys**

In `.env` (or your shell):

```bash
OPENAI_API_KEY=your_openai_key
GEMINI_API_KEY=your_gemini_key
```

---

### 5. **Validate MCP Servers**

Check that your MCP servers (from `config/mcp_servers.json`) are either mock servers or real endpoints.
Example `mcp_servers.json`:

```json
{
  "server1": "http://localhost:8001",
  "server2": "http://localhost:8002",
  "server3": "http://localhost:8003",
  "server4": "http://localhost:8004"
}
```

You can test them with:

```bash
curl http://localhost:8001/health
```

---

### 6. **Run the Parent Agent**

```bash
python src/main.py
```

This should:

* Load the config from `/config/agents.yaml`
* Dynamically initialize agents (weather, pollution, parent)
* Parent agent will coordinate A2A calls
* RAG + React + Chain-of-Thought features will be used

---

### 7. **Test a Query**

For example:

```python
# inside Python REPL
from src.agents.parent_agent import ParentAgent

parent = ParentAgent()
response = parent.run("What is the weather and pollution in Delhi?")
print(response)
```

---

### 8. **Check Dynamic Extension**

To validate dynamic config:

1. Add a new MCP server or tool in `config/agents.yaml`
2. Re-run `main.py` → the framework should automatically load it **without modifying code**.

---

⚡ That’s it! If all steps run fine, you’ve validated the setup.
