LangGraph Dynamic Agent Framework

A production-friendly, modular Python project scaffold that lets you define agents and MCP tool servers via configuration (YAML). Add a new MCP/tool in config/agents.yaml and the rest loads dynamically — no code changes required.

Features

Config-driven agents (child agents + parent coordinator)
MCP servers managed outside agents (just add URLs in config)
Dynamic tool loading (local helper tools + MCP-provided tools)
LLM selection per agent (OpenAI/Gemini or any provider supported by init_chat_model)
Parent agent orchestration (A2A) with RAG + ReAct + Chain-of-Thought support
Simple RAG pipeline using Chroma (persisted DB)
Project layout

Important: This scaffold uses langchain_mcp_adapters, langgraph, langchain, fastmcp, chromadb (or FAISS), and httpx. See requirements.txt.
How it works (high level)

config/agents.yaml describes each agent: which LLM to use, which MCP servers to connect to, which local tools to attach, and whether to use RAG.
agents/factory.py reads the YAML and dynamically spins up MultiServerMCPClient connections, loads MCP tools, imports local tools, and creates a LangGraph (ReAct) agent per config.
agents/parent_agent.py composes child agents into an A2A orchestrator: it can consult RAG first (if enabled), then route to child agents when appropriate.
To add a new MCP/tool: add the MCP URL and tool names to config/agents.yaml. Restart main.py (no code edits needed).

Quick pointers now that you have the scaffold:

Where to edit: config/agents.yaml — add new mcp_servers, tools (use module:function), and llm entries. The factory will pick them up dynamically on restart.
Run MCPs separately (e.g., python mcps/weather_mcp.py) so the MultiServerMCPClient can connect to them.
RAG: seed rag/documents/*.txt and run main.py (first run triggers setup_vectorstore()).
If you want hot-reload (no restart when agents.yaml changes), say the word and I’ll add a file-watcher implementation.
