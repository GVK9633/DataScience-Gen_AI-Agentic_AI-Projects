## 🤖 Agents

### `agents/weather_agent.py`

```python
from langchain_openai import ChatOpenAI
from tools.weather_tools import get_city_weather, get_country_weather
from config.settings import OPENAI_MODEL

def create_weather_agent():
    llm = ChatOpenAI(model=OPENAI_MODEL)
    tools = {"city": get_city_weather, "country": get_country_weather}

    def run(task: dict):
        if "city" in task:
            return tools["city"](task["city"])
        elif "country" in task:
            return tools["country"](task["country"])
        return "Weather agent: no valid input."

    return run
```

### `agents/pollution_agent.py`

```python
from tools.pollution_tools import get_city_pollution, get_country_pollution
# Simulate Gemini (replace with real SDK when available)
class Gemini:
    def __init__(self, model): self.model = model
    def invoke(self, text): return f"[Gemini-{self.model}] {text}"

def create_pollution_agent():
    llm = Gemini("gemini-1.5-pro")
    tools = {"city": get_city_pollution, "country": get_country_pollution}

    def run(task: dict):
        if "city" in task:
            return tools["city"](task["city"])
        elif "country" in task:
            return tools["country"](task["country"])
        return "Pollution agent: no valid input."

    return run
```

### `agents/parent_agent.py`

```python
from config.settings import AGENT_CONFIG
from agents.agent_factory import AgentFactory

def create_parent_agent():
    factory = AgentFactory()

    def run(task: dict):
        if "topic" not in task:
            return "Parent agent: please specify topic (weather/pollution)."

        topic = task["topic"]
        if topic not in AGENT_CONFIG:
            return f"Parent agent: Unknown topic '{topic}'"

        agent = factory.get_agent(topic)
        return agent(task)

    return run
```

---

### `agents/agent_factory.py`

```python
from agents.weather_agent import create_weather_agent
from agents.pollution_agent import create_pollution_agent

class AgentFactory:
    def __init__(self):
        self._agents = {
            "weather": create_weather_agent,
            "pollution": create_pollution_agent,
        }

    def get_agent(self, name: str):
        if name not in self._agents:
            raise ValueError(f"Unknown agent '{name}'")
        return self._agents[name]()
```

## 🚀 `main.py`

```python
from agents.parent_agent import create_parent_agent

def main():
    parent = create_parent_agent()

    # Example runs
    task1 = {"topic": "weather", "city": "Delhi"}
    task2 = {"topic": "pollution", "country": "India"}
    task3 = {"topic": "unknown"}

    print("Task1:", parent(task1))
    print("Task2:", parent(task2))
    print("Task3:", parent(task3))

if __name__ == "__main__":
    main()
```
