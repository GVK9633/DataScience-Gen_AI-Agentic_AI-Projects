## ✅ What is a gradient?

**A gradient tells how much a model’s output (loss) changes when a parameter changes.**

👉 In one line:

**Gradient = derivative of the loss with respect to a parameter.**

---

## ✅ Very simple intuition

If your model has a weight **w** and loss **L**

[
\text{gradient} = \frac{\partial L}{\partial w}
]

It answers:

> “If I slightly change this weight, how will the loss change?”

---

## ✅ Interview-ready definition

**A gradient is the partial derivative of the loss function with respect to a model parameter, used to update that parameter during training.**

---

## ✅ Why gradient is needed

During training, the optimizer uses gradients to update weights:

```text
new_weight = old_weight - learning_rate × gradient
```

This is how the model learns.

---

## ✅ Tiny numeric example

Suppose:

```text
loss = (w - 5)²
```

Then:

```text
gradient = d(loss)/dw = 2(w - 5)
```

If `w = 3`
→ gradient = `-4`

Meaning:

* increase `w` to reduce loss.

---

## ✅ In deep learning frameworks

Libraries like
**PyTorch** and
**TensorFlow**

automatically compute gradients using **autograd**.

## ✅ Very common follow-up

### Q: When do we compute gradients?

👉 Only during **training**

Example pattern:

```text
forward pass → compute loss → backward() → gradients → optimizer step
```

---

## ✅ One-liner difference

| Term         | Meaning                                         |
| ------------ | ----------------------------------------------- |
| Forward pass | compute predictions                             |
| Loss         | measure error                                   |
| Gradient     | tells how to change weights to reduce that loss |

---

## ⭐ Final one-line answer (perfect for interview)

**A gradient is the derivative of the loss with respect to a model parameter and indicates how that parameter should be adjusted to minimize the loss.**
