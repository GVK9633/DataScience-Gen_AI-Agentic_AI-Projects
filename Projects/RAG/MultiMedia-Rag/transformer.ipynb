## ✅ What is a Transformer?

**A Transformer is a neural network architecture that processes sequences using attention instead of recurrence or convolution.**

👉 One-line answer:

**Transformer = sequence model built mainly on self-attention.**

---

## ✅ Why Transformers were introduced

Before transformers, models like RNN / LSTM:

* processed tokens one by one
* were slow
* struggled with long-range dependencies

Transformers solve this by:

👉 looking at **all tokens at the same time** using attention.

---

## ✅ Core idea (very simple)

For every word (token), the model asks:

> Which other words should I pay attention to?

This is done using **self-attention**.

---

## ✅ Main building blocks of a Transformer

A standard transformer layer has:

1. **Self-Attention**
2. **Feed-Forward Network**
3. **Residual connections + LayerNorm**

In a Transformer layer, Self-Attention lets each token look at all other tokens and compute a weighted mix of their representations to capture context (who is relevant to whom), then a Feed-Forward Network (FFN) applies the same small MLP to each token independently to add non-linearity and feature transformation, and finally Residual connections + LayerNorm stabilize training and preserve original information by adding the input back and normalizing the result.
Example: for the sentence “refund policy applies today”, the token “refund” attends strongly to “policy” in self-attention (context mixing), the FFN then transforms the attended vector of “refund” into a richer feature representation, and residual + LayerNorm compute
output = LayerNorm(x + sublayer(x)), so the model keeps the original token signal while learning a stable, normalized update.

---

## ✅ High-level structure

```text
Input tokens
   ↓
Embedding + Positional Encoding
   ↓
[ Self-Attention + FFN ]  (repeated N times)
   ↓
Output representations
```

---

## ✅ Why positional encoding is needed

Because attention itself does **not know word order**.

So we add position information to embeddings.

---

## ✅ What makes Transformer powerful

* models long-range relationships
* fully parallel (fast on GPU)
* scales very well

## ✅ Very common interview follow-up

### Q. What is the key difference between RNN and Transformer?

**Answer**

RNN:

* processes tokens sequentially

Transformer:

* processes all tokens in parallel using attention

## ⭐ Final interview one-liner

**A transformer is a neural network architecture for sequence modeling that relies on self-attention to capture relationships between tokens instead of using recurrence.**
