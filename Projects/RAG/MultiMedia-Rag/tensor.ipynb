## ✅ What is a tensor?

**A tensor is a multi-dimensional array used to store numbers for machine-learning computations.**

👉 In simple words:

> A tensor is just a **generalized matrix** (data + shape), optimized for GPU / ML operations.

---

## ✅ Very short ladder definition

| Name          | Meaning                     |
| ------------- | --------------------------- |
| Scalar        | single number → `5`         |
| Vector        | 1-D array → `[1,2,3]`       |
| Matrix        | 2-D array → `[[1,2],[3,4]]` |
| 👉 **Tensor** | n-D array (1D, 2D, 3D, …)   |

---

## ✅ Formal interview line

**A tensor is an n-dimensional numerical container that supports fast mathematical operations and automatic differentiation.**

---

## ✅ Why tensors are used (important)

Libraries like
**PyTorch** and
**TensorFlow**
use tensors because:

* they run on CPU / GPU
* they are optimized for large matrix operations
* they integrate with autograd (back-propagation)

---

## ✅ Simple example

```python
import torch

x = torch.tensor([1, 2, 3])
```

Here:

* `x` is a **tensor**
* shape = `(3,)`
* dimension = 1
