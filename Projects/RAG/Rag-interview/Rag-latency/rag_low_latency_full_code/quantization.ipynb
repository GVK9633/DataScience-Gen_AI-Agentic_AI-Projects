# 🧠 What Are Quantized Models?

**Quantized models** are neural network models where **weights (and sometimes activations)** are stored in **lower precision** instead of full 32-bit floating point (FP32).

### 🔹 Goal

* Reduce **model size**
* Reduce **memory usage**
* Improve **inference speed**
* Enable **CPU / edge / low-cost GPU inference**

---

## ❌ Before Quantization

```
FP32 → 32-bit floating point
High accuracy
High memory
Slower inference
```

## ✅ After Quantization

```
INT8 / INT4 → Lower precision
Small size
Fast inference
Slight accuracy loss (usually acceptable)
```

---

# 🔢 Types of Quantized Models

---

## 1️⃣ FP16 / BF16 (Half Precision)

### 🔹 What it is

* Uses 16-bit floating point
* Minimal accuracy loss

### 🔹 Example

```text
FP32 → FP16
```

### 🔹 Used when

* GPU inference
* Training + inference acceleration

📌 Common in:

* NVIDIA GPUs
* Cloud LLM serving

---

## 2️⃣ INT8 Quantization (Most Common)

### 🔹 What it is

* Weights stored as 8-bit integers

### 🔹 Benefits

* ~4× smaller model
* Much faster inference on CPU

### 🔹 Used in

* ONNX Runtime
* TensorRT
* OpenVINO

📌 Popular for **production inference**

---

## 3️⃣ INT4 / 4-bit Quantization (LLM-Focused)

### 🔹 What it is

* 4-bit weight representation
* Massive memory reduction

### 🔹 Techniques

* GPTQ
* AWQ
* NF4 (NormalFloat4)

📌 Used in:

* LLaMA
* Mistral
* Falcon
* Qwen

---

## 4️⃣ Dynamic Quantization

### 🔹 How it works

* Weights quantized ahead of time
* Activations quantized **at runtime**

### 🔹 Pros

* Easy to apply
* Good speedup on CPUs

### 🔹 Cons

* Slightly slower than static quantization

---

## 5️⃣ Static Quantization

### 🔹 How it works

* Weights + activations quantized **offline**
* Requires calibration data

### 🔹 Pros

* Fastest inference
* Best performance

### 🔹 Cons

* More setup

---

## 6️⃣ Quantization-Aware Training (QAT)

### 🔹 What it is

* Model trained while simulating quantization

### 🔹 Benefit

* Highest accuracy among quantized models

### 🔹 Trade-off

* More training cost

# 🎯 When Should You Use Quantized Models?

### ✅ Use When

* Low-latency inference is required
* Running on CPU or limited GPU
* Cost optimization is important
* Deploying on edge devices

### ❌ Avoid When

* Training large models
* Accuracy is extremely critical
* Plenty of GPU memory available
