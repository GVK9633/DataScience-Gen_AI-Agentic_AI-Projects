# 🎯 RAG Evaluation – Interview-Ready Guide (Policy Example)

## Scenario (Use This in Interview)

**Use case:**

> *“Customer return policy Q&A system using RAG”*

**Question:**

> *“What is the return policy for sneakers?”*

**Knowledge Base Documents:**

| Doc ID | Content                                 |
| ------ | --------------------------------------- |
| D1     | Sneakers can be returned within 30 days |
| D2     | Shoes exchange allowed within 15 days   |
| D3     | Apparel refunds take 5–7 days           |
| D4     | Electronics warranty is 1 year          |

**Relevant Docs:** `D1, D2`

---

# 🟢 PART 1: RETRIEVAL METRICS (Before LLM)

These evaluate **how good retrieval is**.

---

## 1️⃣ Precision@K

**Definition (Interview):**

> Out of the top-K retrieved documents, how many are relevant?

### Example

Top-3 retrieved:

```
[D1, D4, D2]
```

Relevant:

* D1 ✅
* D2 ✅

### Calculation

```
Precision@3 = 2 / 3 = 0.66
```

**Interpretation:**
Some noise exists, but relevant docs are present.

---

## 2️⃣ Recall@K

**Definition:**

> Out of all relevant documents, how many did we retrieve?

### Example

Relevant docs = `{D1, D2}`
Retrieved = `{D1, D4, D2}`

### Calculation

```
Recall@3 = 2 / 2 = 1.0
```

**Interpretation:**
No important policy information was missed.

---

## 3️⃣ MRR (Mean Reciprocal Rank)

**Definition:**

> How early does the first relevant document appear?

### Example

Ranking:

```
1 → D4 ❌
2 → D1 ✅
```

```
MRR = 1 / 2 = 0.5
```

**Interpretation:**
Relevant info is not ranked first → needs tuning.

---

## 🟡 Retrieval Insight (Say This)

> “In RAG systems, **high recall is more important than high precision**, because the LLM can ignore irrelevant context but cannot hallucinate missing facts.”

---

# 🔵 PART 2: GENERATION METRICS (After LLM)

Now the retrieved docs go to the LLM.

### Context Sent to LLM

```
Sneakers can be returned within 30 days.
Shoes exchange allowed within 15 days.
```

### LLM Answer

> “Sneakers can be returned within 30 days.”

---

## 4️⃣ Faithfulness

**Definition:**

> Is the answer strictly supported by retrieved context?

### Example

* Answer matches context ✅
* No hallucination ✅

**Faithfulness = High**

❌ Bad example:

> “Sneakers can be returned within 60 days.”

---

## 5️⃣ Answer Relevance

**Definition:**

> Does the answer address the user’s question?

### Example

Query:

> “What is the return policy for sneakers?”

Answer:

> “Sneakers can be returned within 30 days.”

✅ Relevant

❌ Bad:

> “The company also sells electronics.”

---

## 6️⃣ Context Relevance

**Definition:**

> Were the retrieved documents useful for answering the question?

### Good Context

```
Sneakers return policy
Shoes exchange rules
```

### Bad Context

```
Electronics warranty
Office locations
```

# 🧪 PART 3: RAGAS TOOL (Interview Favorite)

**RAGAS** = **RAG Evaluation Framework**

### What RAGAS Measures

| Metric             | Needs Ground Truth |
| ------------------ | ------------------ |
| Faithfulness       | ❌                  |
| Answer Relevance   | ❌                  |
| Context Relevance  | ❌                  |
| Precision / Recall | ✅                  |
