# 🔍 **How to Explain Each Block (Interview-Style)**

## 1️⃣ Query Expansion

> “Instead of a single query, we generate multiple semantic variations using an LLM.”

✔ Improves recall
✔ Reduces wording bias

---

## 2️⃣ Parallel Retrieval

> “Each query retrieves documents independently from vector and keyword stores.”

✔ Covers synonyms
✔ Handles large document sets

---

## 3️⃣ Fusion & Re-Ranking (CORE OF RAG FUSION)

> “We merge results using Reciprocal Rank Fusion, which promotes documents appearing across multiple queries.”

✔ High recall
✔ Noise reduction

---

## 4️⃣ Context Selection

> “We deduplicate and limit context to avoid token overflow and confusion.”

---

## 5️⃣ Grounded Answer Generation

> “The LLM is strictly instructed to answer only from the fused context.”

---

## 6️⃣ Hallucination Detection

> “Even after RAG Fusion, we validate grounding using semantic and LLM-based checks.”

---
