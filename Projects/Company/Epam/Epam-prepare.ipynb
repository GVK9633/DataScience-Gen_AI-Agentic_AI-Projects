# 3. Highest priority: Agentic AI and Multi-Agent Architecture

## Memorize these 10 interview answers

Q1. What is Agentic AI?

An AI system that uses an LLM to reason about a goal, select tools, perform actions, and repeat steps until it reaches a result or a stopping condition.

Memory trick: Goal → Reason → Tool → Observe → Repeat.

Q2. What is a multi-agent architecture?

It consists of multiple specialized AI agents that collaborate through an orchestrator to solve a complex task.

Example: Policy Agent retrieves return rules, Order Agent validates order details, and Fraud Agent checks suspicious return patterns.

Q3. What is an orchestrator?

It controls the workflow, decides which agent or tool runs next, manages shared state, and handles errors and stopping conditions.

Example: The Supervisor sends a return request to the Policy Agent and Order Validation Agent before allowing a refund workflow to continue.

Q4. Why use LangGraph?

LangGraph models an agent workflow as a graph of nodes, edges, and shared state. It supports conditional routing, loops, checkpointing, and human approval workflows.

Remember: Nodes = actions; Edges = transitions; State = shared workflow data.

Q5. What is tool calling?

Tool calling allows an LLM to request a predefined function or API with structured arguments; application code validates and executes the request and returns the result to the model.

Example: The Order Agent calls `get_order_status(order_id)` to check whether an order was delivered.

Q6. What is MCP?

Model Context Protocol is a standard way for AI applications to discover and interact with external tools and resources through MCP servers.

Remember: MCP host/client connects to the MCP server; the server exposes tools and resources.

Q7. MCP vs tool calling?

Tool calling is the mechanism by which an LLM requests a function; MCP is a protocol for exposing and accessing tools and resources consistently across compatible clients and servers.

Q8. What is agent memory?

Memory preserves useful information across steps or conversations, such as current workflow state, conversation history, or user preferences.

Types: Short-term memory for the current task; long-term memory for information retained across sessions.

Q9. What is the ReAct pattern?

ReAct combines reasoning and action: the model determines an action, receives an observation from a tool, and uses that result to decide the next step.

Q10. When should you use multiple agents?

Use multiple agents when tasks have clearly separable responsibilities, different tools or permissions, or independently manageable workflows. Prefer a single agent when it is simpler, cheaper, and sufficiently reliable.

## RAG interview questions and answers

Q1. What is chunking?

Breaking a document into smaller, searchable pieces so retrieval can identify relevant context instead of passing an entire document to the LLM.

Trade-off: small chunks improve precision but may lose context; large chunks preserve context but can introduce irrelevant information.

Q2. What are embeddings?

Embeddings are numerical vectors representing semantic information; similar meanings are generally located near each other in vector space.

Q3. What is vector search?

It retrieves documents by comparing the query embedding with document embeddings using a similarity or distance measure.

Q4. What is hybrid search?

Hybrid search combines lexical search, such as BM25, with semantic vector search to retrieve both exact keyword matches and meaning-related content.

Q5. What is reranking?

A reranker scores the initially retrieved candidates against the query more precisely and reorders them so the most relevant context reaches the LLM.

Q6. What is top-K?

Top-K is the number of candidate documents or chunks returned by a retrieval step.

Q7. How do you improve poor RAG answers?

Check document parsing, chunking, embeddings, retrieval recall, reranking, context quality, prompt instructions, and the final generated answer separately.

Q8. How do you handle tables and images in PDFs?

Use layout-aware parsing and table extraction for structured tables, and OCR or vision models for scanned pages and meaningful images. Preserve table headers, row relationships, and source metadata.

Q9. How do you handle millions of documents?

Use scalable ingestion pipelines, batch embeddings, partitioned or distributed indexes, metadata filters, incremental updates, caching, and measured retrieval latency.

Q10. How do you prevent hallucinations in RAG?

Retrieve relevant and trustworthy evidence, constrain answers to the supplied context, provide citations where possible, evaluate grounding, and abstain when evidence is insufficient.

## RAG evaluation metrics: memorize these definitions

| Metric             | One-line definition                                                                                       |
| ------------------ | --------------------------------------------------------------------------------------------------------- |
| Precision\@K       | Fraction of the top K retrieved results that are relevant.                                                |
| Recall\@K          | Fraction of all relevant results found in the top K results.                                              |
| MRR                | Reciprocal rank of the first relevant result, averaged over queries.                                      |
| NDCG               | Measures ranking quality, giving more credit to relevant results ranked higher.                           |
| Faithfulness       | Measures whether claims in an answer are supported by the retrieved context.                              |
| Answer relevance   | Measures whether the response addresses the user's question.                                              |
| Answer correctness | Measures whether the answer is factually correct against a trusted reference.                             |
| Hallucination rate | Fraction of answers or claims identified as unsupported or fabricated under the chosen evaluation method. |

Memory trick: Retrieval quality = Precision, Recall, MRR, NDCG. Generation quality = Faithfulness, Correctness, Relevance.

One important distinction: a faithful answer can still be incorrect if the retrieved document itself contains incorrect information.

# 5. AI Evaluation and Observability

Interviewers want to know whether you can operate AI systems in production, not just build a demo.

## Memorize these answers

Q: What is the difference between evaluation and observability?

Evaluation measures the quality of the system's outputs against defined criteria. Observability uses logs, metrics, and traces to understand what the system did and why it succeeded or failed.

Q: What do you monitor in a production agent?

Remember Q-L-C-R-S:

- Q — Quality: correctness, faithfulness, tool success.
- L — Latency: total response time and individual step latency.
- C — Cost: input/output tokens, model calls, embedding and infrastructure costs.
- R — Reliability: errors, timeouts, retries, availability.
- S — Security: authorization failures, data leakage, suspicious tool usage.

Q: How do you investigate a wrong answer?

> “I trace the request end to end. I check whether the source data is correct, retrieval returns the right context, the prompt is appropriate, the LLM follows the context, and the tool outputs are valid. I reproduce the failure, add a regression test, fix the root cause, and monitor the result.”

Q: How do you evaluate an agent?

Test final-answer quality, tool-selection accuracy, argument validity, task-completion rate, workflow correctness, safety, latency, and cost. Include normal, edge, adversarial, and failure scenarios.

Q: What is tracing?

Tracing records the sequence and timing of operations for a request, such as API calls, agent transitions, retrieval, model calls, and tool execution.

Production tip: Avoid logging raw prompts, retrieved documents, credentials, or customer PII indiscriminately. Use redaction, access controls, retention limits, and safe identifiers.

## Essential coding patterns

1\. Reverse an array in-place

Pattern: Two pointers; swap from both ends.

```
def reverse_array(arr):    left, right = 0, len(arr) - 1    while left < right:        arr[left], arr[right] = arr[right], arr[left]        left += 1        right -= 1    return arr
```

Time: \\(O(n)\\). Auxiliary space: \\(O(1)\\).

2\. Find duplicates using a set

```
def find_duplicates(arr):    seen = set()    duplicates = set()    for x in arr:        if x in seen:            duplicates.add(x)        else:            seen.add(x)    return list(duplicates)
```

Average time: \\(O(n)\\). Auxiliary space: \\(O(n)\\).

3\. Count character frequencies

```
from collections import Counterdef char_frequency(s):    return Counter(s)
```

Time: \\(O(n)\\). Space: \\(O(k)\\), where \\(k\\) is the number of distinct characters.

4\. Find the second-largest distinct number

```
def second_largest(arr):    largest = second = None    for x in arr:        if largest is None or x > largest:            if x != largest:                second = largest            largest = x        elif x != largest and (            second is None or x > second        ):            second = x    return second
```

Time: \\(O(n)\\). Auxiliary space: \\(O(1)\\). Returns `None` if no second-largest distinct value exists.

5\. Two Sum

```
def two_sum(nums, target):    seen = {}    for i, x in enumerate(nums):        needed = target - x        if needed in seen:            return [seen[needed], i]        seen[x] = i    return []
```

Average time: \\(O(n)\\). Auxiliary space: \\(O(n)\\).

## Metrics: the questions interviewers love

- Precision: Of predicted positives, how many were actually positive?
- Recall: Of actual positives, how many did the model identify?
- F1 score: Harmonic mean of precision and recall.
- Accuracy: Fraction of all predictions that are correct.
- ROC-AUC: Measures how well a classifier ranks positives above negatives across thresholds.

Example: In fraud detection, recall is important because missed fraud is costly. Precision also matters because too many false alerts burden investigators.

For imbalanced data, consider the precision–recall curve, class-specific recall, and the business cost of false positives and false negatives.

### Most likely security questions

Q: How do you prevent prompt injection?

> “I treat retrieved documents and user inputs as untrusted, separate instructions from data, validate tool arguments, restrict tool permissions, and enforce authorization outside the LLM. For sensitive actions, I require explicit approval and maintain audit trails.”

Q: What is RBAC?

Role-Based Access Control grants permissions according to assigned roles, such as customer, support agent, and administrator.

Q: How do you secure a RAG system?

Enforce access controls during retrieval, apply tenant and document-level filters, protect embeddings and source documents, encrypt data, redact PII, and test for cross-user data leakage.

Q: Can you trust an LLM to decide whether a refund is authorized?

No. The LLM may interpret the request, but deterministic backend rules and trusted systems must validate eligibility, permissions and transaction limits.

Q: What is least privilege?

Grant each user, service or agent only the permissions necessary to perform its assigned task.

### ECS vs EKS

- ECS: AWS-native container orchestration; often simpler when the application does not need Kubernetes.
- EKS: Kubernetes-based orchestration; useful when you need Kubernetes capabilities and ecosystem compatibility, but it adds operational complexity.

### How do you reduce Bedrock costs?

1. Select the smallest model that meets quality requirements.
2. Limit unnecessary agent steps and repeated LLM calls.
3. Use retrieval and context filtering to reduce tokens.
4. Cache safe, reusable results where appropriate.
5. Set token and iteration limits.
6. Measure cost per successful task, not only cost per API call.

### How do you improve production reliability?

Use bounded retries with exponential backoff and jitter, request timeouts, idempotency for state-changing operations, circuit breakers where appropriate, health checks, and graceful fallback behavior.

Important: Retrying a payment or refund operation without idempotency can execute the same transaction more than once.
