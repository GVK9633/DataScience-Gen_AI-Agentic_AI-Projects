## 2. LLM / GenAI

### Q5. What is an LLM?

**Answer:**
An **LLM (Large Language Model)** is a neural-network model trained on large amounts of text to understand and generate human-like language.

Examples include GPT, Claude, and Gemini.

---

### Q6. What is a token?

**Answer:**
A token is a small unit of text processed by an LLM.

For example:

```text
"Hello world"
```

may be represented as multiple tokens.

Tokens are important because LLMs have **context-window and usage limits based on tokens**.

---

### Q7. What is temperature?

**Answer:**
Temperature controls the randomness of the model's output.

* Low temperature → more predictable
* High temperature → more creative/random

For factual applications, a lower temperature is often preferred.

---

### Q8. RAG vs Fine-tuning?

**Answer:**

| RAG                             | Fine-tuning                            |
| ------------------------------- | -------------------------------------- |
| Retrieves external information  | Changes model behavior/weights         |
| Knowledge can be updated easily | Updating knowledge requires retraining |
| Good for company documents      | Good for specialized behavior/style    |
| Usually cheaper                 | Can be more expensive                  |

# 3. RAG ⭐

### Q9. What does RAG stand for?

**Answer:**

**Retrieval-Augmented Generation**

It combines:

```text
User Question
      ↓
Retriever
      ↓
Relevant Documents
      ↓
LLM
      ↓
Answer
```

---

### Q10. Why are embeddings used in RAG?

**Answer:**
Embeddings convert text into **numerical vectors** that capture semantic meaning.

This allows us to find documents that are **meaningfully similar** to the user's question.

---

### Q11. What is chunking?

**Answer:**
Breaking a large document into smaller pieces before creating embeddings.

Example:

```text
100-page PDF
      ↓
Chunks
      ↓
500 tokens each
      ↓
Embeddings
      ↓
Vector DB
```

---

### Q12. What happens if chunks are too large?

**Answer:**
The retrieved chunk may contain too much irrelevant information, increasing context usage and potentially making it harder for the LLM to focus on the relevant information.

---

### Q13. What happens if chunks are too small?

**Answer:**
Important context may be split across different chunks, causing retrieval to miss the complete meaning.

---

### Q14. What is a vector database?

**Answer:**
A database designed to store and search vector embeddings efficiently.

Examples:

* FAISS
* Chroma
* Pinecone
* Weaviate
* Milvus

---

### Q15. What is semantic search?

**Answer:**
Searching based on **meaning**, rather than only exact keywords.

Example:

```text
Question:
"How can I get my money back?"

Document:
"Customers can request a refund..."
```

Even though "money back" and "refund" are different words, semantic search can recognize their similar meaning.

---

### Q16. How can you reduce hallucinations in RAG?

**Answer:**

* Improve retrieval quality
* Use appropriate chunking
* Use reranking
* Give the LLM retrieved context
* Instruct it to answer only from provided context
* Return "I don't know" when information isn't available
* Evaluate retrieval and generation separately

# 4. Agentic AI ⭐⭐⭐

### Q17. What is an AI agent?

**Answer:**
An AI agent is a system where an LLM can **reason about a task, decide what action to take, use tools, observe results, and continue until the task is completed**.

---

### Q18. AI Agent vs Agentic AI?

**Answer:**

**AI agent** = an individual system that can perform tasks.

**Agentic AI** = a broader approach/system where AI can autonomously plan, make decisions, use tools, interact with systems, and complete multi-step tasks.

---

### Q19. What is tool calling?

**Answer:**
Tool calling allows an LLM to request an external function/API.

Example:

```text
User
 ↓
LLM
 ↓
"Call weather API"
 ↓
Weather API
 ↓
Result
 ↓
LLM
 ↓
Final answer
```

---

### Q20. What is the ReAct pattern?

**Answer:**

```text
Reason
  ↓
Act
  ↓
Observe
  ↓
Reason
  ↓
Act
```

It allows an agent to iteratively reason, use tools, observe results, and continue.

---

### Q21. Agent vs normal LLM chain?

**Answer:**

**Chain:**

```text
Input → Step 1 → Step 2 → Step 3 → Output
```

The workflow is generally predefined.

**Agent:**

```text
Input → LLM decides next action
              ↓
           Tool
              ↓
           Result
              ↓
       LLM decides again
```

The agent can dynamically choose actions.

# 5. LangChain / LangGraph ⭐

### Q22. What is LangChain?

**Answer:**
A framework for building applications using LLMs, prompts, tools, retrievers, agents, and other components.

---

### Q23. What is LangGraph?

**Answer:**
LangGraph is designed for building **stateful, multi-step, controllable agent workflows** using graphs.

---

### Q24. What is a node in LangGraph?

**Answer:**
A node represents a unit of work/function.

Example:

```text
START
  ↓
Agent Node
  ↓
Tool Node
  ↓
Agent Node
  ↓
END
```

---

### Q25. What is state in LangGraph?

**Answer:**
State contains the information that is passed and updated between nodes during the workflow.

For example:

```python
state = {
    "question": "...",
    "documents": [],
    "answer": ""
}
```

---

### Q26. Why use LangGraph for agents?

**Answer:**
It provides better control over:

* State
* Multi-step workflows
* Conditional routing
* Loops
* Human-in-the-loop
* Tool execution
* Agent orchestration

# 6. Agent Memory

### Q27. What is short-term memory?

**Answer:**
Information needed during the current conversation/task.

Example:

```text
User: My name is Vijay.
User: What is my name?
```

The agent uses the conversation context to answer.

---

### Q28. What is long-term memory?

**Answer:**
Information stored for use across future conversations or tasks.

Example:

```text
User preference → stored in database
```

# 8. Important scenario questions

### Q31. Your RAG system retrieves irrelevant documents. What would you investigate?

**Answer:**

1. Chunk size
2. Chunk overlap
3. Embedding model
4. Query quality
5. Vector similarity method
6. Number of retrieved documents (`top-k`)
7. Metadata filtering
8. Reranking

---

### Q32. Your agent keeps calling the same tool repeatedly. What could you do?

**Answer:**

* Set a maximum iteration limit
* Detect repeated tool calls
* Improve the agent's instructions
* Add state tracking
* Add termination conditions
* Return an error/fallback when the limit is reached

---

### Q33. How would you evaluate a RAG system?

**Answer:**

Evaluate both:

**Retrieval:**

* Precision
* Recall
* Hit rate
* Context relevance

**Generation:**

* Faithfulness
* Answer relevance
* Correctness
* Groundedness

---

### Q34. How would you secure an Agentic AI system?

**Answer:**

* Protect API keys
* Authentication/authorization
* Input validation
* Tool permission controls
* Limit tool access
* Prevent prompt injection
* Sanitize retrieved content
* Logging and auditing
* Rate limiting

---

### Q35. What is prompt injection?

**Answer:**
An attack where malicious instructions are inserted into user input or retrieved content to manipulate the LLM into ignoring its intended instructions or performing unauthorized actions.
