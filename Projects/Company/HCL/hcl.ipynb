## 2. Top interview questions with short answers

### A. LLM and prompt engineering

Q1. What is an LLM?

An LLM (Large Language Model) is a neural network trained on large amounts of text to understand and generate language. It predicts the next token based on the context.

Q2. What is prompt engineering?

Prompt engineering is designing clear instructions, context, examples, constraints and output formats to improve an LLM's response.

Q3. What is the difference between system and user prompts?

* System prompt: defines the model's role, rules, boundaries and behavior.

* User prompt: contains the user's actual request or question.

Q4. What are temperature and hallucination?

* Temperature controls randomness in token selection; lower values generally produce more predictable responses.

* Hallucination occurs when the model produces incorrect or unsupported information. Grounding, retrieval, validation and abstaining when evidence is insufficient help reduce it.

Q5. RAG vs fine-tuning?

| RAG                                                   | Fine-tuning                                             |
| ----------------------------------------------------- | ------------------------------------------------------- |
| Retrieves external information at query time          | Updates model behavior through additional training      |
| Useful for changing documents and company knowledge   | Useful for task-specific patterns, style and behavior   |
| Knowledge can be updated by changing the indexed data | Updating knowledge may require another training process |

Interview tip: RAG and fine-tuning are not mutually exclusive; a production application can use both.

Q7. What are embeddings?

Embeddings are numerical vectors representing the semantic meaning of text, images or other data. Similar meanings tend to have nearby vectors in the embedding space.

Q8. What is semantic search?

Semantic search retrieves content based on meaning rather than exact keyword matches. It commonly uses embeddings and vector similarity.

Q9. What is hybrid search?

Hybrid search combines keyword search, such as BM25, with vector search to capture both exact terms and semantic similarity.

Q10. How do you improve poor RAG answers?

1. Check document parsing and chunk quality.

2. Check whether retrieval returns the correct passages.

3. Tune top-k, filters, embeddings and hybrid search.

4. Add reranking when useful.

5. Improve the prompt and require evidence-based answers.

6. Evaluate retrieval quality and answer correctness separately.

Q11. How do you process tables and images in PDFs?

Use layout-aware document parsing, table extraction and OCR or vision models as appropriate. Preserve table headers, row relationships and page metadata so retrieved evidence remains understandable.

### C. Google Vertex AI, Gemini and Google ADK

The JD specifically names Google's stack, so make this a major preparation area.

Q12. What is Vertex AI?

Vertex AI is Google's managed platform for building, deploying and operating machine-learning and generative AI applications, including access to Gemini models and supporting AI services.

Q13. What is Google ADK?

Google Agent Development Kit (ADK) is a framework for building, orchestrating, testing and deploying AI agents. It supports tools and multi-agent designs.

Q14. What is the difference between an LLM and an AI agent?

An LLM generates responses from context. An AI agent combines a model with instructions, tools, state and an execution loop to carry out tasks.

Q15. What are the main types of agents/workflows in Google ADK?

* LlmAgent: uses an LLM to reason about a task and select actions or tools.

* SequentialAgent: executes sub-agents in a defined order.

* ParallelAgent: runs independent sub-agents in parallel.

* LoopAgent: repeats a workflow according to its stopping conditions.

Q16. What is a callback in ADK?

A callback is a function that runs at a defined point in an agent's lifecycle, allowing you to inspect or modify behavior, log information, apply guardrails or control execution.

Q17. What are sessions and state?

* Session: represents a conversation or interaction history.

* State: stores data needed across steps, such as extracted order details or an approval status.

Q18. How do you connect an agent to an external API?

Define a tool with a clear input/output contract, validate its parameters, authenticate the API request, handle timeouts and errors, and return structured results to the agent. Add authorization and logging.

### D. MCP, A2A and LangGraph

Q19. What is MCP?

MCP (Model Context Protocol) is a standard protocol for connecting AI applications to external tools, data sources and resources.

Example: An agent uses an MCP-connected order service to retrieve an order's delivery status.

Q20. What are the main components of MCP?

* MCP Host: The AI application hosting the interaction.

* MCP Client: Maintains a connection to an MCP server.

* MCP Server: Exposes tools, resources and prompts.

* Tools: Actions such as checking an order or creating a ticket.

Q21. MCP vs A2A?

| MCP                                                        | A2A                                                              |
| ---------------------------------------------------------- | ---------------------------------------------------------------- |
| Connects AI applications to tools and context              | Enables agents to communicate and delegate tasks to other agents |
| Example: an agent calls an order API through an MCP server | Example: a supervisor delegates fraud analysis to a fraud agent  |

Remember: MCP connects agents to capabilities; A2A connects agents to agents.

Q22. What is LangGraph?

LangGraph is a framework for building stateful, graph-based agent workflows. It uses nodes for processing steps, edges for transitions, and conditional edges for routing decisions. It also supports persistence and human-in-the-loop patterns.

Q23. LangChain vs LangGraph?

* LangChain provides building blocks for LLM applications, including models, prompts, retrievers and tools.

* LangGraph provides explicit control over multi-step, stateful workflows and agent execution.

Q24. What is ReAct?

ReAct means Reasoning and Acting. An agent alternates between deciding what to do, calling a tool, observing its result and deciding the next step until it can finish the task.

Q25. What is human-in-the-loop (HITL)?

HITL introduces human review or approval at critical steps. For example, an agent can recommend a refund, but a human must approve refunds above a defined threshold.

### E. Multi-agent system design

Q26. What is a multi-agent system?

A multi-agent system contains multiple agents with different responsibilities that coordinate to complete a larger task.

Example:

* Supervisor Agent — manages execution and delegation.

* Policy Agent — retrieves policy documents.

* Order Agent — checks order and delivery details.

* Risk Agent — evaluates potential abuse.

* Response Agent — prepares the final answer.

Q27. Why not use a single agent?

A single agent may be simpler and cheaper for straightforward tasks. Multiple agents can help separate responsibilities and support parallel work, but add orchestration complexity, latency, cost and failure points.

Q28. How do you prevent agents from getting stuck in loops?

Set maximum iterations, execution timeouts, tool-call limits and explicit completion criteria. Detect repeated tool calls, log execution traces and provide a safe fallback.

Q29. What is the difference between a workflow and an agent?

A workflow follows predefined steps and routing rules. An agent can dynamically choose actions based on its instructions, model output and observations. Production systems often combine deterministic workflows with agent-based decisions.

### Q35. Reverse an array in place

Python

Run

```
def reverse_array(arr):
    left, right = 0, len(arr) - 1

    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1

    return arr

print(reverse_array(["h", "e", "l", "l", "o"]))
```

* Time complexity: O(n)O(n)O(n)

* Auxiliary space: O(1)O(1)O(1)

* Key concept: two pointers and in-place swapping.

### Q39. What is the difference between `async` and synchronous code?

Synchronous code normally waits for an operation to finish before continuing. Asynchronous code can yield control while waiting for I/O, allowing other tasks to progress. This is useful for concurrent API calls, subject to the libraries and runtime used.

## 2. How to implement it using Google ADK and Python

The implementation has three main steps.

1. Expose each specialist agent as an A2A service.

2. Register the remote agents with the Orchestrator using `RemoteA2aAgent`.

3. Let the Orchestrator delegate work, collect results and make the final decision.

Google ADK documents both exposing agents with `to_a2a()` and consuming remote agents with `RemoteA2aAgent`.

### Step 1 — Expose the Policy Agent

Create a Policy Agent using your existing RAG implementation. You do not need to rewrite the retriever or vector database; wrap your existing functionality inside the agent.

Python

Run

```
# policy_service/agent.py

from google.adk.agents import Agent
from google.adk.a2a.utils.agent_to_a2a import to_a2a

policy_agent = Agent(
    name="policy_agent",
    model="gemini-2.5-flash",
    instruction="""
    You are a Return Policy Agent.
    Retrieve the applicable return policy using the
    existing RAG pipeline.
    Return the relevant policy conditions and evidence.
    Do not approve refunds or execute transactions.
    """,
)

# Expose this agent using A2A
a2a_app = to_a2a(policy_agent, port=8001)
```

### Step 3 — Connect remote agents to the Orchestrator

This is the key A2A integration point.

Python

Run

```
# orchestrator/agent.py

from google.adk.agents import Agent
from google.adk.agents.remote_a2a_agent import (
    RemoteA2aAgent,
    AGENT_CARD_WELL_KNOWN_PATH,
)

policy_agent = RemoteA2aAgent(
    name="policy_agent",
    description="Retrieves return policy information",
    agent_card=(
        "http://localhost:8001"
        + AGENT_CARD_WELL_KNOWN_PATH
    ),
)

order_agent = RemoteA2aAgent(
    name="order_validation_agent",
    description="Validates order and delivery details",
    agent_card=(
        "http://localhost:8002"
        + AGENT_CARD_WELL_KNOWN_PATH
    ),
)

root_agent = Agent(
    name="return_orchestrator",
    model="gemini-2.5-flash",
    instruction="""
    You are the Return Policy Orchestrator.

    1. Understand the customer's return request.
    2. Delegate policy questions to policy_agent.
    3. Delegate order checks to order_validation_agent.
    4. Combine the findings.
    5. Apply the authoritative business eligibility rules.
    6. Explain the result and ask for human approval
       when required.

    Never issue a refund solely based on an LLM response.
    """,
    sub_agents=[policy_agent, order_agent],
)
```

This illustrates the documented remote-agent registration pattern. Check the installed ADK version and its A2A integration mode before using the code unchanged; the A2A integration is documented as experimental in the Python ADK documentation.

## 3. How the agents communicate at runtime

Suppose the customer asks, “Can I return order 123?”

1. Angular sends the customer's question to your existing backend.

2. The Orchestrator identifies the return eligibility task and delegates to the relevant remote agents.

3. The Policy Agent returns the applicable policy evidence; the Order Agent returns verified order facts.

4. The Orchestrator applies the business rules to both results and handles conflicts or missing information.

5. The backend returns the final response to Angular. Any refund execution remains subject to authorization and required approval.

Important implementation detail: Registering remote agents as sub-agents does not itself guarantee that both will run on every request. ADK's `RemoteA2aAgent` integration routes to a remote agent per turn; if policy and order checks are both mandatory, use explicit orchestration that invokes both agents and combines their results. Independent checks can run in parallel.
