### **Question 1: Paired t-test for Husband vs Wife IQ**

**Given data:**

| Husband (H) | 117 | 105 | 97 | 105 | 123 | 109 | 86 | 78 | 103 | 107 |
|-------------|-----|-----|----|-----|-----|-----|----|----|-----|-----|
| Wife (W)    | 106 | 98  | 87 | 104 | 116 | 95  | 90 | 69 | 108 | 85  |

**Step 1: Define hypotheses**  
- Null hypothesis \( H_0: \mu_d = 0 \) (no difference, i.e., husbands are not more intelligent)  
- Alternative hypothesis \( H_1: \mu_d > 0 \) (husbands are more intelligent)  
This is a **one-tailed paired t-test**.

**Step 2: Compute differences \( d = H - W \)**

| H | W | d = H - W |
|---|---|-----------|
| 117 | 106 | 11 |
| 105 | 98  | 7  |
| 97  | 87  | 10 |
| 105 | 104 | 1  |
| 123 | 116 | 7  |
| 109 | 95  | 14 |
| 86  | 90  | -4 |
| 78  | 69  | 9  |
| 103 | 108 | -5 |
| 107 | 85  | 22 |

**Step 3: Calculate mean and standard deviation of d**

- Sum of d = 11+7+10+1+7+14−4+9−5+22 = 72  
- Mean \( \bar{d} = \frac{72}{10} = 7.2 \)

Now compute squared deviations:

| d | \( d - \bar{d} \) | \( (d - \bar{d})^2 \) |
|---|-------------------|----------------------|
| 11 | 3.8  | 14.44 |
| 7  | -0.2 | 0.04  |
| 10 | 2.8  | 7.84  |
| 1  | -6.2 | 38.44 |
| 7  | -0.2 | 0.04  |
| 14 | 6.8  | 46.24 |
| -4 | -11.2| 125.44|
| 9  | 1.8  | 3.24  |
| -5 | -12.2| 148.84|
| 22 | 14.8 | 219.04|

Sum of squared deviations = 14.44+0.04+7.84+38.44+0.04+46.24+125.44+3.24+148.84+219.04 = **603.6**

Sample variance \( s_d^2 = \frac{603.6}{10-1} = \frac{603.6}{9} = 67.0667 \)  
Sample standard deviation \( s_d = \sqrt{67.0667} \approx 8.189 \)

**Step 4: Compute t-statistic**  
\[
t = \frac{\bar{d}}{s_d / \sqrt{n}} = \frac{7.2}{8.189 / \sqrt{10}} = \frac{7.2}{8.189 / 3.1623} = \frac{7.2}{2.589} \approx 2.781
\]

**Step 5: Critical value**  
Degrees of freedom = \( n-1 = 9 \)  
For one-tailed test at \( \alpha = 0.05 \), \( t_{0.05, 9} = 1.833 \) (from t-table)

**Step 6: Decision**  
Since \( t = 2.781 > 1.833 \), we **reject \( H_0 \)**.

**Conclusion:** There is sufficient evidence at the 5% level to conclude that husbands are more intelligent than their wives.

### **Question 2: Two-way ANOVA (without replication)**

**Data:**

| Subject        | Method A | Method B | Method C |
|----------------|----------|----------|----------|
| Mathematics    | 78       | 82       | 85       |
| Physics        | 74       | 79       | 81       |
| Chemistry      | 80       | 84       | 86       |
| Programming    | 85       | 88       | 91       |

---

**a) Identify the two factors**  
- **Factor 1:** Teaching Method (3 levels: A, B, C)  
- **Factor 2:** Subject (4 levels: Mathematics, Physics, Chemistry, Programming)

---

**b) Null hypotheses**  
- For Teaching Method:  
  \( H_0^{Method} \): There is no significant difference in average marks due to teaching method.  
- For Subject:  
  \( H_0^{Subject} \): There is no significant difference in average marks due to subject.

---

**c) Perform two-way ANOVA (without replication)**

Let rows = Subjects (4), columns = Methods (3)

**Step 1: Row totals (Subjects)**

| Subject        | Total |
|----------------|-------|
| Mathematics    | 78+82+85 = 245 |
| Physics        | 74+79+81 = 234 |
| Chemistry      | 80+84+86 = 250 |
| Programming    | 85+88+91 = 264 |

**Step 2: Column totals (Methods)**

| Method | Total |
|--------|-------|
| A      | 78+74+80+85 = 317 |
| B      | 82+79+84+88 = 333 |
| C      | 85+81+86+91 = 343 |

**Step 3: Grand total**  
Grand total = 245+234+250+264 = 993  
Number of observations \( N = 4 \times 3 = 12 \)

**Step 4: Correction factor (CF)**  
\[
CF = \frac{(\text{Grand total})^2}{N} = \frac{993^2}{12} = \frac{986049}{12} = 82170.75
\]

**Step 5: Total Sum of Squares (SST)**  
Square each observation and sum:

| Observation | Square |
|-------------|--------|
| 78          | 6084   |
| 82          | 6724   |
| 85          | 7225   |
| 74          | 5476   |
| 79          | 6241   |
| 81          | 6561   |
| 80          | 6400   |
| 84          | 7056   |
| 86          | 7396   |
| 85          | 7225   |
| 88          | 7744   |
| 91          | 8281   |

Sum of squares = 6084+6724+7225+5476+6241+6561+6400+7056+7396+7225+7744+8281 = **82413**

\[
SST = 82413 - 82170.75 = 242.25
\]

**Step 6: Sum of Squares for Subjects (SSR)**  
\[
SSR = \frac{\sum (\text{Row total})^2}{k} - CF
\]
where \( k = \) number of columns = 3

Row totals squared:  
245² = 60025, 234² = 54756, 250² = 62500, 264² = 69696  
Sum = 60025+54756+62500+69696 = 246977  
\[
SSR = \frac{246977}{3} - 82170.75 = 82325.6667 - 82170.75 = 154.9167
\]

**Step 7: Sum of Squares for Methods (SSC)**  
\[
SSC = \frac{\sum (\text{Column total})^2}{r} - CF
\]
where \( r = \) number of rows = 4

Column totals squared:  
317² = 100489, 333² = 110889, 343² = 117649  
Sum = 100489+110889+117649 = 329027  
\[
SSC = \frac{329027}{4} - 82170.75 = 82256.75 - 82170.75 = 86
\]

**Step 8: Error Sum of Squares (SSE)**  
\[
SSE = SST - SSR - SSC = 242.25 - 154.9167 - 86 = 1.3333
\]

**Step 9: Degrees of freedom**  
- df(Subject) = r−1 = 3  
- df(Method) = k−1 = 2  
- df(Error) = (r−1)(k−1) = 3×2 = 6  
- df(Total) = N−1 = 11

**Step 10: Mean Squares**  
\[
MSR = \frac{SSR}{3} = \frac{154.9167}{3} = 51.6389
\]  
\[
MSC = \frac{86}{2} = 43
\]  
\[
MSE = \frac{1.3333}{6} = 0.2222
\]

**Step 11: F-ratios**  
\[
F_{Subject} = \frac{MSR}{MSE} = \frac{51.6389}{0.2222} \approx 232.38
\]  
\[
F_{Method} = \frac{MSC}{MSE} = \frac{43}{0.2222} \approx 193.5
\]

**Step 12: Critical values from F-table at 5%**  
- For Subject: \( F_{0.05, 3, 6} = 4.76 \)  
- For Method: \( F_{0.05, 2, 6} = 5.14 \)

**Step 13: Decision**  
- For Subject: \( 232.38 > 4.76 \) → Reject \( H_0^{Subject} \)  
- For Method: \( 193.5 > 5.14 \) → Reject \( H_0^{Method} \)

---

**d) Interpretation**  
Both factors significantly affect student performance because the calculated F-values are much greater than the critical values. Therefore, we conclude that **both teaching method and subject have a significant effect on average marks** at the 5% level of significance.
