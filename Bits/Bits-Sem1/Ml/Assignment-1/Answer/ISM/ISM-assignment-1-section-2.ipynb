## Question 1: Probability of liking at least one drink

Using the **Inclusion-Exclusion Principle**:

$$P(T \cup C \cup J) = P(T) + P(C) + P(J) - P(T \cap C) - P(C \cap J) - P(T \cap J) + P(T \cap C \cap J)$$

**Given:**
- P(T) = 0.50, P(C) = 0.40, P(J) = 0.30
- P(T∩C) = 0.20, P(C∩J) = 0.15, P(T∩J) = 0.10
- P(T∩C∩J) = 0.05

**Substituting:**

$$P(T \cup C \cup J) = 0.50 + 0.40 + 0.30 - 0.20 - 0.15 - 0.10 + 0.05$$

$$= 1.20 - 0.45 + 0.05 = \boxed{0.80}$$

**∴ The probability that a person likes at least one drink = 0.80 (80%)**

## Question 2: Binomial Distribution (p = 1/3, n = 6, q = 2/3)

The binomial formula: $P(X = r) = \binom{n}{r} p^r q^{n-r}$

### i) Exactly once (r = 1):

$$P(X=1) = \binom{6}{1}\left(\frac{1}{3}\right)^1\left(\frac{2}{3}\right)^5 = 6 \times \frac{1}{3} \times \frac{32}{243}$$

$$= \frac{192}{729} = \boxed{\frac{64}{243} \approx 0.2634}$$

### ii) At least once:

$$P(X \geq 1) = 1 - P(X = 0)$$

$$P(X=0) = \binom{6}{0}\left(\frac{1}{3}\right)^0\left(\frac{2}{3}\right)^6 = 1 \times 1 \times \frac{64}{729} = \frac{64}{729}$$

$$P(X \geq 1) = 1 - \frac{64}{729} = \boxed{\frac{665}{729} \approx 0.9121}$$

### iii) At most 5 times:

$$P(X \leq 5) = 1 - P(X = 6)$$

$$P(X=6) = \binom{6}{6}\left(\frac{1}{3}\right)^6\left(\frac{2}{3}\right)^0 = \frac{1}{729}$$

$$P(X \leq 5) = 1 - \frac{1}{729} = \boxed{\frac{728}{729} \approx 0.9986}$$

## Question 3: Bayes' Theorem — Scholarship Problem

**Given prior probabilities:**
- P(Urban) = 0.50, P(Rural) = 0.30, P(International) = 0.20

**Missing info assumption:** The question doesn't give scholarship rates per category, so we assume **equal likelihood** of receiving a scholarship across all groups — meaning we work purely with the given proportions as the probability distribution among scholarship recipients.

> *(If scholarship probabilities per category were given, Bayes' theorem would be applied differently.)*

### a) P(Rural | Scholarship):

$$P(\text{Rural} | \text{Scholarship}) = \frac{0.30}{0.50 + 0.30 + 0.20} = \frac{0.30}{1.00} = \boxed{0.30}$$

### b) P(Urban | Scholarship):

$$P(\text{Urban} | \text{Scholarship}) = \frac{0.50}{1.00} = \boxed{0.50}$$

### c) Which category benefits most?

Urban students have the **highest probability = 0.50**, so **Urban applicants benefit the most** from scholarships in absolute numbers.

### d) Fairness Interpretation:

The scholarship distribution **mirrors the applicant pool proportions** (50:30:20). This suggests the selection process is **proportionally fair** — no category is over- or under-represented relative to their application share. However, if the goal is to support **underrepresented or disadvantaged groups** (rural/international), the process may need **affirmative weighting** to be considered equitable.

### e) P(Rural or International | Scholarship):

Since Rural and International are **mutually exclusive** categories:

$$P(\text{Rural} \cup \text{International}) = P(\text{Rural}) + P(\text{International})$$

$$= 0.30 + 0.20 = \boxed{0.50}$$

**∴ There is a 50% chance that a randomly chosen scholarship recipient is either rural or international.**
