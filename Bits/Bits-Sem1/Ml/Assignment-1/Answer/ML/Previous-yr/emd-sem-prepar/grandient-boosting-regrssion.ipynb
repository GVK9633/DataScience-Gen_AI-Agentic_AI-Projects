## Concept: Gradient Boosting Regression

**One-line meaning:** Gradient Boosting builds a prediction model **step-by-step**, where each new weak learner tries to correct the errors (residuals) made by the current model.

**Why important:** It improves prediction by focusing each new tree on the **remaining errors**.

For this question, remember the flow:

$$
\boxed{\text{Mean} \rightarrow \text{Residuals} \rightarrow \text{Tree leaves} \rightarrow \text{Update prediction}}
$$

---

# (a) Compute \(F_0\) and pseudo-residuals

### Step 1: Find initial prediction \(F_0\)

The question says:

> \(F_0\) = mean of all actual \(y\)

Actual delivery times:

$$
34,\ 52,\ 28,\ 46,\ 40
$$

Add them:

$$
34+52+28+46+40=200
$$

There are 5 jobs:

$$
F_0=\frac{200}{5}
$$

$$
\boxed{F_0=40}
$$

So initially, the model predicts **40 minutes for every job**.

---

### Step 2: Calculate residuals

Formula given:

$$
\boxed{r_i=y_i-F_0(x_i)}
$$

Since \(F_0=40\):

### J1

Actual:

$$
y=34
$$

$$
r_1=34-40=\boxed{-6}
$$

### J2

$$
r_2=52-40=\boxed{12}
$$

### J3

$$
r_3=28-40=\boxed{-12}
$$

### J4

$$
r_4=46-40=\boxed{6}
$$

### J5

$$
r_5=40-40=\boxed{0}
$$

### Final table

| Job | Actual \(y\) | \(F_0\) | Residual \(r=y-F_0\) |
| --- | -----------: | ------: | -------------------: |
| J1  |           34 |      40 |               **-6** |
| J2  |           52 |      40 |               **12** |
| J3  |           28 |      40 |              **-12** |
| J4  |           46 |      40 |                **6** |
| J5  |           40 |      40 |                **0** |

---

# (b) Compute leaf predictions

The tree says:

### Rule 1

If:

$$
Traffic=Low
$$

→ Leaf \(L_A\)

Only **J3 and J4** have Low traffic.

Their residuals:

$$
J3=-12
$$

$$
J4=6
$$

Leaf prediction = average residual:

$$
L_A=\frac{-12+6}{2}
$$

$$
=\frac{-6}{2}
$$

$$
\boxed{L_A=-3}
$$

---

### Rule 2: High traffic and Distance \(\leq4\)

→ Leaf \(L_B\)

High traffic jobs:

* J1: High, distance 3
* J2: High, distance 6
* J5: High, distance 5

Only J1 has distance \(\leq4\).

Therefore:

$$
L_B=r_1=-6
$$

$$
\boxed{L_B=-6}
$$

---

### Rule 3: High traffic and Distance \(>4\)

→ Leaf \(L_C\)

J2 and J5:

* J2: residual \(12\)
* J5: residual \(0\)

Average:

$$
L_C=\frac{12+0}{2}
$$

$$
\boxed{L_C=6}
$$

### Final leaf values

$$
\boxed{L_A=-3}
$$

$$
\boxed{L_B=-6}
$$

$$
\boxed{L_C=6}
$$

---

# (c) Updated prediction and new residual for J2

J2 has:

* Traffic = **High**
* Distance = **6**

Since:

$$
6>4
$$

J2 goes to:

$$
\boxed{L_C}
$$

and:

$$
L_C=6
$$

The learning rate is:

$$
\eta=0.2
$$

The updated model is:

$$
\boxed{F_1(x)=F_0(x)+\eta h_1(x)}
$$

For J2:

$$
F_1(J2)=40+(0.2)(6)
$$

$$
=40+1.2
$$

$$
\boxed{F_1(J2)=41.2}
$$

---

### New residual for J2

Actual J2 time:

$$
y=52
$$

New prediction:

$$
F_1=41.2
$$

Residual:

$$
r_{\text{new}}=y-F_1
$$

$$
=52-41.2
$$

$$
\boxed{r_{\text{new}}=10.8}
$$

---

# ⭐ Final Answers

### (a)

$$
\boxed{F_0=40}
$$

Residuals:

$$
\boxed{[-6,\ 12,\ -12,\ 6,\ 0]}
$$

### (b)

$$
\boxed{L_A=-3}
$$

$$
\boxed{L_B=-6}
$$

$$
\boxed{L_C=6}
$$

### (c)

For J2:

$$
\boxed{F_1(J2)=41.2}
$$

$$
\boxed{\text{New residual}=10.8}
$$

### Easy exam memory trick

$$
\boxed{\text{Initial prediction = Mean}}
$$

$$
\boxed{\text{Residual = Actual - Prediction}}
$$

$$
\boxed{\text{Leaf value = Average residual}}
$$

$$
\boxed{\text{New prediction = Old prediction + Learning rate × Leaf}}
$$
