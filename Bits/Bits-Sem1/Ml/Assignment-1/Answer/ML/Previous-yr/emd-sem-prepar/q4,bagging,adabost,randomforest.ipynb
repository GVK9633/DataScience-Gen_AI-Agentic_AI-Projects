This question is about **ensemble learning**, especially the difference between **Bagging, Random Forest, and AdaBoost**. Let me explain it in simple language so you can answer the exam.

## (a) Bagging vs AdaBoost

### 1. Bagging — Bootstrap Aggregating

Imagine you have **100 decision trees**.

Bagging creates different training datasets by randomly sampling the original data **with replacement**.

Example:

```text
Original data: 20,000 customers

        ↓
   Random sampling
   with replacement
        ↓
 ┌──────┬──────┬──────┐
Tree 1  Tree 2  Tree 3 ... Tree 100
```

Each tree sees a **different sample of customers**.

So the main source of diversity is:

$$
\boxed{\text{Data-level randomness}}
$$

And the trees are generally trained **independently/in parallel**.

### 2. AdaBoost

AdaBoost works differently.

It trains weak learners **one after another**.

```text
Tree 1
  ↓
Find mistakes
  ↓
Give more weight to mistakes
  ↓
Tree 2
  ↓
Find remaining mistakes
  ↓
Give more weight
  ↓
Tree 3 ...
```

So if Tree 1 incorrectly predicts some customers, AdaBoost gives those customers **more importance** for the next learner.

Therefore:

$$
\boxed{\text{AdaBoost = error-driven focus}}
$$

### ⭐ Main difference

| Bagging                            | AdaBoost                              |
| ---------------------------------- | ------------------------------------- |
| Randomly samples training data     | Focuses on previous mistakes          |
| Models trained independently       | Models trained sequentially           |
| Diversity comes from data sampling | Diversity comes from error correction |
| Reduces variance                   | Mainly reduces bias                   |

**Exam answer:**

> Bagging creates diversity through **random bootstrap sampling of training data**, whereas AdaBoost creates diversity through **sequential, error-driven reweighting of misclassified examples**.

# (b) Bagging vs Random Forest

Both use randomness, but Random Forest adds **another level of randomness**.

### Bagging

Different trees get different bootstrap samples:

$$
\boxed{\text{Randomness = training data}}
$$

### Random Forest

Random Forest does:

1. Random bootstrap samples of data
2. Randomly selects a subset of **features** at each split

For example, suppose there are 30 features:

```text
30 features
     ↓
Randomly choose some features
     ↓
Decision Tree split
```

Another tree may choose a different subset.

Therefore:

$$
\boxed{\text{Random Forest = random data + random features}}
$$

This reduces correlation between trees and can improve generalization.

### ⭐ Exam answer

> Random Forest introduces additional randomness by randomly selecting a subset of features at each split, in addition to bootstrap sampling of data used by standard Bagging. This makes trees less correlated and improves generalization.

# (c) Majority voting

There are **3 independent classifiers**.

Each has:

$$
P(\text{correct})=0.7
$$

$$
P(\text{wrong})=0.3
$$

For the ensemble to be correct, at least **2 out of 3** classifiers must be correct.

Therefore:

$$
P(\text{ensemble correct})
=
P(2\text{ correct})+P(3\text{ correct})
$$

### Exactly 2 correct

$$
\binom32(0.7)^2(0.3)
$$

$$
=3(0.49)(0.3)
$$

$$
=0.441
$$

### All 3 correct

$$
(0.7)^3
$$

$$
=0.343
$$

### Add them

$$
0.441+0.343
$$

$$
\boxed{0.784}
$$

So ensemble accuracy is:

$$
\boxed{78.4\%}
$$

Individual classifier accuracy:

$$
70\%
$$

Therefore:

$$
\boxed{78.4\%>70\%}
$$

### Conclusion

Yes, the ensemble is **better than an individual classifier**, assuming the classifiers' predictions are independent.
