## 📚 Key Concepts (1-2 lines for quick revision)

| Concept | Explanation |
|---------|-------------|
| **GMM E-step** | Computes **responsibilities** – probability that each point belongs to each component, using Bayes' theorem. |
| **GMM M-step** | Updates **parameters** (π, μ, σ²) using weighted averages based on responsibilities. |
| **Log-likelihood** | Measures how well the model fits the data. Higher = better fit. |
| **K-Means** | Hard clustering – each point assigned to **one** nearest centroid. GMM gives **soft** probabilities. |

## 📐 Important Formulas

**Gaussian Density:**
\[
N(x|\mu,\sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)
\]

**Responsibility (E-step):**
\[
\gamma_{ik} = \frac{\pi_k \cdot N(x_i|\mu_k,\sigma_k^2)}{\sum_{j=1}^{2} \pi_j \cdot N(x_i|\mu_j,\sigma_j^2)}
\]

**Log-likelihood:**
\[
\ell = \sum_{i=1}^{n} \log\left(\sum_{k=1}^{2} \pi_k \cdot N(x_i|\mu_k,\sigma_k^2)\right)
\]

**M-step Updates:**
\[
\pi_k = \frac{\sum_{i=1}^{n} \gamma_{ik}}{n}, \quad
\mu_k = \frac{\sum_{i=1}^{n} \gamma_{ik} \cdot x_i}{\sum_{i=1}^{n} \gamma_{ik}}, \quad
\sigma_k^2 = \frac{\sum_{i=1}^{n} \gamma_{ik} \cdot (x_i - \mu_k)^2}{\sum_{i=1}^{n} \gamma_{ik}}
\]

## ✅ (c) M-step: Updated Parameters

### Step 1: Compute sums of responsibilities

\[

### Step 2: Updated mixing weights

\[

\[

### Step 4: Updated variances

## ✅ (d) K-Means with k=2

### Step 1: Initial centroids
\[
c_1 = -3, \quad c_2 = 2
\]

### Step 2: Assign each point to nearest centroid

| \(x_i\) | Distance to \(c_1=-3\) | Distance to \(c_2=2\) | Assigned Cluster |
|---------|----------------------|----------------------|------------------|
| -4 | 1 | 6 | **1** |
| -2 | 1 | 4 | **1** |
| 0 | 3 | 2 | **2** |
| 3 | 6 | 1 | **2** |

### Step 3: Update centroids (mean of assigned points)

**Cluster 1:** Points \{-4, -2\}

\[
c_1 = \frac{-4 + (-2)}{2} = \frac{-6}{2} = \boxed{-3.0}
\]

**Cluster 2:** Points \{0, 3\}

\[
c_2 = \frac{0 + 3}{2} = \frac{3}{2} = \boxed{1.5}
\]

## 📊 Comparison: GMM Means vs K-Means Centroids

| Component | GMM Mean | K-Means Centroid |
|-----------|----------|------------------|
| 1 | -2.889 | -3.0 |
| 2 | 1.559 | 1.5 |

**Reason for difference:**  
GMM uses **soft assignments** (probabilities) – points like \(x=-2\) have 99.93% probability for Component 1 but still **0.07%** for Component 2, which pulls the mean slightly upward. K-Means uses **hard assignments** – each point belongs to **exactly one** cluster, so \(x=-2\) contributes fully to Cluster 1 only.
