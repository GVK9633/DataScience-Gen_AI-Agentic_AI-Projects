## Concept 1: Hypothesis Function \( h_\theta(x) \)

**What it is:** The prediction made by the model.

For **simple linear regression** (one input variable):
\[
h_\theta(x) = \theta_0 + \theta_1 x
\]

- \( \theta_0 \) = intercept (where line crosses y-axis)
- \( \theta_1 \) = slope (steepness of line)
- \( x \) = input feature

**Example from problem:**  
If \( \theta_0 = 0, \theta_1 = 1 \), then \( h_\theta(x) = 0 + 1 \cdot x = x \)

So for \( x = 1 \), prediction = 1; for \( x = 2 \), prediction = 2; etc.

---

## Concept 2: Cost Function \( J(\theta_0, \theta_1) \)

**What it is:** Measures how **bad** your model is. Lower cost = better model.

**Without regularization (standard MSE):**
\[
J(\theta_0, \theta_1) = \frac{1}{2m} \sum_{i=1}^m (h_\theta(x_i) - y_i)^2
\]

**With regularization (L2 regularization / Ridge):**
\[
J(\theta_0, \theta_1) = \frac{1}{2m} \sum_{i=1}^m (h_\theta(x_i) - y_i)^2 + \frac{\lambda}{2m} \theta_1^2
\]

| Term | Meaning |
|------|---------|
| \( m \) | Number of training examples (here, 3) |
| \( h_\theta(x_i) - y_i \) | Prediction error |
| \( \lambda \) | Regularization strength (here, 2) |
| \( \frac{\lambda}{2m} \theta_1^2 \) | Penalty for large \( \theta_1 \) |

**Note:** We usually **do NOT regularize \( \theta_0 \)** (the intercept), only \( \theta_1 \) and higher coefficients.

---

## Concept 3: Regularization Term

**Why use it:** Prevents overfitting by keeping coefficients small.

**Effect:**
- Large \( \lambda \) → strong penalty → coefficients shrink toward zero
- Small \( \lambda \) → weak penalty → almost like normal regression
- \( \lambda = 0 \) → standard linear regression

**In your problem:** \( \lambda = 2, m = 3 \), so \( \frac{\lambda}{2m} = \frac{2}{6} = \frac{1}{3} \)

Thus:
\[
J(\theta_0, \theta_1) = \frac{1}{6} \sum_{i=1}^3 (h_\theta(x_i) - y_i)^2 + \frac{1}{3} \theta_1^2
\]

---

## Concept 4: Partial Derivatives \( \frac{\partial J}{\partial \theta_0} \) and \( \frac{\partial J}{\partial \theta_1} \)

**What they are:** The slope (rate of change) of the cost function with respect to each parameter.

**Why needed:** Gradient descent uses them to know **how much** and **in which direction** to change \( \theta_0 \) and \( \theta_1 \).

### Derivative with respect to \( \theta_0 \):
\[
\frac{\partial J}{\partial \theta_0} = \frac{1}{m} \sum_{i=1}^m (h_\theta(x_i) - y_i)
\]

### Derivative with respect to \( \theta_1 \):
\[
\frac{\partial J}{\partial \theta_1} = \frac{1}{m} \sum_{i=1}^m (h_\theta(x_i) - y_i) x_i + \frac{\lambda}{m} \theta_1
\]

**Derivation logic:**
- Derivative of \( (h_\theta(x) - y)^2 \) is \( 2(h_\theta(x) - y) \times \frac{\partial h}{\partial \theta} \)
- \( \frac{\partial h}{\partial \theta_0} = 1 \), \( \frac{\partial h}{\partial \theta_1} = x \)
- The \( \frac{1}{2m} \) cancels the 2 from derivative
- For \( \theta_1 \), derivative of \( \frac{\lambda}{2m} \theta_1^2 \) is \( \frac{\lambda}{m} \theta_1 \)

---

## Concept 5: Gradient Descent Update Rule

**What it is:** The formula to update parameters repeatedly to reduce cost.

\[
\theta_0 := \theta_0 - \alpha \frac{\partial J}{\partial \theta_0}
\]
\[
\theta_1 := \theta_1 - \alpha \frac{\partial J}{\partial \theta_1}
\]

Where:
- \( := \) means "update to" (assignment)
- \( \alpha \) = learning rate (step size) = 0.05 in your problem

**Plugging the derivatives:**

\[
\theta_0 := \theta_0 - \alpha \cdot \frac{1}{m} \sum_{i=1}^m (h_\theta(x_i) - y_i)
\]

\[
\theta_1 := \theta_1 - \alpha \left[ \frac{1}{m} \sum_{i=1}^m (h_\theta(x_i) - y_i) x_i + \frac{\lambda}{m} \theta_1 \right]
\]

---

## Concept 6: Batch Gradient Descent

**What it is:** Uses **all training examples** to compute the gradient in each update.

**Steps:**
1. Compute predictions \( h_\theta(x_i) \) for all \( i \) using current \( \theta_0, \theta_1 \)
2. Calculate errors \( h_\theta(x_i) - y_i \) for all \( i \)
3. Sum them up (with or without multiplying by \( x_i \))
4. Update \( \theta_0 \) and \( \theta_1 \) using the formulas
5. Repeat until convergence

**In your problem:** You only do **1 iteration** (one update).

---

## Concept 7: Summation Notation \( \sum \)

**What it means:** Add up multiple terms.

\[
\sum_{i=1}^m (h_\theta(x_i) - y_i)
\]
Means: For i = 1, 2, 3, compute \( h_\theta(x_i) - y_i \) and add them together.

**Example from problem** (initial \( \theta_0=0, \theta_1=1 \)):

| i | \( x_i \) | \( y_i \) | \( h(x_i)=x_i \) | \( h - y \) |
|---|-----------|-----------|------------------|-------------|
| 1 | 1 | 2 | 1 | -1 |
| 2 | 2 | 5 | 2 | -3 |
| 3 | 3 | 7 | 3 | -4 |
| **Sum** | | | | **-8** |

So \( \sum (h - y) = -8 \)

Similarly \( \sum (h - y) x_i = (-1 \times 1) + (-3 \times 2) + (-4 \times 3) = -1 -6 -12 = -19 \)

## Concept 8: Substituting Values into Formulas

**Step-by-step process for one iteration:**

### Step A: Calculate sums using current \( \theta_0, \theta_1 \)
- \( \text{sum1} = \sum (h - y) \)
- \( \text{sum2} = \sum (h - y) x_i \)

### Step B: Update \( \theta_0 \)
\[
\theta_0^{\text{new}} = \theta_0^{\text{old}} - \alpha \times \frac{\text{sum1}}{m}
\]

\[
\theta_1^{\text{new}} = \theta_1^{\text{old}} - \alpha \left( \frac{\text{sum2}}{m} + \frac{\lambda}{m} \theta_1^{\text{old}} \right)
