## What is Entropy?

**Entropy** measures the **impurity** or **uncertainty** in a dataset.

- **High entropy** = mixed labels (uncertain, impure)
- **Low entropy** = same labels (certain, pure)
- **Zero entropy** = all labels identical

### Simple analogy:

| Situation | Entropy | Example |
|-----------|---------|---------|
| All repay | **0** | [Repay, Repay, Repay] → Certain |
| Half-half | **1** | [Repay, Default, Repay, Default] → Maximum uncertainty |
| Mixed but skewed | Between 0 and 1 | [Repay, Repay, Repay, Default] → Less uncertain |

---

## The Entropy Formula

\[
H(S) = -\sum_{i=1}^{c} p_i \log_2(p_i)
\]

Where:
- \(c\) = number of classes (here: Repay, Default → 2 classes)
- \(p_i\) = proportion of samples in class \(i\)

### For 2 classes (like your problem):

\[
H(S) = -p_{\text{Repay}} \log_2(p_{\text{Repay}}) - p_{\text{Default}} \log_2(p_{\text{Default}})
\]

---

## Common Entropy Values to Remember

| Proportion (p, 1-p) | Entropy | Meaning |
|---------------------|---------|---------|
| (1.0, 0) or (0, 1.0) | 0 | Pure node |
| (0.9, 0.1) | 0.469 | Mostly pure |
| (0.8, 0.2) | 0.722 | |
| (0.75, 0.25) | 0.811 | |
| (0.67, 0.33) | 0.918 | |
| (0.6, 0.4) | 0.971 | |
| (0.5, 0.5) | **1.0** | Maximum impurity |

---

## Concepts You Must Know to Solve the Question

### 1. **Parent Entropy**
- Entropy of the **entire dataset before splitting**
- Formula: \(H(S) = -p_R \log_2(p_R) - p_D \log_2(p_D)\)

### 2. **Child Entropy**
- Entropy of each **subset** after splitting on an attribute
- Calculate separately for each branch

### 3. **Weighted Entropy**
- Average of child entropies, weighted by the **number of samples** in each child
- Formula: \(H_{\text{weighted}} = \frac{n_1}{n} H_1 + \frac{n_2}{n} H_2 + \dots\)

### 4. **Information Gain**
- How much entropy is **reduced** by the split
- Formula: \(IG = H(S) - H_{\text{weighted}}\)
- **Higher IG = Better attribute**

### 5. **Logarithms base 2**
- You need to know how to compute \(\log_2(x)\)
- Either memorize common values or use: \(\log_2(x) = \frac{\log(x)}{\log(2)}\)

## Common Mistakes to Avoid

| Mistake | Why it's wrong |
|---------|----------------|
| Using natural log (ln) instead of log₂ | Formula specifically uses log₂ for bits |
| Forgetting to weight child entropies | Each child's contribution depends on its size |
| Using training error instead of entropy | These are different concepts |
| Saying IG = 0 means bad | IG = 0 means attribute gives no new information |

## Final Tip for Exam

> **Entropy = messiness (0 = clean, 1 = completely mixed for 2 classes)**  
> **Information Gain = how much cleaner the split makes things**  
> **Higher IG = Better split**

## Step 1: Parent Entropy Calculation (1.5 marks)

**Formula:**  
\[
H(S) = - p_{\text{Repay}} \log_2(p_{\text{Repay}}) - p_{\text{Default}} \log_2(p_{\text{Default}})
\]

**Given:**
- Total samples = 14
- Repay = 9, Default = 5

**Probabilities:**
\[
p_{\text{Repay}} = \frac{9}{14}, \quad p_{\text{Default}} = \frac{5}{14}
\]

**Calculation:**
\[
H(S) = -\frac{9}{14} \log_2\left(\frac{9}{14}\right) - \frac{5}{14} \log_2\left(\frac{5}{14}\right)
\]

\[
\frac{9}{14} = 0.643, \quad \frac{5}{14} = 0.357
\]

\[
\log_2(0.643) = \frac{\log(0.643)}{\log(2)} \approx \frac{-0.441}{0.301} \approx -1.465
\]
\[
\log_2(0.357) = \frac{\log(0.357)}{\log(2)} \approx \frac{-0.447}{0.301} \approx -1.485
\]

\[
H(S) = -[0.643 \times (-1.465)] - [0.357 \times (-1.485)]
\]
\[
H(S) = -[-0.942] - [-0.530]
\]
\[
H(S) = 0.942 + 0.530 = 1.472
\]

**Parent Entropy:**
\[
\boxed{H(S) = 1.472 \text{ bits}}
\]

---

## Step 2: Child Entropy Calculations (2 marks)

### Child 1: Income = High

- Repay = 6, Default = 2
- Total = 8

\[
p_{\text{Repay}} = \frac{6}{8} = 0.75, \quad p_{\text{Default}} = \frac{2}{8} = 0.25
\]

\[
H(\text{High}) = -0.75 \log_2(0.75) - 0.25 \log_2(0.25)
\]

\[
\log_2(0.75) = \frac{\log(0.75)}{\log(2)} \approx \frac{-0.125}{0.301} \approx -0.415
\]
\[
\log_2(0.25) = -2
\]

\[
H(\text{High}) = -[0.75 \times (-0.415)] - [0.25 \times (-2)]
\]
\[
H(\text{High}) = -[-0.311] - [-0.5]
\]
\[
H(\text{High}) = 0.311 + 0.5 = 0.811
\]

**Child 1 Entropy:** \(\boxed{0.811 \text{ bits}}\)

---

### Child 2: Income = Low

- Repay = 3, Default = 3
- Total = 6

\[
p_{\text{Repay}} = \frac{3}{6} = 0.5, \quad p_{\text{Default}} = 0.5
\]

\[
H(\text{Low}) = -0.5 \log_2(0.5) - 0.5 \log_2(0.5)
\]
\[
\log_2(0.5) = -1
\]
\[
H(\text{Low}) = -[0.5 \times (-1)] - [0.5 \times (-1)] = -[-0.5] - [-0.5]
\]
\[
H(\text{Low}) = 0.5 + 0.5 = 1.0
\]

**Child 2 Entropy:** \(\boxed{1.0 \text{ bits}}\)

---

## Step 3: Weighted Entropy Calculation (1 mark)

**Formula:**
\[
H_{\text{weighted}} = \frac{|S_{\text{High}}|}{|S|} \times H(\text{High}) + \frac{|S_{\text{Low}}|}{|S|} \times H(\text{Low})
\]

\[
|S_{\text{High}}| = 8, \quad |S_{\text{Low}}| = 6, \quad |S| = 14
\]

\[
H_{\text{weighted}} = \frac{8}{14} \times 0.811 + \frac{6}{14} \times 1.0
\]

\[
\frac{8}{14} = 0.571, \quad \frac{6}{14} = 0.429
\]

\[
H_{\text{weighted}} = 0.571 \times 0.811 + 0.429 \times 1.0
\]
\[
H_{\text{weighted}} = 0.463 + 0.429 = 0.892
\]

**Weighted Entropy:** \(\boxed{0.892 \text{ bits}}\)

---

## Step 4: Information Gain Calculation (1 mark)

**Formula:**
\[
IG = H(S) - H_{\text{weighted}}
\]

\[
IG = 1.472 - 0.892 = 0.58
\]

**Information Gain:** \(\boxed{0.58 \text{ bits}}\)

---

## Step 5: Interpretation (0.5 mark)

**Answer:**  
Yes, **Income Level is a good attribute to split on**.  
The information gain is **0.58 bits**, which is reasonably high (more than half the parent entropy). The split creates a purer child node for "High" income (entropy 0.811) while keeping "Low" income at maximum entropy (1.0). This split will help classify loan repayment better than no split.

## Similar Question

Using **entropy-based information gain**, determine whether **Contract Type** is a good attribute to split on.

## Step 1: Parent Entropy Calculation (1.5 marks)

**Formula:**  
\[
H(S) = -p_{\text{Stay}} \log_2(p_{\text{Stay}}) - p_{\text{Churn}} \log_2(p_{\text{Churn}})
\]

**Given:**  
- Total = 20, Stay = 12, Churn = 8  

\[
p_{\text{Stay}} = \frac{12}{20} = 0.6, \quad p_{\text{Churn}} = \frac{8}{20} = 0.4
\]

**Calculation:**  
\[
\log_2(0.6) = \frac{\log(0.6)}{\log(2)} \approx \frac{-0.222}{0.301} \approx -0.737
\]  
\[
\log_2(0.4) = \frac{\log(0.4)}{\log(2)} \approx \frac{-0.398}{0.301} \approx -1.322
\]

\[
H(S) = -[0.6 \times (-0.737)] - [0.4 \times (-1.322)]
\]  
\[
H(S) = -[-0.442] - [-0.529]
\]  
\[
H(S) = 0.442 + 0.529 = 0.971
\]

**Parent Entropy:**  
\[
\boxed{H(S) = 0.971 \text{ bits}}
\]

## Step 2: Child Entropy Calculations (2 marks)

### Child 1: Contract Type = Monthly

- Stay = 6, Churn = 6, Total = 12  

\[
p_{\text{Stay}} = \frac{6}{12} = 0.5, \quad p_{\text{Churn}} = 0.5
\]  

\[
H(\text{Monthly}) = -0.5 \log_2(0.5) - 0.5 \log_2(0.5)
\]  
\[
\log_2(0.5) = -1
\]  
\[
H(\text{Monthly}) = -[0.5 \times (-1)] - [0.5 \times (-1)] = 0.5 + 0.5 = 1.0
\]

**Child 1 Entropy:** \(\boxed{1.0 \text{ bits}}\)

---

### Child 2: Contract Type = Yearly

- Stay = 6, Churn = 2, Total = 8  

\[
p_{\text{Stay}} = \frac{6}{8} = 0.75, \quad p_{\text{Churn}} = \frac{2}{8} = 0.25
\]  

\[
\log_2(0.75) \approx -0.415, \quad \log_2(0.25) = -2
\]  

\[
H(\text{Yearly}) = -[0.75 \times (-0.415)] - [0.25 \times (-2)]
\]  
\[
H(\text{Yearly}) = -[-0.311] - [-0.5] = 0.311 + 0.5 = 0.811
\]

**Child 2 Entropy:** \(\boxed{0.811 \text{ bits}}\)

## Step 3: Weighted Entropy Calculation (1 mark)

**Formula:**  
\[
H_{\text{weighted}} = \frac{|\text{Monthly}|}{|\text{Total}|} \times H(\text{Monthly}) + \frac{|\text{Yearly}|}{|\text{Total}|} \times H(\text{Yearly})
\]

\[
H_{\text{weighted}} = \frac{12}{20} \times 1.0 + \frac{8}{20} \times 0.811

\[
\frac{12}{20} = 0.6, \quad \frac{8}{20} = 0.4
\]  
\[
H_{\text{weighted}} = 0.6 \times 1.0 + 0.4 \times 0.811 = 0.6 + 0.3244 = 0.9244
\]

**Weighted Entropy:** \(\boxed{0.924 \text{ bits}}\)

## Step 4: Information Gain Calculation (1 mark)

**Formula:**  
\[
IG = H(S) - H_{\text{weighted}}
\]  

\[
IG = 0.971 - 0.924 = 0.047
\]

**Information Gain:** \(\boxed{0.047 \text{ bits}}\)

## Step 5: Interpretation (0.5 mark)

**Answer:**  
**No, Contract Type is NOT a good attribute to split on.**  

The information gain is only **0.047 bits**, which is very close to zero. This means the split does not significantly reduce impurity. The Monthly branch is already at maximum entropy (1.0), and the Yearly branch is only slightly purer. Splitting on Contract Type provides almost no useful information for predicting churn.
