Here’s a **3‑to‑5 line explanation** of intercept, slope, and coefficient:

- **Intercept** (\( \beta_0 \)): The value of \( y \) when all input variables (\( x \)) are zero. It’s the starting point or baseline.

- **Slope** (\( \beta_1 \)): In **simple linear regression** (only one \( x \)), the slope tells how much \( y \) changes when \( x \) increases by 1 unit. It represents the steepness of the straight line.

- **Coefficient** ( \( \beta_1, \beta_2, … \) ): A general term for any multiplier attached to an input variable. In multiple or polynomial regression, you have many coefficients; the slope is just one special case of a coefficient.

**Short memory aid:** Intercept = starting value, Slope = steepness (for one \( x \)), Coefficient = any multiplier (including slope).

## Linear Regression vs. Polynomial Regression

| Feature | **Linear Regression** | **Polynomial Regression** |
|---------|----------------------|---------------------------|
| **Definition** | A statistical method that models the relationship between dependent variable \(y\) and independent variable \(x\) by fitting a **straight line** to the data. | A special case of linear regression where the relationship between \(y\) and \(x\) is modeled as an **\(n\)-th degree polynomial** (a curved line). |
| **Equation form** | \( y = \beta_0 + \beta_1 x \) | \( y = \beta_0 + \beta_1 x + \beta_2 x^2 + \dots + \beta_n x^n \) |
| **Degree** | Always **1** | **2 or higher** (2, 3, 10, etc.) |
| **Shape of curve** | Straight line 📏 | Curved line (bends) 〰️ |
| **Flexibility** | Low (high bias) | High (low bias) |
| **Risk** | Underfitting (too simple) | Overfitting (too complex) |
| **Training error** | Higher (if data is curved) | Lower (fits more closely) |
| **Validation error** | Often stable | Increases after optimal degree |
| **Number of parameters** | 2 (intercept + slope) | \(n+1\) (more coefficients) |
| **When to use** | When relationship is roughly linear | When relationship is curved / non‑linear |

---

## Quick one‑line definitions (to memorize)

- **Linear Regression:** Fits a straight line to model the relationship between \(x\) and \(y\).  
- **Polynomial Regression:** Fits a curve (degree ≥ 2) to model non‑linear relationships.
