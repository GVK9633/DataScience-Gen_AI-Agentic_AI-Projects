```python
import numpy as np

def train_linear(X, y, lr=0.01, epochs=100):
    N, d = X.shape
    weights = np.zeros((d, 2))
    
    for i in range(epochs):
        y_pred = X @ weights
        error = y_pred - y  
        trsns = X.T
        mult = trsns @ error
        gradient = (2/N) * (X.T @ error)
        weights = weights - lr * gradient
    
    return weights

# ===== MINIMAL EXAMPLE =====
# Create tiny dataset
X = np.array([[1, 2], [3, 4], [5, 6]])  # 3 samples, 2 features
y = np.array([[5], [11], [17]])          # y = 2*x1 + 3*x2

# CALL THE FUNCTION
weights = train_linear(X, y, lr=0.01, epochs=500)

# Show results
print("Learned weights:", weights.flatten())
print("Expected: [2, 3]")
```

### Exam Shortcut

| Can a Single Perceptron Implement It? |
| ------------------------------------- |
| AND ✅                                 |
| OR ✅                                  |
| NOT ✅                                 |
| NAND ✅                                |
| NOR ✅                                 |
| XOR ❌                                 |
| XNOR ❌                                |

**Remember:**

* **AND / OR / NAND / NOR / NOT** → **Linearly separable** → **Single perceptron**.
* **XOR / XNOR** → **Not linearly separable** → **Require a multi-layer perceptron (hidden layer)**.

Activation functions decide **whether a neuron should be activated**. They take the neuron's input

[
z = w_1x_1 + w_2x_2 + b
]

and produce the output

[
y = f(z)
]

Without an activation function, a neural network behaves like a simple linear model and cannot learn complex patterns.

---

# Activation Functions Comparison

| Activation Function | Formula                                  | Output Range | Used For                  |
| ------------------- | ---------------------------------------- | ------------ | ------------------------- |
| Sigmoid             | (\sigma(z)=\frac{1}{1+e^{-z}})           | 0 to 1       | Binary classification     |
| Tanh                | (\tanh(z)=\frac{e^z-e^{-z}}{e^z+e^{-z}}) | -1 to +1     | Hidden layers             |
| ReLU                | (f(z)=\max(0,z))                         | 0 to ∞       | Most deep learning models |

---

# 1. Sigmoid Activation Function

## Formula

[
\sigma(z)=\frac{1}{1+e^{-z}}
]

## Output Range

[
0 \le y \le 1
]

It converts any real number into a value between **0 and 1**.

### Small Example

Suppose

[
z=2
]

[
\sigma(2)=\frac{1}{1+e^{-2}}
]

Since

[
e^{-2}\approx0.135
]

[
\sigma(2)=\frac{1}{1+0.135}
=\frac{1}{1.135}
\approx0.881
]

**Output = 0.881**

Another example:

If

[
z=-2
]

[
\sigma(-2)=\frac{1}{1+e^2}
]

[
=\frac{1}{1+7.389}
\approx0.119
]

|  2 |          0.881 |

### Use

* Binary classification
* Probability prediction

# 2. Tanh Activation Function

## Formula

[
\tanh(z)=\frac{e^z-e^{-z}}{e^z+e^{-z}}
]

## Output Range

[
-1 \le y \le 1
]

Unlike Sigmoid, Tanh outputs **negative and positive values**.

### Small Example

Suppose

[
z=1
]

[
\tanh(1)\approx0.762
]

If

[
z=-1
]

[
\tanh(-1)\approx-0.762
]

### Use

* Hidden layers
* Data centered around zero

# 3. ReLU (Rectified Linear Unit)

## Formula

[
f(z)=
\begin{cases}
0,&z<0\
z,&z\ge0
\end{cases}
]

## Output Range

[
0 \text{ to } \infty
]

It simply removes negative values.

### Small Example

If

[
z=-4
]

Output

[
0
]

If

[
z=5
]

Output

[
5
]

### Use

* Hidden layers in deep neural networks
* Fast training
* Most commonly used activation function

# Advantages and Disadvantages

| Activation  | Advantages                                                    | Disadvantages                                                                     |
| ----------- | ------------------------------------------------------------- | --------------------------------------------------------------------------------- |
| **Sigmoid** | Outputs probabilities (0–1), good for binary classification   | Can suffer from vanishing gradients for large positive/negative inputs            |
| **Tanh**    | Output is centered around 0, often trains better than Sigmoid | Also suffers from vanishing gradients                                             |
| **ReLU**    | Simple, fast, and helps deep networks train efficiently       | Neurons can become inactive if they always receive negative inputs ("dying ReLU") |

---

# Exam Memory Trick

| Activation  | Output Range | Best Used For                                       |
| ----------- | ------------ | --------------------------------------------------- |
| **Sigmoid** | **0 to 1**   | Binary classification (output layer)                |
| **Tanh**    | **-1 to 1**  | Hidden layers when zero-centered outputs are useful |
| **ReLU**    | **0 to ∞**   | Hidden layers in most deep learning models          |

### Easy way to remember

* **Sigmoid** → Looks like an **S-shaped curve**, gives values between **0 and 1**.
* **Tanh** → Similar to Sigmoid but centered at **0**, gives values between **-1 and 1**.
* **ReLU** → **Negative becomes 0; positive stays the same**. It is the most commonly used activation function in modern deep learning.
