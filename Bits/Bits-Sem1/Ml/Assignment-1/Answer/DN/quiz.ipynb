Q1. Which of the following statements about the perceptron convergence theorem is TRUE?
ANS: c. Perceptrons are guaranteed to converge if the data is linearly separable

Q2. Assertion (A): Linear classifier w=[1,-2], bias=0 has positive slope. Reason (R): slope = -w1/w2
ANS: d. Both A and R are true, and R is the correct explanation of A

Q3. Why is factor 1/2 included in squared error loss function?
ANS: a. To simplify the derivative during gradient computation

Q4. Which dataset would a single perceptron be UNABLE to classify correctly?
ANS: d. Points arranged in two concentric circles

Q5. Precision focuses on minimizing:
ANS: b. False positives

Q6. Features on vastly different scales — what problem arises during gradient descent?
ANS: c. Some weight updates dominate others, causing slow or erratic convergence

Q7. Training method commonly used to optimize weights in a linear neural network for regression:
ANS: d. Batch Gradient Descent (GD)

Q8. What does weighted sum w1*x1 + w2*x2 + b represent geometrically?
ANS: a. A hyperplane in the input space

Q9. Which dataset split is used to tune hyperparameters?
ANS: a. Validation

Q10. w1=2, w2=-1, w0=3 — which point lies on the positive side of the decision boundary?
ANS: a. (0,0) [gives: 3 > 0, positive side]

Q11. Perceptron with w1=-1, w0=0.5, input (1,0) — what is the output?
ANS: a. 0 (since -1x1 + 0.5 = -0.5 < 0)

Q12. In a confusion matrix, what does "False Positive" represent?
ANS: b. Predicted class 1, actual class 0

Q13. Sigmoid derivative o'(z) = o(z)(1-o(z)) — maximum value and at what z?
ANS: a. Maximum is 0.25 at z = 0

Q14. The bias term w_0 in a single neuron linear regression model allows the model to:
ANS: c. Shift the hyperplane away from the origin

Q15. Core difference between Perceptron (classification) and linear regression neuron (regression):
ANS: d. Uses an identity activation to output a continuous value

Q16. Supervised Learning task that answers "how much or how many" questions:
ANS: a. Regression

Q17. Which is an example of a regression problem?
ANS: d. House Price Prediction

Q18. In z = w^T x + b for logistic regression, what is 'z' called?
ANS: d. The logit or pre-activation

Q19. Dataset: (x=1,y=2),(x=2,y=4),(x=3,y=5). GD with lr=0.1, initial weights=0. Value of w1 after first iteration?
ANS: d. 0.833

Q20. What determines the dimensionality in a dataset?
ANS: a. The number of features (d)
