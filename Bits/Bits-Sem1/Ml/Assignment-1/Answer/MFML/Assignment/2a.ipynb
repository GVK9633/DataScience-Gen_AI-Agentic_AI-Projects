```python
# ============================================================================
# Q2(a): LU DECOMPOSITION USING ELEMENTARY MATRICES (No linalg)
# ============================================================================

import numpy as np

print("="*80)
print("Q2(a): LU DECOMPOSITION USING ELEMENTARY MATRICES")
print("="*80)

# STEP 1: Generate SPD matrix
n = int(input("\nEnter size of matrix n: "))
np.random.seed(42)
B = np.random.randint(-5, 6, (n, n))
A = B.T @ B + n * np.eye(n)
A = (A + A.T) / 2

print("\n--- STEP 1: Matrix A (Symmetric Positive Definite) ---")
for row in A:
    print(" ", [round(x, 2) for x in row])

# STEP 2: Function to create elementary matrix
def elementary_matrix(n, row, col, multiplier):
    """E = I with -multiplier at (row, col)"""
    E = np.eye(n)
    E[row, col] = -multiplier
    return E

# STEP 3: Function to multiply matrices (without linalg)
def matrix_multiply(X, Y):
    """Multiply two matrices manually"""
    m, n = X.shape
    n, p = Y.shape
    result = np.zeros((m, p))
    for i in range(m):
        for j in range(p):
            for k in range(n):
                result[i, j] += X[i, k] * Y[k, j]
    return result

# STEP 4: Function to invert lower triangular matrix (without linalg)
def invert_lower_triangular(L):
    """Inverse of lower triangular matrix using forward substitution"""
    n = L.shape[0]
    L_inv = np.eye(n)
    
    for i in range(n):
        for j in range(i):
            sum_k = 0
            for k in range(j, i):
                sum_k += L[i, k] * L_inv[k, j]
            L_inv[i, j] = -sum_k / L[i, i]
    return L_inv

# STEP 5: LU Decomposition with elementary matrices
U = A.copy()
elementary_matrices = []
elimination_steps = []

print("\n--- STEP 3: Elimination Process ---")

for col in range(n - 1):
    print(f"\n  Processing Column {col}:")
    for row in range(col + 1, n):
        if abs(U[row, col]) > 1e-10:
            multiplier = U[row, col] / U[col, col]
            print(f"    R{row} → R{row} - ({multiplier:.4f}) × R{col}")
            
            E = elementary_matrix(n, row, col, multiplier)
            elementary_matrices.append(E)
            elimination_steps.append((row, col, multiplier))
            
            # Apply elimination: U = E × U
            U = matrix_multiply(E, U)

# STEP 6: Build L from product of elementary matrices inverse
# First compute product of all elementary matrices: E_total = E_k × ... × E_2 × E_1
E_product = np.eye(n)
for E in elementary_matrices:
    E_product = matrix_multiply(E, E_product)

# Then invert to get L (since L = (E_total)⁻¹)
L = invert_lower_triangular(E_product)

print("\n--- STEP 4: Results ---")
print("\nL (Lower Triangular):")
for row in L:
    print(" ", [round(x, 4) for x in row])

print("\nU (Upper Triangular):")
for row in U:
    print(" ", [round(x, 4) for x in row])

# STEP 7: Verification A = L × U (manually)
print("\n--- STEP 5: Verification A = L × U ---")
LU_product = matrix_multiply(L, U)

print("\nL × U:")
for row in LU_product:
    print(" ", [round(x, 2) for x in row])

print("\nOriginal A:")
for row in A:
    print(" ", [round(x, 2) for x in row])

# Check if equal
diff = 0
for i in range(n):
    for j in range(n):
        diff = max(diff, abs(A[i, j] - LU_product[i, j]))

if diff < 1e-10:
    print("\n✓ VERIFIED: A = L × U")
else:
    print(f"\n✗ Verification failed (max difference = {diff:.2e})")

# Display Elementary Matrices
print("\n--- Elementary Matrices Created ---")
for i, (E, step) in enumerate(zip(elementary_matrices, elimination_steps)):
    row, col, mult = step
    print(f"\nE{i+1}: R{row} → R{row} - ({mult:.4f}) × R{col}")
    for r in range(n):
        print("  ", [round(E[r][c], 4) for c in range(n)])

print("\n" + "="*80)
print("END OF Q2(a)")
print("="*80)
```

```python
# ============================================================================
# Q2(a): LU DECOMPOSITION USING ELEMENTARY MATRICES
# ============================================================================

import numpy as np

print("="*80)
print("Q2(a): LU DECOMPOSITION USING ELEMENTARY MATRICES")
print("="*80)

# STEP 1: Generate SPD matrix
# n=3
n = int(input("\nEnter size of matrix n: "))
print(f"Enter size of matrix n: {n}")
np.random.seed(42)
B = np.random.randint(-5, 6, (n, n))
A = B.T @ B + n * np.eye(n)
A = (A + A.T) / 2

print("\n--- STEP 1: Matrix A (Symmetric Positive Definite) ---")
for row in A:
    print(" ", [round(x, 2) for x in row])

# STEP 2: Function to create elementary matrix
print("\n--- STEP 2: create elementary matrix---")
def elementary_matrix(n, row, col, multiplier):
    """E = I with -multiplier at (row, col)"""
    E = np.eye(n)
    E[row, col] = -multiplier
    return E

# STEP 3: LU Decomposition with elementary matrices
U = A.copy()
elementary_matrices = []
elimination_steps = []

print("\n--- STEP 3: Elimination Process ---")

for col in range(n - 1):
    print(f"\n  Processing Column {col}:")
    for row in range(col + 1, n):
        if abs(U[row, col]) > 1e-10:
            multiplier = U[row, col] / U[col, col]
            print(f"    R{row} → R{row} - ({multiplier:.4f}) × R{col}")
            
            E = elementary_matrix(n, row, col, multiplier)
            elementary_matrices.append(E)
            elimination_steps.append((row, col, multiplier))
            U = E @ U

# STEP 4: Build L from product of elementary matrices inverse
E_product = np.eye(n)
for E in elementary_matrices:
    E_product = E @ E_product
L = np.linalg.inv(E_product)

print("\n--- STEP 4: Results ---")
print("\nL (Lower Triangular):")
for row in L:
    print(" ", [round(x, 4) for x in row])

print("\nU (Upper Triangular):")
for row in U:
    print(" ", [round(x, 4) for x in row])

# STEP 5: Verification
print("\n--- STEP 5: Verification A = L × U ---")
LU_product = L @ U
if np.allclose(A, LU_product, rtol=1e-10):
    print("✓ VERIFIED: A = L × U")
else:
    print("✗ Verification failed")

# Display Elementary Matrices
print("\n--- Elementary Matrices Created ---")
for i, (E, step) in enumerate(zip(elementary_matrices, elimination_steps)):
    row, col, mult = step
    print(f"\nE{i+1}: R{row} → R{row} - ({mult:.4f}) × R{col}")
    for r in range(n):
        print("  ", [round(E[r][c], 4) for c in range(n)])

print("\n" + "="*80)
print("END OF Q2(a)")
print("="*80)
```

```python
# ============================================================================
# Q2(a): LU DECOMPOSITION USING ELEMENTARY MATRICES
# For Symmetric Positive Definite Matrix A
# ============================================================================

import numpy as np

# ============================================================================
# STEP 1: Generate SPD Matrix A = B^T B + nI
# ============================================================================

n = int(input("Enter size of matrix n: "))
np.random.seed(42)
B = np.random.randint(-5, 6, (n, n))
A = B.T @ B + n * np.eye(n)
A = (A + A.T) / 2

print("\n--- Matrix A (Symmetric Positive Definite) ---")
for row in A:
    print([round(x, 2) for x in row])

# ============================================================================
# STEP 2: Function to Create Elementary Matrix
# ============================================================================

def elementary_matrix(n, row, col, multiplier):
    """E = I with -multiplier at (row, col)"""
    E = np.eye(n)
    E[row, col] = -multiplier
    return E

def inverse_elementary(n, row, col, multiplier):
    """E^{-1} = I with +multiplier at (row, col)"""
    E_inv = np.eye(n)
    E_inv[row, col] = multiplier
    return E_inv

# ============================================================================
# STEP 3: LU Decomposition Using Elementary Matrices
# ============================================================================

U = A.copy()
L = np.eye(n)
elementary_matrices = []
inverse_matrices = []

for col in range(n - 1):
    for row in range(col + 1, n):
        if abs(U[row, col]) > 1e-10:
            multiplier = U[row, col] / U[col, col]
            
            # Create elementary matrix
            E = elementary_matrix(n, row, col, multiplier)
            E_inv = inverse_elementary(n, row, col, multiplier)
            
            elementary_matrices.append(E)
            inverse_matrices.append(E_inv)
            
            # Apply elimination
            U = E @ U

# Build L from inverse elementary matrices
for E_inv in inverse_matrices:
    L = L @ E_inv

# ============================================================================
# STEP 4: Display Results
# ============================================================================

print("\n--- L (Lower Triangular) ---")
for row in L:
    print([round(x, 4) for x in row])

print("\n--- U (Upper Triangular) ---")
for row in U:
    print([round(x, 4) for x in row])

# ============================================================================
# STEP 5: Verification A = L × U
# ============================================================================

print("\n--- Verification: A = L × U ---")
LU_product = L @ U
if np.allclose(A, LU_product, rtol=1e-10):
    print("✓ VERIFIED: A = L × U")
else:
    print("✗ Verification failed")

# ============================================================================
# STEP 6: Display Elementary Matrices (Deliverable)
# ============================================================================

print(f"\n--- Elementary Matrices Created: {len(elementary_matrices)} ---")
for i, E in enumerate(elementary_matrices):
    print(f"\nE{i+1}:")
    for row in E:
        print(" ", [round(x, 4) for x in row])

print("\n" + "="*60)
print("END OF Q2(a)")
print("="*60)
```
