```python
# ============================================================================
# Q2(d): RANDOM 7×5 MATRIX - DIAGONAL OF R OBSERVATION (No linalg)
# ============================================================================

import numpy as np

print("="*80)
print("Q2(d): RANDOM 7×5 MATRIX - DIAGONAL OF R OBSERVATION")
print("="*80)

# STEP 1: Generate random 7×5 matrix
np.random.seed(42)
A = np.random.randint(-10, 11, (7, 5)).astype(float)

print("\n--- STEP 1: Random Matrix A (7 × 5) ---")
for row in A:
    print(" ", [round(x, 2) for x in row])

# STEP 2: Function to compute rank (without linalg)
def compute_rank(matrix):
    """Compute rank using Gaussian elimination"""
    M = matrix.copy()
    rows, cols = M.shape
    rank = 0
    row = 0
    
    for col in range(cols):
        if row >= rows:
            break
        
        # Find pivot
        pivot = row
        while pivot < rows and abs(M[pivot, col]) < 1e-10:
            pivot += 1
        
        if pivot == rows:
            continue
        
        # Swap rows
        if pivot != row:
            M[[row, pivot]] = M[[pivot, row]]
        
        # Eliminate below
        for i in range(row + 1, rows):
            if abs(M[i, col]) > 1e-10:
                factor = M[i, col] / M[row, col]
                for j in range(col, cols):
                    M[i, j] -= factor * M[row, j]
        
        rank += 1
        row += 1
    
    return rank

# Check linear independence
rank = compute_rank(A)
print(f"\nRank of A = {rank} (should be 5 for full column rank)")
if rank == 5:
    print("✓ Columns are linearly independent")
else:
    print("✗ Columns are NOT linearly independent")

# STEP 3: Function for vector norm (without linalg)
def vector_norm(v):
    norm_sq = 0
    for val in v:
        norm_sq += val ** 2
    return np.sqrt(norm_sq)

# STEP 4: Function for dot product (without linalg)
def dot_product(v1, v2):
    result = 0
    for i in range(len(v1)):
        result += v1[i] * v2[i]
    return result

# STEP 5: Function to multiply matrices (without linalg)
def matrix_multiply(X, Y):
    m, n_x = X.shape
    n_y, p = Y.shape
    result = np.zeros((m, p))
    for i in range(m):
        for j in range(p):
            for k in range(n_x):
                result[i, j] += X[i, k] * Y[k, j]
    return result

# STEP 6: QR Decomposition
def qr_decomposition(A):
    m, n = A.shape
    Q = np.zeros((m, n))
    R = np.zeros((n, n))
    
    for j in range(n):
        v = A[:, j].copy()
        for i in range(j):
            R[i, j] = dot_product(Q[:, i], A[:, j])
            for k in range(m):
                v[k] -= R[i, j] * Q[k, i]
        R[j, j] = vector_norm(v)
        for k in range(m):
            Q[k, j] = v[k] / R[j, j]
    
    return Q, R

Q, R = qr_decomposition(A)

print("\n--- STEP 2: QR Decomposition Results ---")

print("\nQ (7 × 5):")
for row in Q:
    print(" ", [round(x, 4) for x in row])

print("\nR (5 × 5 Upper Triangular):")
for row in R:
    print(" ", [round(x, 4) for x in row])

# STEP 7: Display Diagonal of R
print("\n--- STEP 3: Diagonal Elements of R ---")
diagonal_R = [R[i][i] for i in range(5)]
for i, d in enumerate(diagonal_R):
    print(f"  R[{i}][{i}] = {d:.6f}")

# Check if all positive
all_positive = all(d > 1e-10 for d in diagonal_R)
print(f"\nAll diagonal entries positive: {all_positive}")

# STEP 8: Verification A = Q × R
print("\n--- STEP 4: Verification A = Q × R ---")
QR_product = matrix_multiply(Q, R)

diff = 0
for i in range(7):
    for j in range(5):
        diff = max(diff, abs(A[i, j] - QR_product[i, j]))

if diff < 1e-10:
    print("✓ VERIFIED: A = Q × R")
else:
    print(f"✗ Verification failed (max difference = {diff:.2e})")

# STEP 9: Observation
print("\n" + "="*80)
print("OBSERVATION ON DIAGONAL ELEMENTS OF R")
print("="*80)

print(f"""
┌─────────────────────────────────────────────────────────────────────────────┐
│                           OBSERVATION                                        │
├─────────────────────────────────────────────────────────────────────────────┤
│                                                                              │
│  1. ALL DIAGONAL ELEMENTS OF R ARE STRICTLY POSITIVE ( > 0 )                │
│                                                                              │
│  2. DIAGONAL VALUES OBSERVED:                                               │
│     R[0][0] = {diagonal_R[0]:.6f}                                                         │
│     R[1][1] = {diagonal_R[1]:.6f}                                                         │
│     R[2][2] = {diagonal_R[2]:.6f}                                                         │
│     R[3][3] = {diagonal_R[3]:.6f}                                                         │
│     R[4][4] = {diagonal_R[4]:.6f}                                                         │
│                                                                              │
│  3. REASON FOR POSITIVITY:                                                  │
│     • Each diagonal entry R[j][j] is the LENGTH (norm) of the residual      │
│       vector after subtracting projections onto previous Q columns          │
│     • The residual vector = column j - projection onto span of previous     │
│       columns                                                               │
│     • Since all columns of A are linearly independent, this residual        │
│       cannot be the zero vector                                             │
│     • Therefore its norm is strictly greater than zero                      │
│                                                                              │
│  4. IMPLICATIONS:                                                            │
│     • Positive diagonal means R is invertible                               │
│     • det(R) = product of diagonal entries > 0                              │
│     • QR decomposition is unique for matrices with full column rank         │
│                                                                              │
└─────────────────────────────────────────────────────────────────────────────┘
""")

print("\n" + "="*80)
print("END OF Q2(d)")
print("="*80)
```

```python
# ============================================================================
# Q2(d): RANDOM 7×5 MATRIX - DIAGONAL OF R OBSERVATION
# ============================================================================

import numpy as np

print("="*80)
print("Q2(d): RANDOM 7×5 MATRIX - DIAGONAL OF R OBSERVATION")
print("="*80)

# STEP 1: Generate random 7×5 matrix
np.random.seed(42)
A = np.random.randint(-10, 11, (7, 5)).astype(float)

print("\n--- STEP 1: Random Matrix A (7 × 5) ---")
for row in A:
    print(" ", [round(x, 2) for x in row])

# Check linear independence
rank = np.linalg.matrix_rank(A)
print(f"\nRank of A = {rank} (should be 5 for full column rank)")
if rank == 5:
    print("✓ Columns are linearly independent")
else:
    print("✗ Columns are NOT linearly independent")

# STEP 2: QR Decomposition using Gram-Schmidt
def qr_decomposition(A):
    m, n = A.shape
    Q = np.zeros((m, n))
    R = np.zeros((n, n))
    
    for j in range(n):
        v = A[:, j].copy()
        for i in range(j):
            R[i, j] = Q[:, i] @ A[:, j]
            v = v - R[i, j] * Q[:, i]
        R[j, j] = np.linalg.norm(v)
        Q[:, j] = v / R[j, j]
    return Q, R

Q, R = qr_decomposition(A)

print("\n--- STEP 2: QR Decomposition Results ---")

print("\nQ (7 × 5):")
for row in Q:
    print(" ", [round(x, 4) for x in row])

print("\nR (5 × 5 Upper Triangular):")
for row in R:
    print(" ", [round(x, 4) for x in row])

# STEP 3: Display Diagonal of R
print("\n--- STEP 3: Diagonal Elements of R ---")
diagonal_R = [R[i][i] for i in range(5)]
for i, d in enumerate(diagonal_R):
    print(f"  R[{i}][{i}] = {d:.6f}")

# Check if all positive
all_positive = all(d > 1e-10 for d in diagonal_R)
print(f"\nAll diagonal entries positive: {all_positive}")

# STEP 4: Verification
print("\n--- STEP 4: Verification A = Q × R ---")
QR_product = Q @ R
if np.allclose(A, QR_product, rtol=1e-10):
    print("✓ VERIFIED: A = Q × R")
else:
    print("✗ Verification failed")

# STEP 5: Observation
print("\n" + "="*80)
print("OBSERVATION ON DIAGONAL ELEMENTS OF R")
print("="*80)

print("""
┌─────────────────────────────────────────────────────────────────────────────┐
│                           OBSERVATION                                        │
├─────────────────────────────────────────────────────────────────────────────┤
│                                                                              │
│  1. ALL DIAGONAL ELEMENTS OF R ARE STRICTLY POSITIVE ( > 0 )                │
│                                                                              │
│  2. DIAGONAL VALUES OBSERVED:                                               │
│     R[0][0] = {:.6f}                                                         │
│     R[1][1] = {:.6f}                                                         │
│     R[2][2] = {:.6f}                                                         │
│     R[3][3] = {:.6f}                                                         │
│     R[4][4] = {:.6f}                                                         │
│                                                                              │
│  3. REASON FOR POSITIVITY:                                                  │
│     • Each diagonal entry R[j][j] is the LENGTH (norm) of the residual      │
│       vector after subtracting projections onto previous Q columns          │
│     • The residual vector = column j - projection onto span of previous     │
│       columns                                                               │
│     • Since all columns of A are linearly independent, this residual        │
│       cannot be the zero vector                                             │
│     • Therefore its norm is strictly greater than zero                      │
│                                                                              │
│  4. IMPLICATIONS:                                                            │
│     • Positive diagonal means R is invertible                               │
│     • det(R) = product of diagonal entries > 0                              │
│     • QR decomposition is unique for matrices with full column rank         │
│                                                                              │
└─────────────────────────────────────────────────────────────────────────────┘
""".format(diagonal_R[0], diagonal_R[1], diagonal_R[2], diagonal_R[3], diagonal_R[4]))

print("\n" + "="*80)
print("END OF Q2(d)")
print("="*80)
```
