```python
# ============================================================================
# Q2(b): CHOLESKY DECOMPOSITION (A = L × Lᵀ) - No linalg
# ============================================================================

import numpy as np

print("="*80)
print("Q2(b): CHOLESKY DECOMPOSITION (A = L × Lᵀ)")
print("="*80)

# STEP 1: Generate SPD matrix (same as Q2a)
n = int(input("\nEnter size of matrix n: "))
np.random.seed(42)
B = np.random.randint(-5, 6, (n, n))
A = B.T @ B + n * np.eye(n)
A = (A + A.T) / 2

print("\n--- STEP 1: Matrix A (Symmetric Positive Definite) ---")
for row in A:
    print(" ", [round(x, 2) for x in row])

# STEP 2: Cholesky Decomposition (without linalg)
def cholesky_decomposition(A):
    """Compute Cholesky decomposition A = L × Lᵀ"""
    n = A.shape[0]
    L = np.zeros((n, n))
    
    print("\n--- STEP 2: Cholesky Algorithm Process ---")
    
    for i in range(n):
        for j in range(i + 1):
            if i == j:
                # Diagonal element
                sum_sq = 0
                for k in range(j):
                    sum_sq += L[i][k] ** 2
                L[i][j] = np.sqrt(A[i][i] - sum_sq)
                print(f"  L[{i}][{i}] = sqrt({A[i][i]:.2f} - {sum_sq:.4f}) = {L[i][j]:.4f}")
            else:
                # Off-diagonal element
                sum_prod = 0
                for k in range(j):
                    sum_prod += L[i][k] * L[j][k]
                L[i][j] = (A[i][j] - sum_prod) / L[j][j]
                print(f"  L[{i}][{j}] = ({A[i][j]:.2f} - {sum_prod:.4f}) / {L[j][j]:.4f} = {L[i][j]:.4f}")
    
    return L

L = cholesky_decomposition(A)

print("\n--- STEP 3: L (Lower Triangular from Cholesky) ---")
for row in L:
    print(" ", [round(x, 4) for x in row])

# STEP 4: Function to multiply matrices (without linalg)
def matrix_multiply(X, Y):
    m, n = X.shape
    n, p = Y.shape
    result = np.zeros((m, p))
    for i in range(m):
        for j in range(p):
            for k in range(n):
                result[i, j] += X[i, k] * Y[k, j]
    return result

# STEP 5: Verification A = L × Lᵀ
print("\n--- STEP 4: Verification A = L × Lᵀ ---")
L_T = L.T
LLT_product = matrix_multiply(L, L_T)

print("\nL × Lᵀ:")
for row in LLT_product:
    print(" ", [round(x, 2) for x in row])

print("\nOriginal A:")
for row in A:
    print(" ", [round(x, 2) for x in row])

# Check if equal
diff = 0
for i in range(n):
    for j in range(n):
        diff = max(diff, abs(A[i, j] - LLT_product[i, j]))

if diff < 1e-10:
    print("\n✓ VERIFIED: A = L × Lᵀ")
else:
    print(f"\n✗ Verification failed (max difference = {diff:.2e})")

print("\n" + "="*80)
print("END OF Q2(b)")
print("="*80)
```

```python
# ============================================================================
# Q2(b): CHOLESKY DECOMPOSITION (A = L × Lᵀ)
# ============================================================================

import numpy as np

print("="*80)
print("Q2(b): CHOLESKY DECOMPOSITION (A = L × Lᵀ)")
print("="*80)

# STEP 1: Use same SPD matrix from Q2(a)
# n=3
n = int(input("\nEnter size of matrix n: "))
print(f"Enter size of matrix n: {n}")
np.random.seed(42)
B = np.random.randint(-5, 6, (n, n))
A = B.T @ B + n * np.eye(n)
A = (A + A.T) / 2

print("\n--- STEP 1: Matrix A (Symmetric Positive Definite) ---")
for row in A:
    print(" ", [round(x, 2) for x in row])

# STEP 2: Cholesky Decomposition
def cholesky_decomposition(A):
    """Compute Cholesky decomposition A = L × Lᵀ"""
    n = A.shape[0]
    L = np.zeros((n, n))
    
    print("\n--- STEP 2: Cholesky Algorithm Process ---")
    
    for i in range(n):
        for j in range(i + 1):
            if i == j:
                # Diagonal element
                sum_sq = sum(L[i][k] ** 2 for k in range(j))
                L[i][j] = np.sqrt(A[i][i] - sum_sq)
                print(f"  L[{i}][{i}] = sqrt({A[i][i]:.2f} - {sum_sq:.4f}) = {L[i][j]:.4f}")
            else:
                # Off-diagonal element
                sum_prod = sum(L[i][k] * L[j][k] for k in range(j))
                L[i][j] = (A[i][j] - sum_prod) / L[j][j]
                print(f"  L[{i}][{j}] = ({A[i][j]:.2f} - {sum_prod:.4f}) / {L[j][j]:.4f} = {L[i][j]:.4f}")
    
    return L

L = cholesky_decomposition(A)

print("\n--- STEP 3: L (Lower Triangular from Cholesky) ---")
for row in L:
    print(" ", [round(x, 4) for x in row])

# STEP 4: Verification A = L × Lᵀ
print("\n--- STEP 4: Verification A = L × Lᵀ ---")
LLT_product = L @ L.T

print("\nL × Lᵀ:")
for row in LLT_product:
    print(" ", [round(x, 2) for x in row])

print("\nOriginal A:")
for row in A:
    print(" ", [round(x, 2) for x in row])

if np.allclose(A, LLT_product, rtol=1e-10):
    print("\n✓ VERIFIED: A = L × Lᵀ")
else:
    print("\n✗ Verification failed")

print("\n" + "="*80)
print("END OF Q2(b)")
print("="*80)
```
