Absolutely. Before solving the question, let's understand **SVM from the basics**, in simple MFML exam language.

# 1. What is SVM?

**SVM = Support Vector Machine**

SVM is a **supervised machine learning algorithm** mainly used for **classification**.

Its main job is:

> **Find the best boundary that separates two classes.**

For example:

```text
       + + + +       Class +1
       + + +

-------------------  ← Decision boundary

       - - -
       - - - -       Class -1
```

The line separating the two classes is called the **hyperplane**.

---

# 2. What is a hyperplane?

For 2 features \(x_1,x_2\), the hyperplane is a line:

$$
\boxed{w^Tx+b=0}
$$

or:

$$
\boxed{w_1x_1+w_2x_2+b=0}
$$

Where:

* \(w\) = **weight vector**
* \(b\) = **bias**
* \(x\) = input/features

For example:

$$
2x_1+x_2-5=0
$$

is a decision boundary.

---

# 3. What is the main idea of SVM?

There can be many lines that separate two classes.

SVM doesn't just find **any** separating line.

It finds the line with the **maximum margin**.

Imagine:

```text
        + + + + 
        + + +

        | ← margin
--------|----------------
        | ← margin

        - - - -
        - - -
```

The **margin** is the distance between the decision boundary and the closest data points.

SVM tries to make this margin as **large as possible**.

### Easy memory:

> **SVM = Find the separating boundary with the maximum margin.**

---

# 4. What are Support Vectors?

The data points **closest to the decision boundary** are called:

$$
\boxed{\text{Support Vectors}}
$$

Example:

```text
 + + +        + = Class +1

    +   ← Support Vector

--------- ← Hyperplane

    -   ← Support Vector

 - - -        - = Class -1
```

These points are important because they determine the position of the optimal boundary.

### Easy memory:

> **Support vectors = closest points to the hyperplane.**

---

# 5. What is a hard-margin SVM?

Your question specifically asks for **hard-margin SVM**.

Hard-margin SVM is used when the classes are **perfectly linearly separable**.

That means:

> There are no misclassified points.

The constraint is:

$$
\boxed{y_i(w^Tx_i+b)\geq1}
$$

Why \(y_i\)?

Because the classes are represented as:

$$
+1 \quad \text{and} \quad -1
$$

---

# 6. Understand the constraint

Suppose we have a positive point:

$$
y_i=+1
$$

Then:

$$
(+1)(w^Tx_i+b)\geq1
$$

Therefore:

$$
w^Tx_i+b\geq1
$$

For a negative point:

$$
y_i=-1
$$

Then:

$$
(-1)(w^Tx_i+b)\geq1
$$

which means:

$$
w^Tx_i+b\leq-1
$$

So SVM creates two important boundaries:

$$
\boxed{w^Tx+b=1}
$$

and

$$
\boxed{w^Tx+b=-1}
$$

The middle boundary is:

$$
\boxed{w^Tx+b=0}
$$

---

# 7. What is the objective function?

Hard-margin SVM minimizes:

$$
\boxed{\frac12\|w\|^2}
$$

subject to:

$$
\boxed{y_i(w^Tx_i+b)\geq1}
$$

Why minimize \(\|w\|\)?

Because the margin is:

$$
\boxed{\frac{2}{\|w\|}}
$$

Therefore:

* Smaller \(\|w\|\)
* Larger margin

So SVM is effectively trying to:

> **Maximize the margin.**

---

The **+1 points are above** and the **-1 points are below**.

So the best separating line will be somewhere between them.

Because the positive points have:

$$
x_2=2,4
$$

and negative points have:

$$
x_2=0
$$

the optimal boundary will be **horizontal**.

---

# 9. What will \(w\) look like?

Because the separating line is horizontal, the equation doesn't need \(x_1\).

So:

$$
w_1=0
$$

and only \(x_2\) matters:

$$
w=[0,w_2]
$$

The decision boundary is:

$$
w_2x_2+b=0
$$

# 10. What you need to remember for the exam

| Term                | Simple meaning                           |
| ------------------- | ---------------------------------------- |
| **SVM**             | Classification algorithm                 |
| **Hyperplane**      | Decision boundary                        |
| **Margin**          | Distance from boundary to closest points |
| **Support vectors** | Closest points to boundary               |
| **Hard margin**     | No classification errors allowed         |
| **Weight \(w\)**    | Determines orientation of boundary       |
| **Bias \(b\)**      | Moves boundary                           |
| **Objective**       | Minimize \(\frac12\|w\|^2\)              |
| **Constraint**      | \(y_i(w^Tx_i+b)\geq1\)                   |
