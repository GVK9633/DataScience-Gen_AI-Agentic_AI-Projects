# One‑Line Memory Trick for Exam

Eigenvalues tell you "how much" a matrix stretches. Eigenvectors tell you "which direction" it stretches.

8. One‑Line Summary for Beginners
Eigenvectors = special directions that don't change when a matrix is applied.
Eigenvalues = how much the vector stretches/shrinks in that direction.

9. Memory Trick

"Eigen = 'own' in German.
Eigenvectors are the 'own directions' of a matrix — directions that belong only to that matrix."

# 1. What is an Eigenvector?

Simple definition:
An eigenvector of a matrix is a special direction that, when you apply the matrix transformation, does not change direction — it only gets stretched or squashed (scaled).

Even simpler:
Imagine stretching a rubber band. Some lines on the band stay in the same direction — they just get longer. Those lines are eigenvectors.

# 2. What is an Eigenvalue?

Simple definition:
An eigenvalue is the scale factor by which the eigenvector is stretched or squashed.

Even simpler:
If the rubber band stretches to 3 times its original length, the eigenvalue is 3. If it shrinks to half, the eigenvalue is 0.5.

# 4. Real-Life Analogies

# Analogy 1: Stretching a rubber sheet

Draw an arrow on a rubber sheet.

Stretch the sheet in one direction.

Some arrows point in the stretch direction → they just get longer → eigenvectors.

The stretch factor → eigenvalue.

# Analogy 2: A mirror

Mirror reflection flips left-right.

An arrow pointing straight into the mirror stays in the same direction (just reverses) → eigenvector.

Eigenvalue = -1 (because it flips).

# Analogy 3: Spinning a top

The axis of rotation stays fixed → eigenvector.

Speed of spin → eigenvalue.

## 3. The Equation

\[
A \mathbf{v} = \lambda \mathbf{v}
\]

| Symbol | Meaning |
|--------|---------|
| \( A \) | A square matrix (transformation) |
| \( \mathbf{v} \) | Eigenvector (direction that doesn't change) |
| \( \lambda \) (lambda) | Eigenvalue (how much it stretches/shrinks) |

## 5. Small Example (Step-by-Step)

**Matrix (transformation):**

\[
A = \begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix}
\]

We want to find vectors \( \mathbf{v} \) such that:
\[
A \mathbf{v} = \lambda \mathbf{v}
\]

---

### Step 1: Assume \( \mathbf{v} = \begin{bmatrix} x \\ y \end{bmatrix} \)

We solve:
\[
\begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \lambda \begin{bmatrix} x \\ y \end{bmatrix}
\]

---

### Step 2: Find eigenvalues (λ)

Use the characteristic equation:
\[
\det(A - \lambda I) = 0
\]

\[
\det \begin{bmatrix} 2-\lambda & 1 \\ 1 & 2-\lambda \end{bmatrix} = (2-\lambda)^2 - 1 = 0
\]

\[
(2-\lambda)^2 = 1 \Rightarrow 2-\lambda = \pm 1
\]

\[
\lambda = 1 \quad \text{or} \quad \lambda = 3
\]

**Eigenvalues: 1 and 3**

---

### Step 3: Find eigenvectors (directions)

#### For \( \lambda = 1 \):
\[
(A - I)\mathbf{v} = 0
\]

\[
\begin{bmatrix} 1 & 1 \\ 1 & 1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \end{bmatrix}
\]

\[
x + y = 0 \Rightarrow x = -y
\]

So eigenvector = \( \begin{bmatrix} 1 \\ -1 \end{bmatrix} \) (or any multiple).

**Meaning:** Along the direction (1, -1), the matrix **does nothing** (stretch factor = 1).

---

#### For \( \lambda = 3 \):
\[
(A - 3I)\mathbf{v} = 0
\]

\[
\begin{bmatrix} -1 & 1 \\ 1 & -1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \end{bmatrix}
\]

\[
-x + y = 0 \Rightarrow x = y
\]

So eigenvector = \( \begin{bmatrix} 1 \\ 1 \end{bmatrix} \).

**Meaning:** Along the direction (1, 1), the matrix **stretches by factor 3**.

## 6. Visual Check

| Vector | Direction | After transformation | Eigenvalue? |
|--------|-----------|----------------------|-------------|
| (1, -1) | Diagonal down-right | Stays same direction | λ = 1 |
| (1, 1) | Diagonal up-right | Stretches by 3x | λ = 3 |
| (1, 0) | Horizontal | Becomes (2, 1) → changes direction | Not eigenvector |

## 7. Why This Matters in ML

| Use Case | What eigenvectors/eigenvalues do |
|----------|----------------------------------|
| **PCA** | Find directions of maximum variance (eigenvectors of covariance matrix) |
| **SVD** | Eigenvectors of \( A^T A \) and \( AA^T \) |
| **Stability** | Eigenvalues tell if system converges (|λ| < 1 for stability) |
| **Markov chains** | Steady-state distribution = eigenvector for λ = 1 |
| **Neural nets** | Hessian eigenvalues tell if point is min/max/saddle |

## 10. Quick Exam Cheat Sheet

| Term | Definition |
|------|------------|
| **Eigenvector** | Non‑zero vector \( v \) such that \( A v = \lambda v \) |
| **Eigenvalue** | Scalar \( \lambda \) that satisfies the above |
| **Characteristic equation** | \( \det(A - \lambda I) = 0 \) |
| **Trace** | Sum of eigenvalues = sum of diagonal entries |
| **Determinant** | Product of eigenvalues |
