## What is a Vector Space?

**Definition:** A **vector space** is a collection of objects (called **vectors**) where you can:
1. **Add** two vectors together → get another vector
2. **Multiply** a vector by a number (scalar) → get another vector

These operations must follow **10 specific rules (axioms)**.

---

## Why Does This Matter in ML?

| ML Concept | Uses Vector Spaces |
|------------|-------------------|
| **Data points** | Each sample is a vector (e.g., [age, income, house size]) |
| **Features** | Each feature is a dimension in the vector space |
| **Weights** \( \theta \) | Are vectors in the parameter space |
| **Gradient descent** | Moves a weight vector through the space |
| **PCA / Dimensionality reduction** | Projects vectors onto lower-dimensional spaces |
| **Support Vector Machines (SVM)** | Finds hyperplanes in vector spaces |

## The 10 Axioms of a Vector Space

Let's go through each axiom from your image, explained simply.

---

### Axiom 1: Closure under Addition

**Statement:** If \( \mathbf{u} \) and \( \mathbf{v} \) are vectors in \( \mathbb{V} \), then \( \mathbf{u} + \mathbf{v} \) is also in \( \mathbb{V} \).

**Simple meaning:** Adding two vectors doesn't leave the space — the result stays inside.

**Example:** In \( \mathbb{R}^2 \), if \( \mathbf{u} = (1, 2) \) and \( \mathbf{v} = (3, 4) \), then \( \mathbf{u} + \mathbf{v} = (4, 6) \), which is still in \( \mathbb{R}^2 \).

---

### Axiom 2: Commutativity of Addition

**Statement:** \( \mathbf{u} + \mathbf{v} = \mathbf{v} + \mathbf{u} \)

**Simple meaning:** Order doesn't matter when adding.

**Example:** \( (1, 2) + (3, 4) = (4, 6) \) and \( (3, 4) + (1, 2) = (4, 6) \).

---

### Axiom 3: Associativity of Addition

**Statement:** \( (\mathbf{u} + \mathbf{v}) + \mathbf{w} = \mathbf{u} + (\mathbf{v} + \mathbf{w}) \)

**Simple meaning:** Grouping doesn't matter when adding.

**Example:** \( ((1,2) + (3,4)) + (5,6) = (4,6) + (5,6) = (9,12) \)  
Same as \( (1,2) + ((3,4) + (5,6)) = (1,2) + (8,10) = (9,12) \)

---

### Axiom 4: Additive Identity (Zero Vector)

**Statement:** There exists a vector \( \mathbf{0} \) such that \( \mathbf{u} + \mathbf{0} = \mathbf{u} \)

**Simple meaning:** Adding zero doesn't change the vector.

**Example:** In \( \mathbb{R}^2 \), \( \mathbf{0} = (0, 0) \).  
\( (1, 2) + (0, 0) = (1, 2) \)

---

### Axiom 5: Additive Inverse (Negative Vector)

**Statement:** For every vector \( \mathbf{u} \), there exists \( -\mathbf{u} \) such that \( \mathbf{u} + (-\mathbf{u}) = \mathbf{0} \)

**Simple meaning:** Every vector has a "negative" that cancels it out.

**Example:** If \( \mathbf{u} = (1, 2) \), then \( -\mathbf{u} = (-1, -2) \).  
\( (1, 2) + (-1, -2) = (0, 0) \)

---

### Axiom 6: Closure under Scalar Multiplication

**Statement:** If \( c \) is a scalar and \( \mathbf{u} \) is a vector, then \( c\mathbf{u} \) is also in \( \mathbb{V} \).

**Simple meaning:** Multiplying a vector by a number stays inside the space.

**Example:** \( 3 \times (1, 2) = (3, 6) \), still in \( \mathbb{R}^2 \).

---

### Axiom 7: Distributivity of Scalar Multiplication over Vector Addition

**Statement:** \( c(\mathbf{u} + \mathbf{v}) = c\mathbf{u} + c\mathbf{v} \)

**Simple meaning:** Scalar can be distributed over vector addition.

**Example:**  
\( 2 \times ((1,2) + (3,4)) = 2 \times (4,6) = (8,12) \)  
\( 2 \times (1,2) + 2 \times (3,4) = (2,4) + (6,8) = (8,12) \)

---

### Axiom 8: Distributivity over Scalar Addition

**Statement:** \( (c + d)\mathbf{u} = c\mathbf{u} + d\mathbf{u} \)

**Simple meaning:** Scalars can be distributed over addition.

**Example:**  
\( (2 + 3) \times (1,2) = 5 \times (1,2) = (5,10) \)  
\( 2 \times (1,2) + 3 \times (1,2) = (2,4) + (3,6) = (5,10) \)

---

### Axiom 9: Associativity of Scalar Multiplication

**Statement:** \( c(d\mathbf{u}) = (cd)\mathbf{u} \)

**Simple meaning:** Order of multiplying scalars doesn't matter.

**Example:**  
\( 2 \times (3 \times (1,2)) = 2 \times (3,6) = (6,12) \)  
\( (2 \times 3) \times (1,2) = 6 \times (1,2) = (6,12) \)

---

### Axiom 10: Scalar Identity

**Statement:** \( 1 \times \mathbf{u} = \mathbf{u} \)

**Simple meaning:** Multiplying by 1 doesn't change the vector.

**Example:** \( 1 \times (1, 2) = (1, 2) \)

| Functions on a domain | All continuous functions | Kernel methods |
