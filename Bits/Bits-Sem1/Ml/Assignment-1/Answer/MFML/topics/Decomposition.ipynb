## What is Decomposition?

In mathematics and machine learning, **decomposition** means **breaking a complex object (matrix, function, problem, dataset) into simpler components** that are easier to understand, analyze, or compute.

Think of it like:

> **Decomposition = Breaking a big problem into smaller pieces.**




## Matrix Decomposition Techniques

**Decomposition** means expressing a matrix as a product of simpler matrices to make computations easier.

[
A = \text{(simpler matrices)}
]

These techniques are widely used in **Machine Learning, Data Science, Computer Vision, PCA, Recommendation Systems, and Numerical Computing**.

---

# 1. Eigenvalue Decomposition (EVD)

Also called **Eigen Decomposition**.

A square matrix (A) can be decomposed as:

[
A = PDP^{-1}
]

where:

* (P) = matrix of eigenvectors
* (D) = diagonal matrix of eigenvalues
* (P^{-1}) = inverse of eigenvector matrix

### Example

[
A=
\begin{bmatrix}
2 & 0\
0 & 3
\end{bmatrix}
]

[
P=
\begin{bmatrix}
1 & 0\
0 & 1
\end{bmatrix}
]

[
D=
\begin{bmatrix}
2 & 0\
0 & 3
\end{bmatrix}
]

Thus,

[
A=PDP^{-1}
]

### Applications

* PCA
* Spectral Clustering
* Dynamical Systems
* Differential Equations

### Limitation

Works only for matrices that have enough independent eigenvectors.

---

# 2. Singular Value Decomposition (SVD)

The most important decomposition in Machine Learning.

Any matrix (A) (square or rectangular) can be decomposed as:

[
A = U\Sigma V^T
]

where:

* (U) = left singular vectors
* (\Sigma) = singular values (diagonal matrix)
* (V^T) = right singular vectors

### Example

For

[
A=
\begin{bmatrix}
3 & 0\
4 & 5
\end{bmatrix}
]

SVD gives

[
A = U\Sigma V^T
]

(The actual calculation is lengthy and usually done by software.)

### Applications

* PCA
* Recommendation Systems
* Image Compression
* NLP
* Latent Semantic Analysis (LSA)

### Advantages

✅ Works for any matrix

✅ Numerically stable

✅ Most widely used decomposition

---

# 3. Cholesky Decomposition

Used for **symmetric positive-definite matrices**.

A matrix (A) is decomposed as:

[
A = LL^T
]

where:

* (L) = lower triangular matrix
* (L^T) = transpose of (L)

---

### Example

[
A=
\begin{bmatrix}
4 & 2\
2 & 2
\end{bmatrix}
]

Cholesky decomposition:

[
L=
\begin{bmatrix}
2 & 0\
1 & 1
\end{bmatrix}
]

Check:

[
LL^T
====

\begin{bmatrix}
2 & 0\
1 & 1
\end{bmatrix}
\begin{bmatrix}
2 & 1\
0 & 1
\end{bmatrix}
=============

\begin{bmatrix}
4 & 2\
2 & 2
\end{bmatrix}
]

### Applications

* Solving linear systems
* Gaussian Processes
* Covariance Matrices
* Optimization

### Advantages

✅ Faster than LU decomposition

✅ Numerically stable

---

# 4. Spectral Decomposition

A special form of Eigenvalue Decomposition for **symmetric matrices**.

If (A) is symmetric:

[
A = Q\Lambda Q^T
]

where:

* (Q) = orthonormal eigenvectors
* (\Lambda) = diagonal matrix of eigenvalues

---

### Example

[
A=
\begin{bmatrix}
2 & 1\
1 & 2
\end{bmatrix}
]

Eigenvalues:

[
\lambda_1=3,\quad \lambda_2=1
]

Eigenvectors:

[
q_1=
\frac1{\sqrt2}
\begin{bmatrix}
1\
1
\end{bmatrix}
,\quad
q_2=
\frac1{\sqrt2}
\begin{bmatrix}
1\
-1
\end{bmatrix}
]

Then:

[
A=Q\Lambda Q^T
]

### Applications

* PCA
* Spectral Clustering
* Graph Analysis
* Quantum Mechanics

---

# Relationship Between EVD and Spectral Decomposition

### EVD

[
A=PDP^{-1}
]

* General square matrix
* Eigenvectors need not be orthogonal

### Spectral Decomposition

[
A=Q\Lambda Q^T
]

* Symmetric matrix only
* Eigenvectors are orthonormal

**Every spectral decomposition is an eigen decomposition, but not every eigen decomposition is spectral decomposition.**

# Quick Comparison Table

| Method   | Formula          | Matrix Type                 | Main Use            |
| -------- | ---------------- | --------------------------- | ------------------- |
| EVD      | (A=PDP^{-1})     | Square                      | Eigen analysis      |
| SVD      | (A=U\Sigma V^T)  | Any matrix                  | PCA, Compression    |
| Cholesky | (A=LL^T)         | Symmetric Positive Definite | Fast linear solving |
| Spectral | (A=Q\Lambda Q^T) | Symmetric                   | PCA, Clustering     |

### Memory Trick

* **EVD** → Eigenvalues + Eigenvectors
* **SVD** → Any Matrix
* **Cholesky** → Symmetric Positive Definite Matrix
* **Spectral** → Symmetric Matrix + Orthonormal Eigenvectors
* **PCA** → Built using EVD or SVD of the covariance/data matrix.
