## Problem 1: Check Consistency (2 equations, 2 unknowns)

**Question:**  
Determine if the following system is consistent or inconsistent:

\[
\begin{cases}
x + 2y = 5 \\
2x + 4y = 10
\end{cases}
\]

---

### Answer:

**Step 1:** Write augmented matrix

\[
\left[\begin{array}{cc|c}
1 & 2 & 5 \\
2 & 4 & 10
\end{array}\right]
\]

**Step 2:** Eliminate below first pivot  
\( R_2 \leftarrow R_2 - 2R_1 \)

\[
\left[\begin{array}{cc|c}
1 & 2 & 5 \\
0 & 0 & 0
\end{array}\right]
\]

**Step 3:** Check for inconsistency  
No row of form \([0\ 0\ |\ \text{non-zero}]\) → **Consistent**

\[
\boxed{\text{Consistent (infinitely many solutions)}}
\]

## Problem 3: Check Consistency (3 equations, 3 unknowns)

**Question:**  
Test consistency of the system:

\[
\begin{cases}
x + y + z = 6 \\
2x + 3y + 4z = 20 \\
3x + 4y + 5z = 26
\end{cases}
\]

---

### Answer:

**Step 1:** Augmented matrix

\[
\left[\begin{array}{ccc|c}
1 & 1 & 1 & 6 \\
2 & 3 & 4 & 20 \\
3 & 4 & 5 & 26
\end{array}\right]
\]

**Step 2:** \( R_2 \leftarrow R_2 - 2R_1 \), \( R_3 \leftarrow R_3 - 3R_1 \)

\[
\left[\begin{array}{ccc|c}
1 & 1 & 1 & 6 \\
0 & 1 & 2 & 8 \\
0 & 1 & 2 & 8
\end{array}\right]
\]

**Step 3:** \( R_3 \leftarrow R_3 - R_2 \)

\[
\left[\begin{array}{ccc|c}
1 & 1 & 1 & 6 \\
0 & 1 & 2 & 8 \\
0 & 0 & 0 & 0
\end{array}\right]
\]

**Step 4:** Check inconsistency  
No row \([0\ 0\ 0\ |\ \text{non-zero}]\) → **Consistent**

\[
\boxed{\text{Consistent (infinitely many solutions)}}
\]

## Problem 4: Check Consistency (3 equations, 3 unknowns)

**Question:**  
Test consistency:

\[
\begin{cases}
x + y + z = 3 \\
2x + 2y + 2z = 7 \\
x + 2y + 3z = 10
\end{cases}
\]

---

### Answer:

**Step 1:** Augmented matrix

\[
\left[\begin{array}{ccc|c}
1 & 1 & 1 & 3 \\
2 & 2 & 2 & 7 \\
1 & 2 & 3 & 10
\end{array}\right]
\]

**Step 2:** \( R_2 \leftarrow R_2 - 2R_1 \), \( R_3 \leftarrow R_3 - R_1 \)

\[
\left[\begin{array}{ccc|c}
1 & 1 & 1 & 3 \\
0 & 0 & 0 & 1 \\
0 & 1 & 2 & 7
\end{array}\right]
\]

**Step 3:** Check inconsistency  
Row 2: \([0\ 0\ 0\ |\ 1]\) → \(0 = 1\) → **Inconsistent**

\[
\boxed{\text{Inconsistent (no solution)}}
\]

## Problem 5: Find Condition for Consistency

**Question:**  
For what value(s) of \(k\) is the following system consistent?

\[
\begin{cases}
x + y = 2 \\
2x + ky = 4
\end{cases}
\]

---

### Answer:

**Step 1:** Augmented matrix

\[
\left[\begin{array}{cc|c}
1 & 1 & 2 \\
2 & k & 4
\end{array}\right]
\]

**Step 2:** \( R_2 \leftarrow R_2 - 2R_1 \)

\[
\left[\begin{array}{cc|c}
1 & 1 & 2 \\
0 & k-2 & 0
\end{array}\right]
\]

**Step 3:** Analyze for consistency

- If \(k-2 \neq 0\) (i.e., \(k \neq 2\)):  
  Second row becomes \([0\ k-2\ |\ 0]\) → divide to get \(y = 0\) → Consistent (unique solution)

- If \(k = 2\):  
  Second row becomes \([0\ 0\ |\ 0]\) → \(0 = 0\) → Consistent (infinitely many solutions)

**Step 4:** Conclusion

\[
\boxed{\text{Consistent for all real } k}
\]

## Problem 6: Find Condition for Consistency (with constant)

**Question:**  
Find \(k\) such that system is consistent:

\[
\begin{cases}
x + 2y = 3 \\
2x + 4y = k
\end{cases}
\]

---

### Answer:

**Step 1:** Augmented matrix

\[
\left[\begin{array}{cc|c}
1 & 2 & 3 \\
2 & 4 & k
\end{array}\right]
\]

**Step 2:** \( R_2 \leftarrow R_2 - 2R_1 \)

\[
\left[\begin{array}{cc|c}
1 & 2 & 3 \\
0 & 0 & k-6
\end{array}\right]
\]

**Step 3:** Consistency condition  
No row \([0\ 0\ |\ \text{non-zero}]\) → We need \(k-6 = 0\)

\[
k = 6
\]

If \(k = 6\), second row becomes \([0\ 0\ |\ 0]\) → Consistent (infinitely many solutions)  
If \(k \neq 6\), second row \([0\ 0\ |\ k-6]\) with \(k-6 \neq 0\) → Inconsistent

\[
\boxed{k = 6}
\]

## Summary Table for Quick Revision

| System Type | REF Form | Consistency |
|-------------|----------|-------------|
| Unique solution | All rows have pivots, no zero row on left | ✅ Consistent |
| Infinite solutions | At least one all-zero row (including constant side) | ✅ Consistent |
| No solution | Any row \([0\ 0\ ...\ 0\ |\ \text{non-zero}]\) | ❌ Inconsistent |
