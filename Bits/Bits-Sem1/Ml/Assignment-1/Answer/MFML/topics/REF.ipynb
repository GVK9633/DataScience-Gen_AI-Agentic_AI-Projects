# REF and RREF Rules

### Row Echelon Form (REF)

A matrix is in REF if:

1. All zero rows are at the bottom.
2. The leading entry (pivot) of each nonzero row is to the right of the pivot above it.
3. Entries below each pivot are zero.

### Reduced Row Echelon Form (RREF)

In addition to REF conditions:

4. Each pivot is 1.
5. Each pivot is the only nonzero entry in its column.

# Problem 1

Find REF and RREF of

[
A=
\begin{bmatrix}
1&2&1\
2&4&3\
3&6&4
\end{bmatrix}
]

## Step 1: Make zeros below first pivot

[
R_2 \rightarrow R_2-2R_1
]

[
R_3 \rightarrow R_3-3R_1
]

[
\begin{bmatrix}
1&2&1\
0&0&1\
0&0&1
\end{bmatrix}
]

## Step 2

[
R_3 \rightarrow R_3-R_2
]

[
\begin{bmatrix}
1&2&1\
0&0&1\
0&0&0
\end{bmatrix}
]

This is the **REF**.

---

## Convert to RREF

Remove entry above second pivot:

[
R_1 \rightarrow R_1-R_2
]

[
\boxed{
\begin{bmatrix}
1&2&0\
0&0&1\
0&0&0
\end{bmatrix}
}
]

This is the **RREF**.

# Problem 2

Find RREF of

[
\begin{bmatrix}
1&2&3\
2&4&6\
1&1&1
\end{bmatrix}
]

### Step 1

[
R_2\rightarrow R_2-2R_1
]

[
R_3\rightarrow R_3-R_1
]

[
\begin{bmatrix}
1&2&3\
0&0&0\
0&-1&-2
\end{bmatrix}
]

Swap rows:

[
R_2 \leftrightarrow R_3
]

[
\begin{bmatrix}
1&2&3\
0&-1&-2\
0&0&0
\end{bmatrix}
]

### Step 2

[
R_2\rightarrow -R_2
]

[
\begin{bmatrix}
1&2&3\
0&1&2\
0&0&0
\end{bmatrix}
]

### Step 3

[
R_1\rightarrow R_1-2R_2
]

[
\boxed{
\begin{bmatrix}
1&0&-1\
0&1&2\
0&0&0
\end{bmatrix}
}
]

This is the **RREF**.

# Problem 3 (System of Equations)

Solve using RREF:

[
x+y+z=6
]

[
2x+3y+z=11
]

[
x+2y+3z=14
]

### Augmented Matrix

[
\left[
\begin{array}{ccc|c}
1&1&1&6\
2&3&1&11\
1&2&3&14
\end{array}
\right]
]

### Step 1

[
R_2\rightarrow R_2-2R_1
]

[
R_3\rightarrow R_3-R_1
]

[
\left[
\begin{array}{ccc|c}
1&1&1&6\
0&1&-1&-1\
0&1&2&8
\end{array}
\right]
]

### Step 2

[
R_3\rightarrow R_3-R_2
]

[
\left[
\begin{array}{ccc|c}
1&1&1&6\
0&1&-1&-1\
0&0&3&9
\end{array}
\right]
]

### Step 3

[
R_3\rightarrow \frac13R_3
]

[
\left[
\begin{array}{ccc|c}
1&1&1&6\
0&1&-1&-1\
0&0&1&3
\end{array}
\right]
]

### Step 4

[
R_2\rightarrow R_2+R_3
]

[
R_1\rightarrow R_1-R_3
]

[
\left[
\begin{array}{ccc|c}
1&1&0&3\
0&1&0&2\
0&0&1&3
\end{array}
\right]
]

### Step 5

[
R_1\rightarrow R_1-R_2
]

[
\boxed{
\left[
\begin{array}{ccc|c}
1&0&0&1\
0&1&0&2\
0&0&1&3
\end{array}
\right]
}
]

Therefore,

[
\boxed{x=1,\quad y=2,\quad z=3}
]

# Exam Tips for MFML

Common questions:

1. Convert matrix to REF.
2. Convert matrix to RREF.
3. Find rank using RREF.
4. Solve linear systems using RREF.
5. Determine consistency of a system.
6. Find basis and dimension using pivot columns.

### Quick Rank Rule

After RREF:

[
\text{Rank} = \text{Number of pivots}
]

Example:

[
\begin{bmatrix}
1&0&2\
0&1&3\
0&0&0
\end{bmatrix}
]

There are 2 pivots ⇒

[
\boxed{\text{Rank}=2}
]

These are exactly the types of REF/RREF problems that frequently appear in MFML mid-semester and comprehensive exams.

## Practice Problems

### Q1.

\[
A = \begin{bmatrix}
1 & 2 & 3 \\

\end{bmatrix}

Reduce the following matrix to **Reduced Row Echelon Form (RREF)**:

B = \begin{bmatrix}

\end{bmatrix}
\]

### Q3.
Find the **rank** of the following matrix by converting to REF:
\[
C = \begin{bmatrix}
2 & 1 & 3 \\
4 & 2 & 6 \\

\end{bmatrix}
\]

### Q4.
Determine if the following system is consistent by converting the augmented matrix to REF:
\[
\begin{cases}
x + 2y + z = 4 \\
2x + 5y + 3z = 9 \\
x + 3y + 2z = 5
\end{cases}
\]

### Q5.
Convert to RREF and find all solutions:
\[
\begin{cases}

\end{cases}
\]

## Answer Key with Step-by-Step Solutions

---

### Q1 Solution: REF of A

**Step 1:** \( R_2 \leftarrow R_2 - 2R_1 \), \( R_3 \leftarrow R_3 - 3R_1 \)
\[
\begin{bmatrix}
1 & 2 & 3 \\
0 & 1 & 0 \\
0 & 1 & 0
\end{bmatrix}
\]

**Step 2:** \( R_3 \leftarrow R_3 - R_2 \)
\[
\begin{bmatrix}
1 & 2 & 3 \\
0 & 1 & 0 \\
0 & 0 & 0
\end{bmatrix}
\]

✅ **REF of A**:
\[
\boxed{\begin{bmatrix}
1 & 2 & 3 \\
0 & 1 & 0 \\
0 & 0 & 0
\end{bmatrix}}
\]
Rank = 2

---

### Q2 Solution: RREF of B

**Step 1:** \( R_2 \leftarrow R_2 - 2R_1 \), \( R_3 \leftarrow R_3 - 3R_1 \)
\[
\begin{bmatrix}
1 & 2 & -1 \\
0 & 1 & -1 \\
0 & 1 & -1
\end{bmatrix}
\]

**Step 2:** \( R_3 \leftarrow R_3 - R_2 \)
\[
\begin{bmatrix}
1 & 2 & -1 \\
0 & 1 & -1 \\
0 & 0 & 0
\end{bmatrix}
\]

**Step 3:** \( R_1 \leftarrow R_1 - 2R_2 \)
\[
\begin{bmatrix}
1 & 0 & 1 \\
0 & 1 & -1 \\
0 & 0 & 0
\end{bmatrix}
\]

✅ **RREF of B**:
\[
\boxed{\begin{bmatrix}
1 & 0 & 1 \\
0 & 1 & -1 \\
0 & 0 & 0
\end{bmatrix}}
\]

---

### Q3 Solution: Rank of C

**Step 1:** Swap \( R_1 \) and \( R_3 \) for easier pivot (optional)
\[
\begin{bmatrix}
1 & 0 & 1 \\
4 & 2 & 6 \\
2 & 1 & 3
\end{bmatrix}
\]

**Step 2:** \( R_2 \leftarrow R_2 - 4R_1 \), \( R_3 \leftarrow R_3 - 2R_1 \)
\[
\begin{bmatrix}
1 & 0 & 1 \\
0 & 2 & 2 \\
0 & 1 & 1
\end{bmatrix}
\]

**Step 3:** \( R_2 \leftarrow R_2 / 2 \)
\[
\begin{bmatrix}
1 & 0 & 1 \\
0 & 1 & 1 \\
0 & 1 & 1
\end{bmatrix}
\]

**Step 4:** \( R_3 \leftarrow R_3 - R_2 \)
\[
\begin{bmatrix}
1 & 0 & 1 \\
0 & 1 & 1 \\
0 & 0 & 0
\end{bmatrix}
\]

✅ **Rank** = number of non-zero rows = **2**

\[
\boxed{2}
\]

### Q4 Solution: Consistency Check

**Augmented matrix**:
\[
\left[\begin{array}{ccc|c}
1 & 2 & 1 & 4 \\
2 & 5 & 3 & 9 \\
1 & 3 & 2 & 5
\end{array}\right]
\]

**Step 1:** \( R_2 \leftarrow R_2 - 2R_1 \), \( R_3 \leftarrow R_3 - R_1 \)
\[
\left[\begin{array}{ccc|c}
1 & 2 & 1 & 4 \\
0 & 1 & 1 & 1 \\
0 & 1 & 1 & 1
\end{array}\right]
\]

**Step 2:** \( R_3 \leftarrow R_3 - R_2 \)
\[
\left[\begin{array}{ccc|c}
1 & 2 & 1 & 4 \\
0 & 1 & 1 & 1 \\
0 & 0 & 0 & 0
\end{array}\right]
\]

✅ No row of form \( [0\ 0\ 0\ |\ \text{non-zero}] \) → **Consistent**

\[
\boxed{\text{Consistent}}
\]

---

### Q5 Solution: RREF and Solutions

**Augmented matrix**:
\[
\left[\begin{array}{ccc|c}
1 & 1 & 1 & 6 \\
2 & 3 & 4 & 20 \\
3 & 4 & 5 & 26
\end{array}\right]
\]

**Step 1:** \( R_2 \leftarrow R_2 - 2R_1 \), \( R_3 \leftarrow R_3 - 3R_1 \)
\[
\left[\begin{array}{ccc|c}
1 & 1 & 1 & 6 \\
0 & 1 & 2 & 8 \\
0 & 1 & 2 & 8
\end{array}\right]
\]

**Step 2:** \( R_3 \leftarrow R_3 - R_2 \)
\[
\left[\begin{array}{ccc|c}
1 & 1 & 1 & 6 \\
0 & 1 & 2 & 8 \\
0 & 0 & 0 & 0
\end{array}\right]
\]

**Step 3:** \( R_1 \leftarrow R_1 - R_2 \)
\[
\left[\begin{array}{ccc|c}
1 & 0 & -1 & -2 \\
0 & 1 & 2 & 8 \\
0 & 0 & 0 & 0
\end{array}\right]
\]

**Solution**:  
Let \( z = t \) (free variable)  
\( x - t = -2 \implies x = t - 2 \)  
\( y + 2t = 8 \implies y = 8 - 2t \)

✅ **General solution**:
\[
\boxed{(x, y, z) = (t - 2,\ 8 - 2t,\ t) \quad \text{for } t \in \mathbb{R}}
\]
