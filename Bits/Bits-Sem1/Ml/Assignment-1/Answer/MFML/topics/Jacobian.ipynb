## 1. What is a Jacobian Matrix?

**Definition:**  
The **Jacobian matrix** is a matrix that contains **all first-order partial derivatives** of a **vector-valued function**.

### In simple words:
- If you have a function that takes **multiple inputs** and produces **multiple outputs**, 
- The Jacobian is a matrix that tells you **how each output changes** with respect to **each input**.

## 1. What is a Jacobian Matrix?

**Definition:**  
The **Jacobian matrix** is a matrix that contains **all first-order partial derivatives** of a **vector-valued function**.

### In simple words:
- If you have a function that takes **multiple inputs** and produces **multiple outputs**, 
- The Jacobian is a matrix that tells you **how each output changes** with respect to **each input**.

---

## 2. The Formula

Suppose we have a function:
\[
\mathbf{f}(x_1, x_2, \dots, x_n) =
\begin{bmatrix}
f_1(x_1, \dots, x_n) \\
f_2(x_1, \dots, x_n) \\
\vdots \\
f_m(x_1, \dots, x_n)
\end{bmatrix}
\]

This means:
- **Inputs:** \( n \) variables (\( x_1, x_2, \dots, x_n \))
- **Outputs:** \( m \) functions (\( f_1, f_2, \dots, f_m \))

The **Jacobian matrix** \( J \) is an **\( m \times n \)** matrix:

\[
J = \begin{bmatrix}
\frac{\partial f_1}{\partial x_1} & \frac{\partial f_1}{\partial x_2} & \dots & \frac{\partial f_1}{\partial x_n} \\
\frac{\partial f_2}{\partial x_1} & \frac{\partial f_2}{\partial x_2} & \dots & \frac{\partial f_2}{\partial x_n} \\
\vdots & \vdots & \ddots & \vdots \\
\frac{\partial f_m}{\partial x_1} & \frac{\partial f_m}{\partial x_2} & \dots & \frac{\partial f_m}{\partial x_n}
\end{bmatrix}
\]

**Structure:**
- **Rows** = outputs (\( f_1, f_2, \dots \))
- **Columns** = inputs (\( x_1, x_2, \dots \))
- **Entry \( J_{ij} \)** = How output \( f_i \) changes when input \( x_j \) changes.

---

## 3. Simple Example

### Problem:
Find the Jacobian of:
\[
\mathbf{f}(x, y) = \begin{bmatrix} f_1(x, y) \\ f_2(x, y) \end{bmatrix}
= \begin{bmatrix} x^2 y \\ \sin(x) + y^3 \end{bmatrix}
\]

Here:
- **Inputs:** \( x, y \) (so \( n = 2 \))
- **Outputs:** \( f_1, f_2 \) (so \( m = 2 \))
- Jacobian will be a **\( 2 \times 2 \)** matrix.

---

### Step 1: Find all partial derivatives

#### For \( f_1 = x^2 y \):

| Derivative | Result |
|------------|--------|
| \( \frac{\partial f_1}{\partial x} \) | \( 2xy \) |
| \( \frac{\partial f_1}{\partial y} \) | \( x^2 \) |

---

#### For \( f_2 = \sin(x) + y^3 \):

| Derivative | Result |
|------------|--------|
| \( \frac{\partial f_2}{\partial x} \) | \( \cos(x) \) |
| \( \frac{\partial f_2}{\partial y} \) | \( 3y^2 \) |

---

### Step 2: Arrange in Matrix Form

\[
J = \begin{bmatrix}
\frac{\partial f_1}{\partial x} & \frac{\partial f_1}{\partial y} \\
\frac{\partial f_2}{\partial x} & \frac{\partial f_2}{\partial y}
\end{bmatrix}
= \begin{bmatrix}
2xy & x^2 \\
\cos(x) & 3y^2
\end{bmatrix}
\]

---

### Step 3: Evaluate at a Point (Optional)

If asked to evaluate at \( (x, y) = (1, 2) \):

\[
J(1, 2) = \begin{bmatrix}
2(1)(2) & (1)^2 \\
\cos(1) & 3(2)^2
\end{bmatrix}
= \begin{bmatrix}
4 & 1 \\
\cos(1) & 12
\end{bmatrix}
\]

## 4. Visual Summary

| Function | Inputs | Outputs | Jacobian Size |

| \( f(x,y) = x^2 y \) | 2 | 1 | \( 1 \times 2 \) (gradient) |
| \( \mathbf{f}(x,y) = [x^2 y,\ \sin(x)+y^3]^T \) | 2 | 2 | \( 2 \times 2 \) |
| \( \mathbf{f}(x,y,z) = [xyz,\ x+y+z]^T \) | 3 | 2 | \( 2 \times 3 \) |

## 5. Why is Jacobian Important in ML?

| Application | What Jacobian does |
|-------------|-------------------|
| **Backpropagation** | Computes gradients of loss w.r.t weights |
| **Optimization** | Used in Newton's method (Hessian approximation) |
| **Neural Networks** | Chain rule for multi-output networks |
| **Robotics** | Relates joint velocities to end-effector velocities |
| **Change of Variables** | Used in probability (transforming distributions) |

## 6. Memory Trick for Exam

> **"Jacobian = matrix of all first partial derivatives. Rows = outputs, Columns = inputs."**

| Step | Action |
|------|--------|
| 1 | Identify number of outputs = **rows** |
| 2 | Identify number of inputs = **columns** |
| 3 | Differentiate each output w.r.t each input |
| 4 | Arrange in matrix |

## 7. Quick One-Line Summary

> **"If you have \( m \) outputs and \( n \) inputs, the Jacobian is an \( m \times n \) matrix where entry \( (i,j) \) is \( \partial f_i / \partial x_j \)."**
