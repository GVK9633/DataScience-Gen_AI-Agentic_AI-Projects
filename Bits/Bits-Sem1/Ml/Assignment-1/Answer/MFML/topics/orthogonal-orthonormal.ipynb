## Orthogonal vs Orthonormal

These terms are commonly used in **Linear Algebra**, especially for vectors and matrices.

| Property                              | Orthogonal | Orthonormal |
| ------------------------------------- | ---------- | ----------- |
| Vectors are perpendicular?            | ✅ Yes      | ✅ Yes       |
| Dot product between different vectors | 0          | 0           |
| Length (magnitude) of vectors         | Any length | Exactly 1   |
| Unit vectors required?                | ❌ No       | ✅ Yes       |

---

## 1. Orthogonal Vectors

Two vectors are **orthogonal** if their dot product is zero.

### Definition

[
\mathbf{u}\cdot\mathbf{v}=0
]

### Example

[
\mathbf{u}=(1,1)
]

[
\mathbf{v}=(1,-1)
]

Dot product:

[
(1)(1)+(1)(-1)=1-1=0
]

Since the dot product is 0, the vectors are **orthogonal**.

### Lengths

[
||u||=\sqrt{1^2+1^2}=\sqrt{2}
]

[
||v||=\sqrt{1^2+(-1)^2}=\sqrt{2}
]

The lengths are not 1, so they are **not orthonormal**.

---

## 2. Orthonormal Vectors

Vectors are **orthonormal** if:

1. They are orthogonal.
2. Each vector has unit length.

### Conditions

[
u_i \cdot u_j = 0
\quad (i\neq j)
]

and

[
||u_i||=1
]

### Example

[
u_1=
\begin{bmatrix}
1\
0
\end{bmatrix}
,\quad
u_2=
\begin{bmatrix}
0\
1
\end{bmatrix}
]

Dot product:

[
u_1\cdot u_2=0
]

Lengths:

[
||u_1||=1
]

[
||u_2||=1
]

Therefore, the vectors are **orthonormal**.

---

## Geometric Interpretation

### Orthogonal

Vectors meet at a right angle (90°).

```
      v
      |
      |
      |
------|------ u
```

The vectors can have any length.

---

### Orthonormal

Vectors meet at 90° **and** each has length 1.

```
      ^
      | length = 1
      |
------|-----> length = 1
```

---

## Orthogonal Matrix

A square matrix (Q) is orthogonal if

[
Q^TQ = QQ^T = I
]

Example:

[
Q=
\begin{bmatrix}
0 & -1\
1 & 0
\end{bmatrix}
]

Check:

[
Q^TQ=I
]

So (Q) is an **orthogonal matrix**.

### Important Fact

The columns (and rows) of an orthogonal matrix are actually **orthonormal vectors**.

---

## Example

Consider

[
A=
\begin{bmatrix}
3\
4
\end{bmatrix}
]

Length:

[
||A||=\sqrt{3^2+4^2}=5
]

Normalize it:

[
u=
\frac{A}{||A||}
===============

\begin{bmatrix}
3/5\
4/5
\end{bmatrix}
]

Now

[
||u||=1
]

This process is called **normalization**.

---

## Interview Questions

### Q1. Can orthogonal vectors have different lengths?

✅ Yes.

Example:

[
(2,0),\quad (0,5)
]

They are orthogonal but not orthonormal.

---

### Q2. Is every orthonormal set orthogonal?

✅ Yes.

Orthonormal = Orthogonal + Unit Length.

---

### Q3. Is every orthogonal set orthonormal?

❌ No.

The vectors may not have length 1.

---

## Quick Memory Trick

* **Orthogonal** = **Perpendicular**
* **Orthonormal** = **Perpendicular + Normalized (unit length)**

### Example

[
(2,0),\ (0,3)
]

→ Orthogonal ✔️

→ Orthonormal ❌

---

[
(1,0),\ (0,1)
]

→ Orthogonal ✔️

→ Orthonormal ✔️

**In one line:**
**Orthonormal vectors are orthogonal vectors whose lengths are all equal to 1.**
