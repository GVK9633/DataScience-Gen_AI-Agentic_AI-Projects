# 📘 PART 1: CORE CONCEPTS – QUICK REVISION

## 1. Vector Spaces & Subspaces
| Concept | Key Point |
|---------|-----------|
| **Subspace test** | 0 vector, closed under addition & scalar multiplication |
| **Span** | All linear combinations of vectors |
| **Basis** | Linearly independent + spans the space |
| **Dimension** | Number of vectors in basis |
| **Row space** | span of rows (dimension = rank) |
| **Null space** | solutions to Ax = 0 (dimension = n − rank) |
| **Column space** | span of columns (dimension = rank) |

**Memory trick:**  
> "Rank = number of pivots. Nullity = free variables. Rank + Nullity = number of columns."

---

## 2. Eigenvalues & Eigenvectors
| Concept | Key Point |
|---------|-----------|
| **Eigenvalue equation** | \( A v = \lambda v \) |
| **Characteristic equation** | \( \det(A - \lambda I) = 0 \) |
| **Diagonalization** | \( A = PDP^{-1} \) (if n independent eigenvectors) |
| **Trace** = sum of eigenvalues = sum of diagonal entries |
| **Determinant** = product of eigenvalues |

**Memory trick:**  
> "Eigenvalues tell you how much a matrix stretches. Eigenvectors tell you the direction."

---

## 3. Matrix Decompositions
| Decomposition | Use |
|---------------|-----|
| **LU** | Solve Ax = b quickly |
| **QR** | Least squares, Gram‑Schmidt |
| **SVD** | \( A = U \Sigma V^T \) – PCA, dimensionality reduction, pseudo‑inverse |
| **Eigen decomposition** | \( A = PDP^{-1} \) (square, diagonalizable) |

**SVD memory:**  
> "U rotates, Σ scales, Vᵀ rotates back. Keep top Σ values = drop noise."

## 4. Calculus for ML
| Concept | Key Point |
|---------|-----------|
| **Gradient** | \( \nabla f = (\partial f / \partial x_1, \dots) \) – direction of steepest ascent |
| **Gradient descent** | \( x_{k+1} = x_k - \alpha \nabla f(x_k) \) |
| **Hessian** | Matrix of second derivatives – check convexity |
| **Taylor expansion** | \( f(x+h) \approx f(x) + \nabla f^T h + \frac{1}{2} h^T H h \) |
| **Jacobian** | Matrix of first derivatives for vector functions |

**Memory trick:**  
> "Gradient = first‑order info. Hessian = second‑order info. Use Hessian to check if you're at a min or max."

---

## 5. Logistic Regression
| Concept | Key Point |
|---------|-----------|
| **Log‑odds** | \( \ln \frac{p}{1-p} = \theta^T x \) |
| **Sigmoid** | \( p = \frac{1}{1+e^{-\theta^T x}} \) |
| **Decision boundary** | \( \theta^T x = 0 \) |
| **Interpretation** | Large \( \theta_i \) means feature i strongly influences log‑odds, but final prediction depends on all features + intercept |

**Memory trick:**  
> "Log‑odds = sum of all features × weights + intercept. One big weight doesn't guarantee outcome."

## Problem 1: Subspace Test (2 marks)
**Q:** Check if the set \( W = \{ (x, y, z) \in \mathbb{R}^3 \mid x + 2y - z = 0 \} \) is a subspace of \( \mathbb{R}^3 \).

**Solution:**
- **Zero vector:** (0,0,0) satisfies \( 0+0-0=0 \) → in W ✓
- **Closure under addition:**  
  Let \( u=(x_1,y_1,z_1), v=(x_2,y_2,z_2) \) in W.  
  Then \( x_1+2y_1-z_1=0 \) and \( x_2+2y_2-z_2=0 \).  
  Add: \( (x_1+x_2)+2(y_1+y_2)-(z_1+z_2)=0 \) → \( u+v \in W \) ✓
- **Closure under scalar multiplication:**  
  \( k(x+2y-z)=k(0)=0 \) → \( ku \in W \) ✓

**Answer:** Yes, it's a subspace. ✅

## Problem 2: Basis and Dimension (2 marks)
**Q:** Find a basis and dimension for the null space of:
\[
A = \begin{bmatrix} 1 & 2 & 1 \\ 2 & 4 & 2 \\ 0 & 0 & 0 \end{bmatrix}
\]

**Solution:**
- Row reduce: \( \begin{bmatrix} 1 & 2 & 1 \\ 0 & 0 & 0 \\ 0 & 0 & 0 \end{bmatrix} \)
- Pivot column: column 1 → \( x_1 = -2x_2 - x_3 \)
- Free variables: \( x_2 = s, x_3 = t \)
- General solution:
\[
\mathbf{x} = \begin{bmatrix} -2s - t \\ s \\ t \end{bmatrix} = s \begin{bmatrix} -2 \\ 1 \\ 0 \end{bmatrix} + t \begin{bmatrix} -1 \\ 0 \\ 1 \end{bmatrix}
\]
- **Basis:** \( \{ (-2,1,0)^T, (-1,0,1)^T \} \)
- **Dimension:** 2

## Problem 3: Eigenvalues & Eigenvectors (3 marks)
**Q:** Find eigenvalues and eigenvectors of:
\[
A = \begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix}
\]

**Solution:**
- Characteristic equation: \( \det(A - \lambda I) = 0 \)
\[
\det \begin{bmatrix} 2-\lambda & 1 \\ 1 & 2-\lambda \end{bmatrix} = (2-\lambda)^2 - 1 = \lambda^2 - 4\lambda + 3 = 0
\]
- \( \lambda = 1, 3 \)

**For \( \lambda = 1 \):**
\[
(A - I)v = 0 \Rightarrow \begin{bmatrix} 1 & 1 \\ 1 & 1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = 0 \Rightarrow x + y = 0 \Rightarrow v = (1, -1)
\]

**For \( \lambda = 3 \):**
\[
(A - 3I)v = 0 \Rightarrow \begin{bmatrix} -1 & 1 \\ 1 & -1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = 0 \Rightarrow x = y \Rightarrow v = (1, 1)
\]

**Answer:** \( \lambda_1 = 1, v_1 = (1,-1); \lambda_2 = 3, v_2 = (1,1) \)

## Problem 4: SVD Intuition (2 marks)
**Q:** What is the SVD of a matrix used for in ML?

**Answer:**
- **Dimensionality reduction** (PCA): keep top singular values
- **Noise filtering**: drop small singular values
- **Matrix approximation**: \( A \approx U_k \Sigma_k V_k^T \)
- **Pseudo‑inverse**: for solving non‑square systems

## Problem 5: Logistic Regression Interpretation (2 marks)
**Q:** In logistic regression, \( \theta_2 = +50 \) for CGPA. A student says high CGPA guarantees selection. Is that true?

**Answer:**
- **No.** Log‑odds = \( \theta_0 + \theta_1 \cdot \text{Age} + \theta_2 \cdot \text{CGPA} + \theta_3 \cdot \text{IQ} + \theta_4 \cdot \text{Height} \)
- Final probability depends on **all** terms.
- If intercept or other features are negative enough, high CGPA can still give low log‑odds → not selected.
