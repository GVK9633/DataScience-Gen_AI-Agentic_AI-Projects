```python
# ============================================================================
# Q2(c): QR DECOMPOSITION (A = Q × R) - No linalg
# ============================================================================

import numpy as np

print("="*80)
print("Q2(c): QR DECOMPOSITION (A = Q × R)")
print("="*80)

# STEP 1: Create matrix with m > n
m, n = 6, 4
np.random.seed(42)
A = np.random.randint(-5, 6, (m, n)).astype(float)

print("\n--- STEP 1: Matrix A ({} × {}) with m > n ---".format(m, n))
for row in A:
    print(" ", [round(x, 2) for x in row])

# STEP 2: Function to compute norm (without linalg)
def vector_norm(v):
    """Compute Euclidean norm of a vector"""
    norm_sq = 0
    for val in v:
        norm_sq += val ** 2
    return np.sqrt(norm_sq)

# STEP 3: Function for dot product (without linalg)
def dot_product(v1, v2):
    """Compute dot product of two vectors"""
    result = 0
    for i in range(len(v1)):
        result += v1[i] * v2[i]
    return result

# STEP 4: Function to multiply matrices (without linalg)
def matrix_multiply(X, Y):
    m, n_x = X.shape
    n_y, p = Y.shape
    result = np.zeros((m, p))
    for i in range(m):
        for j in range(p):
            for k in range(n_x):
                result[i, j] += X[i, k] * Y[k, j]
    return result

# STEP 5: QR Decomposition using Gram-Schmidt
def qr_decomposition(A):
    """QR decomposition using Gram-Schmidt process"""
    m, n = A.shape
    Q = np.zeros((m, n))
    R = np.zeros((n, n))
    
    print("\n--- STEP 2: Gram-Schmidt Process ---")
    
    for j in range(n):
        print(f"\n  Processing Column {j}:")
        v = A[:, j].copy()
        print(f"    Initial v = {[round(x, 4) for x in v]}")
        
        for i in range(j):
            R[i, j] = dot_product(Q[:, i], A[:, j])
            print(f"    R[{i}][{j}] = Q[:,{i}]ᵀ × A[:,{j}] = {R[i, j]:.4f}")
            for k in range(m):
                v[k] -= R[i, j] * Q[k, i]
            print(f"    After subtracting projection: v = {[round(x, 4) for x in v]}")
        
        R[j, j] = vector_norm(v)
        print(f"    R[{j}][{j}] = ||v|| = {R[j, j]:.4f}")
        
        for k in range(m):
            Q[k, j] = v[k] / R[j, j]
        print(f"    Q[:,{j}] = v / R[{j}][{j}] = {[round(x, 4) for x in Q[:, j]]}")
    
    return Q, R

Q, R = qr_decomposition(A)

print("\n--- STEP 3: Results ---")
print("\nQ (Orthogonal Matrix, {} × {}):".format(m, n))
for row in Q:
    print(" ", [round(x, 4) for x in row])

print("\nR (Upper Triangular, {} × {}):".format(n, n))
for row in R:
    print(" ", [round(x, 4) for x in row])

# STEP 6: Verification A = Q × R
print("\n--- STEP 4: Verification A = Q × R ---")
QR_product = matrix_multiply(Q, R)

diff = 0
for i in range(m):
    for j in range(n):
        diff = max(diff, abs(A[i, j] - QR_product[i, j]))

if diff < 1e-10:
    print("✓ VERIFIED: A = Q × R")
else:
    print(f"✗ Verification failed (max difference = {diff:.2e})")

# STEP 7: Orthonormality Check QᵀQ = I
print("\n--- STEP 5: Orthonormality Check QᵀQ = I ---")
QTQ = matrix_multiply(Q.T, Q)

print("QᵀQ:")
for row in QTQ:
    print(" ", [round(x, 4) for x in row])

# Check if identity
is_identity = True
for i in range(n):
    for j in range(n):
        expected = 1 if i == j else 0
        if abs(QTQ[i, j] - expected) > 1e-10:
            is_identity = False

if is_identity:
    print("\n✓ QᵀQ = I (Columns are orthonormal)")
else:
    print("\n✗ Orthonormality check failed")

print("\n" + "="*80)
print("END OF Q2(c)")
print("="*80)
```

```python
# ============================================================================
# Q2(c): QR DECOMPOSITION (A = Q × R)
# ============================================================================

import numpy as np

print("="*80)
print("Q2(c): QR DECOMPOSITION (A = Q × R)")
print("="*80)

# STEP 1: Create matrix with m > n
m, n = 6, 4
np.random.seed(42)
A = np.random.randint(-5, 6, (m, n)).astype(float)

print("\n--- STEP 1: Matrix A ({} × {}) with m > n ---".format(m, n))
for row in A:
    print(" ", [round(x, 2) for x in row])

# STEP 2: Gram-Schmidt QR Decomposition
def qr_decomposition(A):
    """QR decomposition using Gram-Schmidt process"""
    m, n = A.shape
    Q = np.zeros((m, n))
    R = np.zeros((n, n))
    
    print("\n--- STEP 2: Gram-Schmidt Process ---")
    
    for j in range(n):
        print(f"\n  Processing Column {j}:")
        v = A[:, j].copy()
        print(f"    Initial v = {[round(x, 4) for x in v]}")
        
        for i in range(j):
            R[i, j] = Q[:, i] @ A[:, j]
            print(f"    R[{i}][{j}] = Q[:,{i}]ᵀ × A[:,{j}] = {R[i, j]:.4f}")
            v = v - R[i, j] * Q[:, i]
            print(f"    After subtracting projection: v = {[round(x, 4) for x in v]}")
        
        R[j, j] = np.linalg.norm(v)
        print(f"    R[{j}][{j}] = ||v|| = {R[j, j]:.4f}")
        Q[:, j] = v / R[j, j]
        print(f"    Q[:,{j}] = v / R[{j}][{j}] = {[round(x, 4) for x in Q[:, j]]}")
    
    return Q, R

Q, R = qr_decomposition(A)

print("\n--- STEP 3: Results ---")
print("\nQ (Orthogonal Matrix, {} × {}):".format(m, n))
for row in Q:
    print(" ", [round(x, 4) for x in row])

print("\nR (Upper Triangular, {} × {}):".format(n, n))
for row in R:
    print(" ", [round(x, 4) for x in row])

# STEP 4: Verification A = Q × R
print("\n--- STEP 4: Verification A = Q × R ---")
QR_product = Q @ R
if np.allclose(A, QR_product, rtol=1e-10):
    print("✓ VERIFIED: A = Q × R")
else:
    print("✗ Verification failed")

# STEP 5: Orthonormality Check
print("\n--- STEP 5: Orthonormality Check QᵀQ = I ---")
QTQ = Q.T @ Q
print("QᵀQ:")
for row in QTQ:
    print(" ", [round(x, 4) for x in row])

if np.allclose(QTQ, np.eye(n), rtol=1e-10):
    print("\n✓ QᵀQ = I (Columns are orthonormal)")
else:
    print("\n✗ Orthonormality check failed")

print("\n" + "="*80)
print("END OF Q2(c)")
print("="*80)
```
