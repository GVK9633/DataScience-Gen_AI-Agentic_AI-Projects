With **10 years of .NET experience**, interviewers typically focus less on syntax and more on architecture, system design, performance, cloud, and real-world problem solving. Here's a prioritized preparation plan.

## 1. C# (Must Know)

Be comfortable explaining concepts with examples.

* OOP principles (SOLID, encapsulation, inheritance, polymorphism, abstraction)
* Collections (`List`, `Dictionary`, `HashSet`, `ConcurrentDictionary`)
* Generics
* Delegates, Events, Func, Action, Predicate
* LINQ (deferred execution, `SelectMany`, `GroupBy`, joins)
* Extension methods
* Exception handling and custom exceptions
* `IDisposable` and `using`
* Garbage Collection (Generations, LOH)
* Value types vs Reference types
* `ref`, `out`, `in`
* `record` vs `class`
* `readonly`, `const`, `static`
* Nullable reference types
* Pattern matching
* Reflection
* Expression Trees (basic understanding)

## High-Frequency Questions for a 10-Year .NET Interview

* Explain Dependency Injection.
* Singleton vs Scoped vs Transient.
* async/await internals.
* Task vs Thread.
* IEnumerable vs IQueryable.
* IEnumerable vs List.
* Interface vs Abstract Class.
* String vs StringBuilder.
* Heap vs Stack.
* Value Type vs Reference Type.
* Garbage Collection.
* Middleware execution order.
* JWT authentication flow.
* EF Core Change Tracker.
* Optimistic vs Pessimistic concurrency.
* SQL Indexes.
* Clustered vs Non-clustered indexes.
* Microservices communication.
* How do you secure APIs?
* How do you improve API performance?
* Explain a production issue you solved.

For someone with your experience, interviewers will often spend 30–50% of the discussion on your projects and architectural decisions. Be ready to explain *why* you chose a particular design, the trade-offs you considered, and the measurable impact (performance, scalability, maintainability, or reliability) of your decisions.
