For a **10-year .NET interview**, you don't need to memorize every design pattern. You should understand **why a pattern is used, when to use it, and be able to explain it with a simple example**.

Here are the most commonly asked patterns.

---

# 1. Singleton Pattern

### Purpose

Only **one instance** of a class should exist.

### Example

A logging service.

```csharp
public sealed class Logger
{
    private static readonly Logger _instance = new Logger();

    private Logger() { }

    public static Logger Instance => _instance;

    public void Log(string message)
    {
        Console.WriteLine(message);
    }
}
```

Usage

```csharp
Logger.Instance.Log("Application Started");
```

**Real-world examples**

* Logger
* Configuration Manager
* Cache Manager

---

# 2. Factory Pattern

### Purpose

Instead of creating objects using `new`, let another class create them.

Without Factory

```csharp
var payment = new CreditCardPayment();
```

With Factory

```csharp
public interface IPayment
{
    void Pay();
}

public class CreditCardPayment : IPayment
{
    public void Pay() => Console.WriteLine("Paid using Credit Card");
}

public class UpiPayment : IPayment
{
    public void Pay() => Console.WriteLine("Paid using UPI");
}

public class PaymentFactory
{
    public static IPayment Create(string type)
    {
        if(type=="Card")
            return new CreditCardPayment();

        return new UpiPayment();
    }
}
```

Usage

```csharp
var payment = PaymentFactory.Create("Card");
payment.Pay();
```

**Real-life**
Amazon supports

* Credit Card
* UPI
* Wallet

Factory decides which object to create.

# 3. Repository Pattern

Most commonly used in .NET.

Purpose

Separate database code from business logic.

Without Repository

```csharp
public class ProductService
{
    public void GetProducts()
    {
        var db = new AppDbContext();
    }
}
```

With Repository

```csharp
public interface IProductRepository
{
    List<Product> GetAll();
}

public class ProductRepository : IProductRepository
{
    public List<Product> GetAll()
    {
        return new List<Product>();
    }
}
```

Service

```csharp
public class ProductService
{
    private readonly IProductRepository repo;

    public ProductService(IProductRepository repository)
    {
        repo = repository;
    }

    public List<Product> GetProducts()
    {
        return repo.GetAll();
    }
}
```

Benefits

* Easy testing
* Loose coupling
* Easier maintenance

# 4. Dependency Injection (DI)

Purpose

Instead of creating objects inside a class, inject them.

Bad

```csharp
public class EmailService
{
    private Logger logger = new Logger();
}
```

Good

```csharp
public class EmailService
{
    private readonly ILogger logger;

    public EmailService(ILogger logger)
    {
        this.logger = logger;
    }
}
```

Register

```csharp
builder.Services.AddScoped<ILogger, Logger>();
```

# 5. Strategy Pattern

Purpose

Choose one algorithm at runtime.

Example

Different discounts.

```csharp
public interface IDiscount
{
    decimal Calculate(decimal amount);
}
```

Silver Customer

```csharp
public class SilverDiscount : IDiscount
{
    public decimal Calculate(decimal amount)
    {
        return amount * 0.9m;
    }
}
```

Gold Customer

```csharp
public class GoldDiscount : IDiscount
{
    public decimal Calculate(decimal amount)
    {
        return amount * 0.8m;
    }
}
```

Usage

```csharp
IDiscount discount = new GoldDiscount();

Console.WriteLine(discount.Calculate(1000));
```

Example

```csharp
Pizza pizza = new PizzaBuilder()
    .WithCheese()
    .WithPaneer()
    .Build();
```

Instead of

```csharp
new Pizza(true,false,true,true,false,false)
```

# 7. Observer Pattern

Purpose

One object notifies multiple objects.

Example

YouTube

```
YouTuber uploads video

↓

Subscribers receive notification
```

.NET example

```csharp
public event Action Notify;

Notify?.Invoke();
```

Real-life

* Email notifications
* Stock alerts
* Push notifications

# 8. Decorator Pattern

Purpose

Add functionality without changing existing class.

Example

Coffee

Basic Coffee

↓

Add Milk

↓

Add Sugar

↓

Add Chocolate

Instead of creating

```
CoffeeWithMilk

CoffeeWithSugar

CoffeeWithMilkSugar
```

Decorator wraps the original object.

# 9. Adapter Pattern

Purpose

Convert one interface into another.

Example

Indian Charger

↓

US Socket

Need Adapter.

Code

```csharp
OldPaymentGateway

↓

Adapter

↓

New Interface
```

Useful when integrating third-party APIs.

# 10. CQRS

Separate

* Read Operations
* Write Operations

Instead of

```
ProductService

GetProducts()

SaveProduct()

DeleteProduct()
```

Use

```
GetProductQuery

CreateProductCommand

DeleteProductCommand
```

Benefits

* Better scalability
* Cleaner architecture

---

# 11. Mediator Pattern

Instead of classes talking to each other directly,

Everything goes through a mediator.

Without

```
Controller
   ↓
Service1
   ↓
Service2
   ↓
Service3
```

With MediatR

```
Controller

↓

Mediator

↓

Handler
```

Very common in Clean Architecture.

---

# 12. Unit of Work

Suppose

```
Insert Customer

Insert Order

Insert Payment
```

All should succeed or fail together.

```csharp
using(var transaction = context.Database.BeginTransaction())
{
    // Save Customer

    // Save Order

    // Save Payment

    transaction.Commit();
}
```

# Interview Tip (Most Important)

Don't just define the pattern. Explain **when** and **why** you would use it. For example:

**Q: Why Repository Pattern?**

A: "Repository Pattern abstracts data access from business logic. My services depend on interfaces instead of EF Core directly, which improves testability, reduces coupling, and makes it easier to switch data sources if needed."
